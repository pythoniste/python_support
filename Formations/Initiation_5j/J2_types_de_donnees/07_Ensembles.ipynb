## 2. Unicité et non-ordre

````python
{1, 2, 3, 2, 1}
````

Les doublons disparaissent automatiquement.

````python
{'c', 'a', 'b'}
````

L'affichage n'a **pas** d'ordre garanti : ne vous fiez pas à l'ordre visible.

### Les éléments doivent être hashables

Comme les clés de dict, les éléments de set doivent être immuables.

````python
try:
    s = {[1, 2]}
except TypeError as err:
    print('erreur :', err)
````

````python
s = {(1, 2), (3, 4)}
print(s)
````

## 3. Ajouter, retirer, vider

````python
s = {1, 2, 3}
s.add(4)
print(s)
````

`add` sur une valeur déjà présente ne fait rien (pas d'erreur).

````python
s.add(2)
print(s)
````

### `remove` vs `discard`

````python
s = {1, 2, 3}
s.remove(2)
print(s)
````

````python
try:
    s.remove(99)
except KeyError as err:
    print('remove lève KeyError :', err)
````

`discard` ne lève pas d'erreur si l'élément est absent.

````python
s.discard(99)
print(s)
````

### `pop` retire un élément arbitraire

````python
s = {1, 2, 3}
elem = s.pop()
print('retiré :', elem, ' restant :', s)
````

### `clear` vide l'ensemble

````python
s = {1, 2, 3}
s.clear()
print(s)
````

## 4. Appartenance et taille

Le test `x in s` est **O(1)** en moyenne. C'est la raison principale d'utiliser un set.

````python
s = {1, 2, 3, 4, 5}
3 in s
````

````python
99 in s
````

````python
len(s)
````

## 5. Opérations ensemblistes

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
````

### Union `|`

````python
a | b
````

### Intersection `&`

````python
a & b
````

### Différence `-`

````python
a - b
````

````python
b - a
````

### Différence symétrique `^` (éléments dans l'un ou l'autre, mais pas les deux)

````python
a ^ b
````

### Versions avec méthodes

Chaque opérateur a un équivalent méthode qui accepte **n'importe quel itérable**, pas seulement un autre set.

````python
a.union([3, 4, 5, 6])
````

````python
a.intersection([3, 4, 5, 6])
````

````python
a.difference([3, 4, 5, 6])
````

### Mises à jour en place

````python
a = {1, 2, 3}
a |= {3, 4}
print(a)
````

````python
a = {1, 2, 3}
a &= {2, 3, 4}
print(a)
````

````python
{1, 2} < {1, 2, 3}    # strictement inclus
````

## 8. `frozenset` — version immuable

Un `frozenset` a les mêmes opérations qu'un set, mais on ne peut ni ajouter ni retirer. Il est **hashable**, donc utilisable comme clé de dict ou élément d'un autre set.

````python
fs = frozenset([1, 2, 3])
print(fs)
````

````python
try:
    fs.add(4)
except AttributeError as err:
    print('erreur :', err)
````

````python
d = {frozenset([1, 2]): 'paire 1-2', frozenset([3, 4]): 'paire 3-4'}
print(d)
````

## 9. Cas d'usage : dédoublonner une liste

L'usage le plus fréquent d'un set : éliminer les doublons d'une séquence.

````python
valeurs = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]
unique = list(set(valeurs))
print(unique)
````

Attention : la conversion en set **perd l'ordre**. Si l'ordre compte, voir l'astuce ci-dessous.

````python
# Dédoublonnage ordonné (Python 3.7+, via dict)
valeurs = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]
list(dict.fromkeys(valeurs))
````

## 10. Synthèse

| Opération | Syntaxe |
|---|---|
| Création | `{1, 2, 3}` / `set()` (pas `{}`) |
| Ajout | `s.add(x)` |
| Retrait strict | `s.remove(x)` (KeyError si absent) |
| Retrait tolérant | `s.discard(x)` |
| Appartenance | `x in s` (O(1)) |
| Union | `a \| b` |
| Intersection | `a & b` |
| Différence | `a - b` |
| Différence symétrique | `a ^ b` |
| Sous-ensemble | `a <= b` |

### Règles à retenir

1. `{}` = dict vide, **pas** set vide. Utiliser `set()`.
2. Les éléments doivent être hashables.
3. Un set n'a pas d'ordre garanti — ne triez pas implicitement.
4. Pour les tests d'appartenance répétés, un set est beaucoup plus rapide qu'une liste.
