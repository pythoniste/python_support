## 1. Rappel : `True`, `False`, `bool`

Python n'a que **deux valeurs booléennes** : `True` et `False` (toujours avec une majuscule).


````python
print(True)
````

````python
print(False)
````

````python
print(type(True))
````

**Fait amusant** : `bool` est une **sous-classe** de `int`. `True` équivaut à `1`, `False` équivaut à `0` dans un contexte numérique.

````python
print(True + True)
````

````python
print(False * 10)
````

On exploite parfois cette propriété pour compter : `True + True + False` = 2 (on verra des idiomes avec les listes).

## 2. Les comparaisons produisent des booléens

Toute comparaison (`==`, `!=`, `<`, `<=`, `>`, `>=`, `is`, `in`) renvoie un booléen.


````python
print(3 < 5)
````

````python
print(3 == 3.0)
````

````python
x = 10
print(x > 5 and x < 20)
````

On peut **stocker** une comparaison dans une variable — c'est souvent plus lisible qu'un `if` imbriqué.

````python
age = 20
majeur = age >= 18
if majeur:
    print("Accès autorisé")
````

## 3. Truthiness : valeurs « vraies » et « fausses »

En Python, **toute valeur** peut être évaluée en contexte booléen (dans un `if`, un `while`, un `and`, etc.). Les valeurs considérées **fausses** sont :

- `False`
- `None`
- `0`, `0.0`, `0j`
- chaîne vide `""`
- conteneurs vides `[]`, `()`, `{}`, `set()` *(à connaître pour plus tard)*

**Tout le reste est vrai.**


````python
x = 0
if x:
    print("x est vrai")
else:
    print("x est faux")
````

````python
x = 42
if x:
    print("x est vrai")
````

````python
nom = ""
if nom:
    print("nom défini :", nom)
else:
    print("nom vide")
````

````python
nom = "Alice"
if nom:
    print("nom défini :", nom)
````

### Idiome : tester la présence d'une valeur

Au lieu d'écrire `if nom != "":`, on écrit simplement `if nom:`. C'est plus concis et plus pythonique.


## 4. `bool(x)` — tester la véracité explicitement

`bool(x)` renvoie `True` ou `False` selon la véracité de `x`. On s'en sert surtout pour **débugger**, rarement en production.


````python
print(bool(0))
````

````python
print(bool(1))
````

````python
print(bool(-5))
````

````python
print(bool(0.0))
````

````python
print(bool(""))
````

````python
print(bool("non"))
````

````python
print(bool(None))
````

## 5. Opérateurs de court-circuit : `and` et `or` retournent des valeurs

**Détail important** : en Python, `and` et `or` ne renvoient pas forcément `True`/`False`. Ils renvoient **l'opérande qui a permis de décider**.


````python
print(3 and 5)
````

````python
print(0 and 5)
````

````python
print(3 or 5)
````

````python
print(0 or 5)
````

````python
print("" or "défaut")
````

### Règles précises

| Expression | Évaluation | Résultat |
|---|---|---|
| `a and b` | si `a` faux, s'arrête à `a` | `a` |
| `a and b` | sinon | `b` |
| `a or b` | si `a` vrai, s'arrête à `a` | `a` |
| `a or b` | sinon | `b` |

### Pattern : valeur par défaut

L'idiome `valeur = entree or "par défaut"` utilise `or` pour fournir une valeur si `entree` est vide.


````python
entree = ""
nom = entree or "Anonyme"
print("Bonjour,", nom)
````

````python
entree = "Alice"
nom = entree or "Anonyme"
print("Bonjour,", nom)
````

**Attention** : ce pattern fait passer `0` ou `0.0` pour une valeur « vide », ce qui peut être faux selon le contexte. On verra mieux dans le notebook sur les fonctions.

## 6. `not` — l'inverse logique

`not x` est `True` si `x` est faux, `False` si `x` est vrai.


````python
print(not True)
````

````python
print(not False)
````

````python
print(not 0)
````

````python
print(not 42)
````

````python
print(not "")
````

Idiome courant : `if not x:` signifie « si x est faux / vide / zéro / None ».

## 7. `any` et `all`

Ces deux primitives acceptent une **séquence** de valeurs et retournent un booléen :

- `all(séquence)` : `True` si **toutes** les valeurs sont vraies. Équivalent de plusieurs `and` enchaînés.
- `any(séquence)` : `True` si **au moins une** valeur est vraie. Équivalent de plusieurs `or` enchaînés.

**Cas de base** :

- `all([])` → `True` *(vacuité)*
- `any([])` → `False`

Pour les utiliser sans connaître les listes, on peut passer une **expression génératrice** basée sur `range(...)` :


````python
print(all(i > 0 for i in range(1, 10)))
````

````python
print(all(i > 5 for i in range(1, 10)))
````

````python
print(any(i > 5 for i in range(1, 10)))
````

````python
print(any(i > 100 for i in range(1, 10)))
````

L'expression `(i > 5 for i in range(1, 10))` est un **générateur** : elle produit des booléens un par un. `any`/`all` s'arrêtent dès qu'ils ont leur réponse (court-circuit). On approfondira les générateurs dans le notebook sur les compréhensions.

### Cas idiomatique : vérifier une propriété sur tous les entiers d'un intervalle

````python
# Tous les entiers de 1 à 100 sont-ils positifs ?
print(all(i > 0 for i in range(1, 101)))
````

````python
# Existe-t-il un multiple de 7 entre 50 et 60 ?
print(any(i % 7 == 0 for i in range(50, 61)))
````

## 8. Synthèse

### Valeurs fausses en Python

- `False`
- `None`
- `0`, `0.0`
- `""`
- conteneurs vides `[]`, `()`, `{}`, `set()`

**Tout le reste est vrai.**

### Règles à retenir

1. **`if x:`** est l'idiome pour tester la présence d'une valeur — plus concis que `if x != 0:`.
2. **`and`/`or` ne renvoient pas nécessairement `True`/`False`** : ils renvoient l'un de leurs opérandes.
3. **Pattern de valeur par défaut** : `nom = entree or "défaut"` (avec les limites).
4. **`all` et `any`** sur une expression génératrice — pratique pour tester une propriété sur un intervalle.
5. **`not`** fonctionne sur n'importe quelle valeur, pas seulement sur un booléen strict.
6. **Comparer à `True`/`False` avec `==`** est un anti-pattern : écrire `if x:` plutôt que `if x == True:`.
