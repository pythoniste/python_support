## 1. Qu'est-ce que Ruff ?

**Ruff** est un **linter + formatter** pour Python, écrit en Rust. Il remplace à lui seul :

- `flake8` et ses plugins (linting) ;
- `isort` (tri des imports) ;
- `pyupgrade` (modernisation automatique) ;
- `black` (formatage).

Principal avantage : **la vitesse** (10 à 100× plus rapide que ses prédécesseurs).

## 2. Installation

Avec `uv` (recommandé) :

```bash

```

Ou avec `pip` :

```bash

```

Vérifier la version :

```bash
ruff --version
```

---


## 3. `ruff check` — linting

```bash
ruff check .
ruff check src/
ruff check mon_fichier.py
```

Chaque règle a un **code** : `E501` (ligne trop longue), `F401` (import inutilisé), `I001` (imports à trier)…

Exemple de sortie :

```
src/demo.py:3:1: F401 [*] `os` imported but unused
src/demo.py:7:80: E501 Line too long (92 > 88)
Found 2 errors.
[*] 1 fixable with the `--fix` option.
```

---


## 4. `ruff format` — formatage

```bash
ruff format .
ruff format --check .
```

Le format est **compatible `black`** : mêmes choix, mêmes 88 caractères par ligne par défaut. `--check` renvoie un exit code non nul sans modifier les fichiers — utile en CI.

---


## 5. Auto-fix

Beaucoup de règles sont réparables automatiquement :

```bash
ruff check --fix .
```

Pour les corrections « possibles mais discutables », ajouter `--unsafe-fixes`.

**Toujours commit avant de lancer un auto-fix**, pour pouvoir annuler proprement.

---


## 6. Configuration dans `pyproject.toml`

Ruff se configure dans `pyproject.toml` sous la section `[tool.ruff]` :

````python
# pyproject.toml
config = '''
[tool.ruff]
line-length = 88
target-version = "py313"

[tool.ruff.lint]
select = [
    "E",   # pycodestyle errors
    "F",   # pyflakes
    "I",   # isort
    "UP",  # pyupgrade
    "B",   # flake8-bugbear
]
ignore = ["E501"]

[tool.ruff.lint.per-file-ignores]
"tests/*" = ["D100", "D103"]
'''
print(config)
````

### Quelques familles de règles utiles

| Code | Famille |
|---|---|
| `E`, `W` | pycodestyle |
| `F` | pyflakes |
| `I` | isort |
| `N` | pep8-naming |
| `D` | pydocstyle |
| `UP` | pyupgrade |
| `B` | bugbear |
| `RUF` | règles spécifiques Ruff |

## 8. Synthèse

| Commande | Effet |
|---|---|
| `ruff check .` | Lint |
| `ruff check --fix .` | Lint + réparer |
| `ruff format .` | Formater |
| `ruff format --check .` | Vérifier sans modifier |

### Règles

1. Ruff tourne à chaque commit (pre-commit) et en CI.
2. La config vit dans `pyproject.toml`, partagée par toute l'équipe.
3. Utiliser `--check` en CI, pas `--fix`.

## Mini-exemples supplémentaires

### Ignorer une ligne ponctuellement

````python
# Pour ignorer une règle sur une ligne précise :
valeur = 42  # noqa: F841
# Sans le 'noqa', Ruff signalerait que 'valeur' n'est pas utilisée.
````

Limiter l'usage de `# noqa` : c'est une **dette technique**. À réserver aux cas où la règle n'a vraiment pas de sens.

### Ignorer un dossier

```toml
[tool.ruff]
exclude = ["migrations", "legacy_code", ".venv"]
```

### Activer toutes les règles... puis trier

```toml
[tool.ruff.lint]
select = ["ALL"]
ignore = [
    "D",     # docstrings — trop strict au début
    "ANN",   # annotations — on le fait progressivement
]
```

### Auto-fix sûr vs auto-fix risqué

```bash
ruff check --fix .          # fixes sûrs
ruff check --fix --unsafe-fixes .  # inclut les réécritures risquées
```

### Statistiques de projet

```bash
ruff check --statistics .
```

Affiche combien d'erreurs par règle — utile pour décider quelle famille activer en priorité.

### Comparaison avec d'autres outils

| Outil historique | Équivalent Ruff |
|---|---|
| `flake8` + plugins | `select = ["E", "F", "W", ...]` |
| `isort` | `select = ["I"]` |
| `pyupgrade` | `select = ["UP"]` |
| `black` | `ruff format` |
| `bandit` (partiel) | `select = ["S"]` |

### Quand Ruff ne suffit pas

Ruff est un **linter syntaxique**. Il ne remplace **pas** :
- `mypy` pour la vérification de types ;
- `bandit` pour une analyse sécurité poussée ;
- les tests unitaires.

Stack typique : **ruff + mypy + pytest**.

### Workflow recommandé

1. Écrire le code.
2. `ruff format .` (formate).
3. `ruff check --fix .` (répare).
4. Reste manuel → corriger.
5. `mypy src/` (types).
6. `pytest` (tests).

## Cheat sheet — Ruff en 10 commandes

```bash
ruff --version                       # version
ruff check .                          # lint
ruff check --fix .                    # lint + auto-fix sûr
ruff check --fix --unsafe-fixes .    # auto-fix plus risqué
ruff check --statistics .            # stats par règle
ruff check --select E501 .            # une règle précise
ruff check --ignore E501 .            # ignorer une règle
ruff format .                         # formater
ruff format --check .                 # vérifier sans modifier
ruff check --watch .                  # mode watch (re-linte)
```

### Règles les plus utiles au quotidien

| Code | Règle | Effet |
|---|---|---|
| `E501` | Line too long | Ligne > N caractères |
| `F401` | Unused import | Import inutilisé |
| `F841` | Unused variable | Variable affectée mais jamais lue |
| `I001` | Unsorted imports | Imports mal triés |
| `UP006` | Non-pep585 annotation | `List[int]` → `list[int]` |
| `UP035` | Import deprecated | `typing.List` → `list` |
| `B008` | Function call in default | Piège du mutable default |
| `B006` | Mutable default | Liste en default argument |
| `E722` | Bare except | `except:` sans type |
| `N802` | Function name | `snake_case` non respecté |

### Exemple de config recommandée

````python
config = '''
[tool.ruff]
line-length = 88
target-version = "py313"
src = ["src", "tests"]

[tool.ruff.lint]
select = [
    "E",   # pycodestyle
    "F",   # pyflakes
    "W",   # warnings
    "I",   # isort
    "N",   # naming
    "UP",  # pyupgrade
    "B",   # bugbear
    "C4",  # comprehensions
    "SIM", # simplify
    "RUF", # ruff-spécifique
]
ignore = []

[tool.ruff.lint.per-file-ignores]
"tests/*" = ["D"]
'''
print(config)
````
