## 1. Pourquoi tester ?

- **Sécurité au refactoring** : changer du code sans casser.
- **Spécification exécutable** : le test décrit ce que la fonction doit faire.
- **Détection précoce des régressions**.
- **Documentation vivante** : un test est un exemple d'utilisation.

Pytest est le standard de facto en Python.

## 2. Premier test

````python
# src/calc/operations.py
def ajouter(a: int, b: int) -> int:
    """Renvoie la somme de a et b."""
    return a + b

# tests/test_operations.py
# from calc.operations import ajouter
def test_ajouter_basique() -> None:
    assert ajouter(2, 3) == 5

# Ici on peut directement l'exécuter :
test_ajouter_basique()
print('test passé')
````

## 3. Convention de nommage

| Élément | Convention |
|---|---|
| Fichier | `test_*.py` ou `*_test.py` |
| Fonction | `test_*` |
| Classe | `Test*` (sans `__init__`) |

Pytest découvre **automatiquement** ces fichiers/fonctions.

## 4. `assert` enrichi

Pytest réécrit les `assert` pour afficher les valeurs en cas d'échec — pas besoin de messages manuels.

````python
def test_multiplication() -> None:
    assert 3 * 4 == 12

test_multiplication()
print('ok')
````

````python
# En cas d'échec, pytest affiche les deux valeurs :
# >   assert 3 * 4 == 13
# E   assert 12 == 13
````

## 5. Tester une exception : `pytest.raises`

````python
import pytest  # disponible si le package est installé

def diviser(a: float, b: float) -> float:
    if b == 0:
        raise ValueError('division par zéro')
    return a / b

def test_diviser_par_zero() -> None:
    with pytest.raises(ValueError, match='zéro'):
        diviser(10, 0)

# try:
#     test_diviser_par_zero()
#     print('ok')
# except ModuleNotFoundError:
#     print('pytest non installé — à lancer dans un vrai projet')
````

## 7. Fixtures — aperçu

Une **fixture** fournit un objet préparé à chaque test qui le demande. Exemple : un dossier temporaire.

````python
# import pytest
# 
# @pytest.fixture
# def donnees() -> list[int]:
#     return [1, 2, 3]
# 
# def test_somme(donnees: list[int]) -> None:
#     assert sum(donnees) == 6
print('fixtures — voir le chapitre intermédiaire')
````

Pytest fournit aussi des fixtures prêtes à l'emploi : `tmp_path` (dossier temporaire), `capsys` (capture stdout), `monkeypatch` (patch à la volée).

## 8. Lancer les tests

```bash
# tout
pytest

pytest tests/
pytest tests/test_operations.py

# un test précis
pytest tests/test_operations.py::test_ajouter_basique

# verbeux et arrêt au premier échec
pytest -v -x

# avec couverture (si pytest-cov installé)
pytest --cov=src/mon_projet
```

## 9. Synthèse

| Besoin | Syntaxe |
|---|---|
| Test simple | `def test_...(): assert ...` |
| Exception attendue | `with pytest.raises(...):` |
| Plusieurs cas | `@pytest.mark.parametrize(...)` |
| Objet préparé | `@pytest.fixture` |

### Règles

1. `tests/` à côté de `src/`, pas dedans.
2. Un test = une vérification.
3. Nom de test descriptif : `test_ajouter_nombres_negatifs_renvoie_somme_negative`.
4. Pas d'état partagé entre tests.

### `pytest -k` — sélection par motif

```bash

```

Exécute les tests dont le nom contient `ajouter` mais pas `zero`.

### `pytest -x --pdb` — débugger au premier échec

Combine l'arrêt au premier échec (`-x`) et l'entrée automatique dans `pdb` (`--pdb`). Très pratique quand on cherche un bug.

### Structure typique

```
mon_projet/
├── pyproject.toml
├── src/
│   └── mon_projet/
│       └── core.py
└── tests/
    └── test_core.py
```

La section `[tool.pytest.ini_options]` du `pyproject.toml` permet de déclarer `testpaths = ["tests"]`.

### Ne pas tester le code des bibliothèques

On ne teste pas le comportement de `json.dumps` ou de `sum`. On teste **notre** code qui les utilise.

### Couverture de code

```bash
pip install pytest-cov
pytest --cov=src/mon_projet --cov-report=term-missing
```

La couverture indique quelles lignes ne sont pas exécutées par les tests. **Viser 80 % est un bon objectif**, 100 % est souvent du luxe.

## Cheat sheet — pytest au quotidien

```bash
# Exécuter
pytest                          # tous les tests
pytest tests/test_core.py       # un fichier
pytest tests/test_core.py::test_ajouter   # un test
pytest -k ajouter               # tests contenant 'ajouter'

# Verbeux
pytest -v                       # nom de chaque test
pytest -vv                      # détails complets

# Fail fast / debug
pytest -x                       # stop au premier échec
pytest -x --pdb                 # + entre dans pdb
pytest --lf                     # rejouer uniquement les échecs de la dernière fois
pytest --ff                     # échecs d'abord, puis le reste

# Couverture (avec pytest-cov)
pytest --cov=src/mon_projet
pytest --cov=src/mon_projet --cov-report=html
```

### Assertions courantes

````python
# assert simple
assert 2 + 2 == 4

# assert avec message d'aide (mais pytest affiche déjà les valeurs)
assert [1, 2, 3] == [1, 2, 3]

# égalité approximative pour float
import math
assert math.isclose(0.1 + 0.2, 0.3, abs_tol=1e-9)

# assertion sur une exception
def diviser(a: float, b: float) -> float:
    if b == 0:
        raise ValueError('/0')
    return a / b

try:
    diviser(1, 0)
except ValueError:
    print('ok')
````
