## 1. Qu'est-ce que `pyproject.toml` ?

C'est le **fichier de configuration central** d'un projet Python moderne. Défini par **PEP 518** et **PEP 621**, il remplace `setup.py`, `setup.cfg`, `requirements.txt`, `Pipfile`, etc.

Il contient trois grandes familles de sections :

- `[build-system]` — comment construire le package ;
- `[project]` — métadonnées du projet (nom, version, dépendances) ;
- `[tool.*]` — config des outils (ruff, pytest, mypy…).

---


## 2. Section `[project]` — PEP 621

````python
exemple = '''
[project]
name = "mon-projet"
version = "0.1.0"
description = "Un petit utilitaire de démonstration."
readme = "README.md"
requires-python = ">=3.13"
license = { text = "MIT" }
authors = [{ name = "Alice", email = "alice@example.com" }]
keywords = ["demo", "formation"]
classifiers = [
    "Programming Language :: Python :: 3",
    "License :: OSI Approved :: MIT License",
]
'''
print(exemple)
````

### Champs obligatoires

Seuls `name` et `version` sont strictement obligatoires — tout le reste est recommandé.

---


## 3. Dépendances

````python
exemple = '''
[project]
dependencies = [
    "httpx>=0.27",
    "rich>=13",
    "pydantic>=2.0,<3.0",
]
'''
print(exemple)
````

### Syntaxe des versions

| Forme | Sens |
|---|---|
| `httpx` | n'importe quelle version |
| `httpx>=0.27` | au moins 0.27 |
| `httpx==0.27.0` | exactement 0.27.0 |
| `httpx>=0.27,<1.0` | 0.27 à 1.0 exclu |
| `httpx~=0.27.0` | compatible release (`>=0.27.0,<0.28`) |

---


## 4. Dépendances optionnelles et de dev

````python
exemple = '''
[project.optional-dependencies]
dev = [
    "ruff>=0.6",
    "mypy>=1.10",
    "pytest>=8",
]
docs = ["sphinx>=7"]
'''
print(exemple)
````

### Avec `uv` : section dédiée

```toml
[dependency-groups]
dev = ["ruff", "pytest", "mypy"]
```

C'est la convention recommandée par `uv` et `hatch`.

---


## 5. Entry points (scripts CLI)

````python
exemple = '''
[project.scripts]
mon-outil = "mon_projet.cli:principal"
'''
print(exemple)
````

Une fois le package installé (`pip install .` ou `uv sync`), la commande `mon-outil` apparaît dans le PATH et invoque la fonction `principal` du module `mon_projet.cli`.

C'est la manière standard de distribuer une commande CLI en Python.

## 6. Build backend

La section `[build-system]` indique **qui** construit le package.

````python
exemple = '''
[build-system]
requires = ["hatchling"]
build-backend = "hatchling.build"
'''
print(exemple)
````

| Backend | Quand |
|---|---|
| **Hatchling** | Moderne, léger, recommandé |
| Setuptools | Historique, toujours valide |
| Flit | Simple, pour petits projets |
| Poetry-core | Si on utilise Poetry |

Pour un projet neuf, **Hatchling**.

---


## 7. Exemple complet

````python
exemple = '''
[build-system]
requires = ["hatchling"]
build-backend = "hatchling.build"

[project]
name = "mon-projet"
version = "0.1.0"
description = "Un petit utilitaire de démonstration."
readme = "README.md"
requires-python = ">=3.13"
license = { text = "MIT" }
authors = [{ name = "Alice", email = "alice@example.com" }]
dependencies = [
    "httpx>=0.27",
    "rich>=13",
]

[project.optional-dependencies]
dev = ["ruff>=0.6", "mypy>=1.10", "pytest>=8"]

[project.scripts]
mon-outil = "mon_projet.cli:principal"

[tool.ruff]
line-length = 88
target-version = "py313"

[tool.ruff.lint]
select = ["E", "F", "I", "UP", "B"]

[tool.pytest.ini_options]
testpaths = ["tests"]
'''
print(exemple)
````

## 8. Synthèse

| Section | Rôle |
|---|---|
| `[build-system]` | Comment builder |
| `[project]` | Métadonnées PEP 621 |
| `[project.dependencies]` | Dépendances runtime |
| `[project.optional-dependencies]` | Groupes optionnels (dev, docs) |
| `[project.scripts]` | Entry points CLI |
| `[tool.ruff]`, `[tool.pytest.ini_options]` | Config des outils |

### Règles

1. Un seul `pyproject.toml` à la racine du projet.
2. Ne plus utiliser `setup.py` pour un projet neuf.
3. Pin les dépendances **par intervalle**, pas en exact (`>=x,<y`).

## Mini-exemples supplémentaires

### Lire un `pyproject.toml` avec `tomllib`

````python
import tomllib
from pathlib import Path

pp = '[project]\nname = "demo"\nversion = "0.1.0"\n'
Path('/tmp/demo_pyproject.toml').write_text(pp, encoding='utf-8')

with open('/tmp/demo_pyproject.toml', 'rb') as f:
    data = tomllib.load(f)
print(data)
````

⚠️ `tomllib` lit **uniquement**. Pour écrire, utiliser `tomli_w` (tiers) ou construire la chaîne à la main.

## Cheat sheet — `pyproject.toml` en un coup d'œil

````python
template = '''
[build-system]
requires = ["hatchling"]
build-backend = "hatchling.build"

[project]
name = "mon-outil"
version = "0.1.0"
description = "..."
readme = "README.md"
requires-python = ">=3.13"
license = { text = "MIT" }
authors = [{ name = "Vous", email = "vous@example.com" }]
dependencies = ["httpx>=0.27"]

[project.optional-dependencies]
dev = ["ruff>=0.6", "pytest>=8", "mypy>=1.10"]

[project.scripts]
mon-outil = "mon_outil.cli:main"

[project.urls]
Homepage = "https://github.com/vous/mon-outil"
Issues = "https://github.com/vous/mon-outil/issues"

[tool.ruff]
line-length = 88

[tool.pytest.ini_options]
testpaths = ["tests"]
'''
print(template)
````

### Versions pin — les 4 formes

| Spécifieur | Sens |
|---|---|
| `httpx` | Toute version |
| `httpx==0.27.0` | Exactement 0.27.0 |
| `httpx>=0.27,<1.0` | Intervalle |
| `httpx~=0.27.0` | Compatible release (≥0.27.0, <0.28) |

### Build backends populaires

| Backend | Recommandation |
|---|---|
| Hatchling | Moderne, léger, **recommandé** |
| Setuptools | Classique, toujours valide |
| Flit | Simple, packages purs |
| Poetry-core | Si vous utilisez Poetry |
| Maturin | Pour Rust + Python |
