## 1. PEP 8 : pourquoi ?

**PEP 8** est le guide de style officiel pour le code Python. C'est la **lingua franca** de la communauté : en le suivant, votre code ressemble à celui de tout le monde, ce qui facilite sa lecture par n'importe qui.

Extrait fameux de PEP 8 : *« A Foolish Consistency is the Hobgoblin of Little Minds »* — ne transformez pas la cohérence en obsession : lisibilité d'abord.

## 2. Indentation et longueur de ligne

- **Indentation : 4 espaces**. Jamais de tabulations.
- **Longueur de ligne : 79 caractères** (ou 99 pour les bases de code modernes, ou 88 comme `black`/`ruff`).

````python
def resume(valeurs: list[int]) -> dict[str, float]:
    """Renvoie un résumé statistique."""
    return {
        'min': float(min(valeurs)),
        'max': float(max(valeurs)),
        'moyenne': sum(valeurs) / len(valeurs),
    }
````

## 3. Lignes blanches

- **2 lignes blanches** entre définitions de fonctions/classes au niveau module ;
- **1 ligne blanche** entre méthodes dans une classe ;
- **1 ligne blanche** pour séparer les blocs logiques dans une fonction.

## 4. Nommage

| Type | Convention | Exemple |
|---|---|---|
| variable / fonction | `snake_case` | `compter_mots`, `prix_ht` |
| constante | `UPPER_SNAKE_CASE` | `TAUX_TVA`, `PI` |
| classe | `PascalCase` | `Compte`, `HttpClient` |
| privé | `_prefixe` | `_helper`, `_cache` |
| évitez | `l`, `O`, `I` | (confusion avec 1 et 0) |

````python
# ❌ mauvais
def CalculerPrix(PrixHT, Taux):
    return PrixHT + PrixHT * Taux

# ✅ bon
TAUX_TVA = 0.20

def calculer_ttc(prix_ht: float, taux: float = TAUX_TVA) -> float:
    return prix_ht + prix_ht * taux
````

## 7. Docstrings — style Google

Chaque fonction publique a une **docstring**. Le style Google est lisible, simple et bien supporté par les outils (Sphinx, VS Code, Ruff).

````python
def retirer(solde: float, montant: float) -> float:
    """Retire un montant d'un solde.

    Args:
        solde: solde actuel (en euros).
        montant: montant à retirer (positif).

    Returns:
        Le nouveau solde après retrait.

    Raises:
        ValueError: si `montant` est négatif ou dépasse le solde.
    """
    if montant < 0 or montant > solde:
        raise ValueError('montant invalide')
    return solde - montant
````

````python
help(retirer)
````

### Sections usuelles

| Section | Quand |
|---|---|
| `Args:` | liste des paramètres |
| `Returns:` | description du retour |
| `Raises:` | exceptions levées |
| `Yields:` | générateurs |
| `Examples:` | code démonstratif |

### Fonction avec beaucoup de paramètres

````python
def creer_utilisateur(
    nom: str,
    email: str,
    age: int,
    ville: str = 'inconnue',
    actif: bool = True,
) -> dict[str, object]:
    """Crée un dictionnaire utilisateur."""
    return {
        'nom': nom,
        'email': email,
        'age': age,
        'ville': ville,
        'actif': actif,
    }

creer_utilisateur('Alice', 'alice@x.fr', 30)
````

### Docstring avec `Examples`

````python
def carre(x: int) -> int:
    """Renvoie le carré de x.

    Args:
        x: un entier.

    Returns:
        x * x.

    Examples:
        >>> carre(5)
        25
        >>> carre(-3)
        9
    """
    return x * x

help(carre)
````

Le bloc `Examples` suit le style `doctest` : les lignes `>>>` peuvent être exécutées automatiquement par `pytest --doctest-modules`.

### Commentaires vs docstrings

- **Commentaire** `# ...` : explique *pourquoi*, pas *quoi*. Pour les lecteurs du code.
- **Docstring** `"""..."""` : explique *quoi* et *comment l'utiliser*. Pour les utilisateurs de la fonction.

## Cheat sheet — PEP 8 en 10 lignes

| Règle | Valeur |
|---|---|
| Indentation | 4 espaces |
| Longueur de ligne | 88 (ruff) ou 79 (PEP 8 strict) |
| Lignes blanches entre fonctions top-level | 2 |
| Lignes blanches entre méthodes | 1 |
| Variable / fonction | `snake_case` |
| Constante | `UPPER_SNAKE_CASE` |
| Classe | `PascalCase` |
| Privé | `_prefixe` |
| Import par ligne | 1 |
| Espaces autour de `=` par défaut d'arg | **pas d'espace** |

````python
# Exemple PEP 8-compliant minimal
MAX_TENTATIVES = 3


def saluer(nom: str, politesse: str = 'Bonjour') -> str:
    """Construit une salutation."""
    return f'{politesse} {nom} !'


class Utilisateur:
    """Un utilisateur simple."""

    def __init__(self, nom: str) -> None:
        self.nom = nom


print(saluer('Alice'))
````

### Checklist docstring

- [ ] Première ligne : une phrase impérative (« Renvoie…», « Calcule…»)
- [ ] Section `Args:` si paramètres
- [ ] Section `Returns:` sauf si `-> None`
- [ ] Section `Raises:` si la fonction peut lever une exception
- [ ] `Examples:` si l'usage n'est pas évident
