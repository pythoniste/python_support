## 1. Qu'est-ce qu'un CSV ?

**Comma-Separated Values** : un format texte où chaque ligne est un enregistrement, les champs sont séparés par un délimiteur (virgule, point-virgule, tabulation…). Très répandu, très simple, et plein de pièges (virgules dans les valeurs, guillemets, sauts de ligne…).

**Ne jamais parser un CSV « à la main »** — toujours passer par `csv` (ou `pandas`).

````python
from pathlib import Path
contenu = 'nom,age,ville\nAlice,30,Paris\nBob,25,Lyon\nChloé,28,Marseille\n'
Path('/tmp/demo.csv').write_text(contenu, encoding='utf-8')
````

## 2. `csv.reader`

`csv.reader` renvoie chaque ligne sous forme de **liste de chaînes**. Pas de conversion auto : `'30'` reste une chaîne.

````python
import csv
with open('/tmp/demo.csv', 'r', encoding='utf-8', newline='') as f:
    lecteur = csv.reader(f)
    for ligne in lecteur:
        print(ligne)
````

### Sauter l'en-tête

Utiliser `next(lecteur)` pour consommer la première ligne.

````python
with open('/tmp/demo.csv', 'r', encoding='utf-8', newline='') as f:
    lecteur = csv.reader(f)
    entete = next(lecteur)
    print('colonnes :', entete)
    for ligne in lecteur:
        print(ligne)
````

---


## 3. `csv.DictReader`

`DictReader` utilise la première ligne comme en-tête et renvoie chaque enregistrement sous forme de `dict[str, str]`. **Beaucoup plus lisible** que l'indexation par position.

````python
import csv
with open('/tmp/demo.csv', 'r', encoding='utf-8', newline='') as f:
    lecteur = csv.DictReader(f)
    for ligne in lecteur:
        print(ligne)
````

On accède aux colonnes par leur nom :

````python
with open('/tmp/demo.csv', 'r', encoding='utf-8', newline='') as f:
    lecteur = csv.DictReader(f)
    for ligne in lecteur:
        print(ligne['nom'], '-', ligne['ville'])
````

## 4. `csv.writer`

````python
import csv
with open('/tmp/out.csv', 'w', encoding='utf-8', newline='') as f:
    w = csv.writer(f)
    w.writerow(['nom', 'age', 'ville'])
    w.writerow(['Alice', 30, 'Paris'])
    w.writerow(['Bob', 25, 'Lyon'])

print(open('/tmp/out.csv', encoding='utf-8').read())
````

### Écrire plusieurs lignes en une fois

````python
lignes = [['David', 22, 'Nice'], ['Eve', 40, 'Toulouse']]
with open('/tmp/out.csv', 'a', encoding='utf-8', newline='') as f:
    w = csv.writer(f)
    w.writerows(lignes)

print(open('/tmp/out.csv', encoding='utf-8').read())
````

## 5. `csv.DictWriter`

````python
import csv

donnees = [
    {'nom': 'Alice', 'age': 30, 'ville': 'Paris'},
    {'nom': 'Bob', 'age': 25, 'ville': 'Lyon'},
]

with open('/tmp/dictout.csv', 'w', encoding='utf-8', newline='') as f:
    w = csv.DictWriter(f, fieldnames=['nom', 'age', 'ville'])
    w.writeheader()
    w.writerows(donnees)

print(open('/tmp/dictout.csv', encoding='utf-8').read())
````

## 6. Dialectes

Un **dialecte** décrit les règles du CSV : délimiteur, caractère de quoting, saut de ligne. Python en propose plusieurs prédéfinis (`excel`, `excel-tab`, `unix`).

````python
import csv
# Variante point-virgule (habituelle en France)
contenu = 'nom;age;ville\nAlice;30;Paris\n'
open('/tmp/demo_fr.csv', 'w', encoding='utf-8', newline='').write(contenu)

with open('/tmp/demo_fr.csv', 'r', encoding='utf-8', newline='') as f:
    lecteur = csv.reader(f, delimiter=';')
    for ligne in lecteur:
        print(ligne)
````

### `csv.Sniffer` — auto-détection (aperçu)

````python
import csv
with open('/tmp/demo_fr.csv', 'r', encoding='utf-8', newline='') as f:
    dialecte = csv.Sniffer().sniff(f.read(1024))
print(dialecte.delimiter)
````

## 7. Pourquoi `newline=''` ?

`csv` gère lui-même les fins de ligne. Si on laisse Python faire la traduction (ce qu'il fait par défaut sur Windows), on obtient des **lignes vides parasites**. **Toujours** ouvrir en `newline=''` quand on utilise `csv`.

---


## 8. Synthèse

| Besoin | Outil |
|---|---|
| Lire (listes) | `csv.reader(f)` |
| Lire (dicts) | `csv.DictReader(f)` |
| Écrire (listes) | `csv.writer(f).writerow(...)` |
| Écrire (dicts) | `csv.DictWriter(f, fieldnames=...)` |
| Délimiteur `;` | `delimiter=';'` |

### Règles

1. Toujours `open(..., newline='', encoding='utf-8')`.
2. `DictReader`/`DictWriter` sont plus lisibles que leur équivalent positionnel.
3. Les valeurs lues sont **toujours des chaînes** — convertir si besoin.

import csv

def compter_lignes_csv(chemin: str) -> int:
    """Renvoie le nombre de lignes hors en-tête."""
    with open(chemin, 'r', encoding='utf-8', newline='') as f:

        next(lecteur, None)  # en-tête

import csv

def moyenne_colonne(chemin: str, colonne: str) -> float:
    """Moyenne d'une colonne numérique d'un CSV."""
    total = 0.0
    n = 0
    with open(chemin, 'r', encoding='utf-8', newline='') as f:
        for ligne in csv.DictReader(f):
            total = total + float(ligne[colonne])
            n = n + 1
    return total / n if n > 0 else 0.0

from pathlib import Path

import csv

def convertir(source: str, destination: str) -> None:
    """Convertit un CSV point-virgule en CSV virgule."""
    with open(source, 'r', encoding='utf-8', newline='') as f_in:
        lecteur = csv.reader(f_in, delimiter=';')
        with open(destination, 'w', encoding='utf-8', newline='') as f_out:
            ecriveur = csv.writer(f_out, delimiter=',')
            for ligne in lecteur:
                ecriveur.writerow(ligne)

convertir('/tmp/fr.csv', '/tmp/en.csv')
print(open('/tmp/en.csv', encoding='utf-8').read())

import csv

def ca_par_produit(chemin: str) -> dict[str, float]:
    """Agrège le chiffre d'affaires par produit."""
    ca: dict[str, float] = {}
    with open(chemin, 'r', encoding='utf-8', newline='') as f:
        for ligne in csv.DictReader(f):
            produit = ligne['produit']
            montant = int(ligne['quantite']) * float(ligne['prix_unitaire'])
            ca[produit] = ca.get(produit, 0.0) + montant
    return ca

from pathlib import Path

    'produit,quantite,prix_unitaire\n'
    'pomme,10,1.5\n'

    'pomme,3,1.5\n', encoding='utf-8')
print(ca_par_produit('/tmp/ventes.csv'))

### Écrire avec `quoting`

````python
import csv
with open('/tmp/quote.csv', 'w', encoding='utf-8', newline='') as f:
    w = csv.writer(f, quoting=csv.QUOTE_ALL)
    w.writerow(['nom', 'message'])
    w.writerow(['Alice', 'bonjour, monde'])
print(open('/tmp/quote.csv', encoding='utf-8').read())
````

`QUOTE_ALL` place des guillemets autour de **chaque** champ. `QUOTE_MINIMAL` (défaut) n'en met qu'autour des champs qui contiennent le séparateur ou un guillemet.

### Parcourir 1000 lignes sans tout charger

````python
import csv
import itertools
from pathlib import Path

lignes = ['nom,score\n'] + [f'eleve_{i},{i % 20}\n' for i in range(1000)]
Path('/tmp/big.csv').write_text(''.join(lignes), encoding='utf-8')

with open('/tmp/big.csv', 'r', encoding='utf-8', newline='') as f:
    lecteur = csv.DictReader(f)
    for ligne in itertools.islice(lecteur, 5):
        print(ligne)
````

### Sauts de ligne dans un champ

````python
import csv
with open('/tmp/multiline.csv', 'w', encoding='utf-8', newline='') as f:
    w = csv.writer(f)
    w.writerow(['nom', 'bio'])
    w.writerow(['Alice', 'ligne 1\nligne 2'])

with open('/tmp/multiline.csv', 'r', encoding='utf-8', newline='') as f:
    for ligne in csv.reader(f):
        print(ligne)
````

C'est la raison exacte pour laquelle `newline=''` est obligatoire : sans ça, les `\n` internes cassent le parseur.
