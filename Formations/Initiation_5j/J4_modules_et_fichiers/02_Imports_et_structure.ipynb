## 1. Un module, c'est quoi exactement ?

Un **module** est un fichier `.py`. Un **package** est un dossier contenant un `__init__.py` (ou considéré comme un package namespace depuis Python 3.3).

Le nom du module = le nom du fichier sans `.py`.

---


## 2. Formes d'import

````python
import math
math.sqrt(16)
````

````python
from math import sqrt
sqrt(16)
````

````python
from math import sqrt as racine
racine(16)
````

### `from module import *` — à proscrire

Importe tous les noms publics du module. Pollue l'espace de noms, rend le code difficile à tracer.

````python
# from math import *   # ❌ à proscrire en production
````

---


## 3. Absolu vs relatif

- **Import absolu** : chemin complet depuis la racine du projet — `from mon_projet.utils import helper`.
- **Import relatif** : depuis le module courant — `from .utils import helper`.

**Convention du cours :** on utilise **uniquement des imports absolus**. Ils sont plus lisibles et résistent aux refactorings.

---


## 4. Structure d'un petit projet

```
mon_projet/
├── pyproject.toml
├── README.md
├── src/
│   └── mon_projet/
│       ├── __init__.py
│       ├── cli.py
│       ├── core.py
│       └── utils.py
└── tests/
    ├── __init__.py
    ├── test_core.py
    └── test_utils.py
```

### Le « src layout »

Placer le package dans `src/` évite les imports accidentels quand on lance Python depuis la racine. C'est la structure **recommandée** par la Python Packaging Authority (PyPA).

## 5. `__name__ == '__main__'`

Chaque module a une variable spéciale `__name__`. Elle vaut :

- `'__main__'` si le module est exécuté directement (`python mon_fichier.py`) ;
- le nom du module sinon (quand il est importé).

````python
print(__name__)  # 'main' dans certains runtimes, '__main__' dans un script
````

### Idiome classique

```python
def principal() -> None:
    print('démarrage...')

if __name__ == '__main__':
    principal()
```

Cela permet d'avoir du code **exécuté** si on lance le fichier, mais **pas** s'il est juste importé.

---


## 6. `sys.path`

Pour trouver un module, Python parcourt la liste `sys.path`. Elle contient :

1. le dossier du script exécuté ;
2. le contenu de `PYTHONPATH` ;
3. les dossiers standard de la distribution (stdlib, site-packages).

Un module installé via `pip`/`uv` atterrit dans `site-packages`, accessible depuis n'importe où.

````python
import sys
sys.path[:3]
````

---


## 7. Pièges d'imports

### Imports circulaires

Si `a.py` importe `b.py` et `b.py` importe `a.py`, Python peut se retrouver à n'importer qu'à moitié l'un des deux modules. **Solution :** déplacer le code partagé dans un troisième module.

### Écraser un nom builtin

Ne nommez pas un fichier `random.py` ou `math.py` dans votre projet — il écrasera le module standard.

---


## 8. Synthèse

| Élément | Rôle |
|---|---|
| `module.py` | Fichier .py = un module |
| `package/__init__.py` | Dossier = package |
| `import module` | Import d'un module complet |
| `from module import x` | Import sélectif |
| `as` | Alias |
| `__name__` | `'__main__'` si exécuté directement |
| `sys.path` | Liste des dossiers de recherche |

### Règles

1. Imports **absolus** uniquement.
2. Un seul **module** par fichier, un seul **sujet** par module.
3. `src/` layout pour tout projet sérieux.
4. `if __name__ == '__main__':` pour le point d'entrée CLI.

**Question 1.** Qu'est-ce qui transforme un dossier en package en Python 3 ?

**Question 2.** Que vaut `__name__` quand un module est exécuté directement ?

<details>
<summary>📖 Réponse</summary>

`'__main__'`. Quand il est importé, `__name__` vaut le nom du module.

</details>

**Question 3.** Pourquoi préférer les imports absolus ?

<details>
<summary>📖 Réponse</summary>

Ils restent valides en cas de refactoring et sont immédiatement compréhensibles sans connaître la position du fichier courant.

</details>

**Question 4.** Où Python cherche-t-il les modules ?

<details>
<summary>📖 Réponse</summary>

Dans les dossiers listés par `sys.path` : dossier du script, `PYTHONPATH`, `site-packages`.

</details>

**Question 5.** Qu'est-ce qu'un import circulaire et comment le résoudre ?

<details>
<summary>📖 Réponse</summary>

`a.py` importe `b.py` qui importe `a.py`. Solution : extraire le code partagé dans un troisième module neutre.

</details>
