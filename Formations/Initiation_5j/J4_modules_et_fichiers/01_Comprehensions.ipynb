## 5. Générateurs : `(x for x in ...)`

Remplacer les crochets `[...]` par des parenthèses `(...)` donne un **générateur**. Un générateur **ne stocke pas** la liste complète en mémoire : il produit les valeurs à la demande.

````python
gen = (x * x for x in range(5))
print(gen)
````

````python
for v in gen:
    print(v)
````

Un générateur **s'épuise** : après la première boucle, il est vide.

````python
list(gen)
````

### Économie de mémoire

````python
import sys

l = [x * x for x in range(10_000)]
g = (x * x for x in range(10_000))
print('list :', sys.getsizeof(l))
print('gen  :', sys.getsizeof(g))
````

### Consommation efficace avec `sum`, `max`, `any`, `all`

Ces fonctions acceptent un générateur sans qu'on ait besoin de parenthèses supplémentaires.

````python
sum(x * x for x in range(10))
````

````python
any(x > 100 for x in range(50))
````

## 6. `yield` — fonctions génératrices (aperçu)

`yield` transforme une fonction en générateur. Chaque `yield` suspend l'exécution et renvoie une valeur.

````python
def compte_a_rebours(n: int):
    while n > 0:
        yield n
        n = n - 1

for v in compte_a_rebours(5):
    print(v)
````

## 7. `itertools` — outils utiles

Le module `itertools` fournit des briques très efficaces pour travailler avec des itérables.

````python
import itertools
````

### `count` — compteur infini

````python
gen = itertools.count(start=10, step=2)
for _ in range(5):
    print(next(gen))
````

### `cycle` — boucle infinie sur un itérable

````python
gen = itertools.cycle(['rouge', 'vert', 'bleu'])
for _ in range(6):
    print(next(gen))
````

### `chain` — enchaîner plusieurs itérables

````python
list(itertools.chain([1, 2, 3], [10, 20], [100]))
````

### `combinations` et `permutations`

````python
list(itertools.combinations('abc', 2))
````

````python
list(itertools.permutations('abc', 2))
````

### `groupby` — regrouper les éléments consécutifs égaux

````python
data = [('a', 1), ('a', 2), ('b', 3), ('b', 4), ('a', 5)]
for cle, groupe in itertools.groupby(data, key=lambda x: x[0]):
    print(cle, list(groupe))
````

## 8. Quand ne **pas** utiliser de compréhension

- quand elle dépasse 1 à 2 lignes : on perd en lisibilité ;
- quand on a besoin d'effets de bord (`print`, modifs d'état) : un `for` est plus clair ;
- quand l'imbrication dépasse 2 niveaux : éclater en fonctions intermédiaires.

````python
# À éviter : trop imbriqué, illisible
# [f(x, y, z) for x in A for y in B if cond(x, y) for z in C if autre_cond(y, z)]
pass
````

## 9. Synthèse

| Forme | Résultat |
|---|---|
| `[f(x) for x in it]` | list |
| `{f(x) for x in it}` | set |
| `{k: v for k, v in it}` | dict |
| `(f(x) for x in it)` | generator |
| `[f(x) for x in it if cond]` | filtré |
| `[f(x, y) for x in a for y in b]` | imbriqué |

### Règles

1. Préférer une compréhension à un `for` + `append` quand c'est lisible.
2. Préférer un **générateur** quand on ne consomme qu'une fois.
3. Ne pas empiler plus de 2 niveaux de `for`.

```python
def pascal(n: int) -> list[list[int]]:
    """n premières lignes du triangle de Pascal."""
    triangle: list[list[int]] = []
    for i in range(n):
        ligne = [1 if j == 0 or j == i
                 else triangle[i - 1][j - 1] + triangle[i - 1][j]
                 for j in range(i + 1)]
        triangle.append(ligne)
    return triangle

for ligne in pascal(6):
    print(ligne)
```

```python
def top_n(notes: dict[str, int], n: int) -> list[tuple[str, int]]:
    """Renvoie les n meilleurs couples (nom, note) triés par note décroissante."""
    classement = sorted(notes.items(), key=lambda item: item[1], reverse=True)
    return classement[:n]

notes = {'Alice': 15, 'Bob': 12, 'Chloé': 18, 'David': 9}
```
