## 1. `print` : les bases

`print(...)` affiche une ou plusieurs valeurs à l'écran, séparées par des espaces, puis passe à la ligne.


````python
print("Bonjour")
````

````python
print(42)
````

````python
print("Age :", 30)
````

````python
print(1, 2, 3, 4, 5)
````

**Observation** : Python met automatiquement un **espace** entre chaque valeur, et un **retour à la ligne** à la fin.

## 2. Plusieurs valeurs avec `sep`

Le paramètre nommé `sep` (**sep**arator) change le caractère mis entre les valeurs. Par défaut c'est `" "`.


````python
print("a", "b", "c")
````

````python
print("a", "b", "c", sep="-")
````

````python
print("a", "b", "c", sep="")
````

````python
print(2026, 4, 14, sep="/")
````

## 3. `end` : contrôler le retour à la ligne

Par défaut, `print` ajoute `"\n"` (retour à la ligne) à la fin. On peut changer ce comportement avec le paramètre `end`.


````python
print("sans retour", end="")
print(" — suite sur la même ligne")
````

````python
print("ligne 1", end=" | ")
print("ligne 2")
````

C'est utile pour afficher sur une seule ligne plusieurs éléments calculés un par un *(vous le verrez avec les boucles)*.

## 4. `flush` : forcer l'affichage immédiat

Par défaut, `print` **bufferise** sa sortie : elle est affichée à la fin de la ligne, ou quand le tampon est plein. Pour des affichages en **temps réel** (barre de progression, log de long calcul), on force le vidage avec `flush=True`.


````python
print("Début du traitement...", flush=True)
````

En Jupyter, le rendu est quasi-immédiat de toute façon. Le `flush` est surtout utile dans un script lancé depuis un terminal.

## 5. `input` : lire du texte

`input(...)` affiche un **prompt** (message) à l'écran et **attend** que l'utilisateur tape quelque chose suivi de `Entrée`. La valeur retournée est **toujours** une chaîne de caractères.


**Note importante** : dans un notebook Jupyter, `input()` ouvre une petite boîte de saisie en haut du notebook. Dans un script lancé en terminal, c'est le curseur qui attend au prompt.

Les exemples ci-dessous sont **ré-exécutables** dans un notebook : Jupyter demandera la valeur au moment de l'exécution.


````python
prenom = input("Votre prénom : ")
````

````python
print("Bonjour,", prenom)
````

Le prompt peut contenir plusieurs mots, y compris des caractères spéciaux :

````python
reponse = input("Tapez o/n : ")
````

````python
print("Vous avez tapé :", reponse)
````

## 6. Convertir la saisie en nombre

Comme `input` renvoie **toujours une chaîne**, il faut explicitement convertir si on veut un nombre.


````python
texte = input("Âge : ")
age = int(texte)
print("L'an prochain, vous aurez :", age + 1)
````

Raccourci : conversion **immédiate** à la lecture.

````python
age = int(input("Âge : "))
print("L'an prochain :", age + 1)
````

Pour un flottant, même principe :

````python
prix = float(input("Prix : "))
ttc = prix * 1.20
print("TTC :", round(ttc, 2))
````

**Attention** : si l'utilisateur tape quelque chose qui n'est pas un nombre valide, `int(...)` lève une `ValueError`. On apprendra à gérer ça avec `try/except` dans le notebook sur les exceptions.


## 7. Échapper les caractères spéciaux (`\n`, `\t`)

Dans une chaîne, certains caractères s'écrivent avec un **antislash** suivi d'une lettre :

| Séquence | Sens |
|---|---|
| `\n` | Retour à la ligne |
| `\t` | Tabulation |
| `\\` | Antislash |
| `\"` | Guillemet double dans une chaîne entre `"..."` |
| `\'` | Guillemet simple dans une chaîne entre `'...'` |


````python
print("ligne 1\nligne 2")
````

````python
print("Nom\tPrénom\tAge")
````

````python
print("C:\\Users\\alice")
````

````python
print("Elle a dit : \"bonjour\".")
````

On approfondira ces concepts dans `03_Chaines.ipynb`, mais ces quatre séquences suffisent pour aligner proprement un affichage simple.

## 8. Synthèse

### Paramètres utiles de `print`

| Paramètre | Défaut | Rôle |
|---|---|---|
| `sep` | `" "` | Séparateur entre les arguments |
| `end` | `"\n"` | Caractère(s) ajouté(s) à la fin |
| `flush` | `False` | Forcer le vidage du buffer |

### Règles à retenir

1. **`print` ajoute un espace entre chaque argument** (paramètre `sep`).
2. **`print` ajoute `\n` à la fin** (paramètre `end`).
3. **`input` renvoie toujours une chaîne** — convertir explicitement si besoin.
4. **`int(input("..."))`** est le pattern pour lire un entier.
5. **`\n` et `\t`** dans une chaîne produisent un saut de ligne et une tabulation.
