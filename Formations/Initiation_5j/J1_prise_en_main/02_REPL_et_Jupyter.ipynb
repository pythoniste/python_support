## 1. Qu'est-ce que le REPL ?

**REPL** = **R**ead – **E**val – **P**rint – **L**oop.

C'est une **boucle interactive** qui :

1. **Lit** une expression tapée par l'utilisateur,
2. l'**évalue**,
3. en **affiche** le résultat,
4. **recommence**.

On l'utilise pour tester rapidement une idée, sans créer de fichier.


## 2. Lancer le REPL

Dans un environnement virtuel activé, taper simplement :

```
python
```

Vous voyez apparaître quelque chose comme :

```
Python 3.14.0 (main, Oct ...)
Type "help", "copyright", "credits" or "license" for more information.
>>>
```

Le `>>>` est l'**invite primaire** : le REPL attend que vous tapiez quelque chose.


## 3. Expressions vs instructions

Dans le REPL, **tout ce que vous tapez est immédiatement évalué et affiché**.

Exemple de session, ligne par ligne :

```
>>> 2 + 3
5
>>> 10 * 7
70
>>> 2 ** 10
1024
```

Python calcule, puis affiche automatiquement le résultat. C'est la grande différence avec un script où il faut `print(...)`.

Quand vous taperez ces mêmes expressions **dans une cellule Jupyter**, le même affichage automatique s'appliquera pour la dernière expression de la cellule.


````python
2 + 3
````

````python
10 * 7
````

````python
2 ** 10
````

Voilà : chaque cellule ci-dessus se comporte comme une ligne de REPL. C'est **exactement** le même modèle d'interaction.

## 4. Quitter le REPL

Trois façons équivalentes :

| Méthode | Plateforme |
|---|---|
| Taper `exit()` | Toutes |
| Taper `quit()` | Toutes |
| `Ctrl` + `D` | Linux, macOS |
| `Ctrl` + `Z` puis `Entrée` | Windows |


## 5. Le REPL amélioré de Python 3.13+

Depuis Python 3.13, le REPL standard est **beaucoup plus ergonomique** :

- **édition multi-ligne** : on peut taper un bloc complet et le relire avec les flèches ;
- **coloration syntaxique** dans le terminal ;
- **complétion** automatique avec `Tab` ;
- **commandes spéciales** : `help`, `exit`, `clear` *(sans parenthèses)*.

Python 3.14 améliore encore cette expérience *(c'est la version que nous utilisons)*.


## 6. Script, REPL, Jupyter — quel outil quand ?

| Outil | Forces | Limites |
|---|---|---|
| **Script** (`python monscript.py`) | Reproductible, versionnable, utilisable en production | Pas interactif, doit être relancé pour chaque essai |
| **REPL** (`python`) | Ultra-rapide pour tester une idée | Session perdue à la fermeture, pas d'historique riche |
| **Jupyter** (notebook) | Mélange code + texte + graphiques, cellules ré-exécutables indépendamment | Pas fait pour la production, les cellules peuvent être ré-exécutées dans le désordre |

**Règle de la formation** : on apprend en **Jupyter** (cellules courtes, feedback immédiat), on livre en **script** (packagé en J5).

## 9. Anatomie d'un notebook

Un **notebook** (extension `.ipynb`) est une **liste de cellules** enregistrées au format JSON.

Deux types de cellules principaux :

| Type | Contenu | Ce qui arrive à l'exécution |
|---|---|---|
| **Code** | Code Python | Exécuté dans le kernel, résultat affiché |
| **Markdown** | Texte mis en forme (titres, liens, listes, ...) | Rendu HTML dans la cellule |

Chaque cellule de code a une **invite** `[ ]` à gauche :

- `[ ]` : jamais exécutée ;
- `[*]` : en cours d'exécution ;
- `[3]` : exécutée, numéro d'exécution = 3.


````python
# Ceci est une cellule de code
# Quand vous l'exécutez, Python évalue chaque ligne
42
````

La dernière expression d'une cellule (ici `42`) est **affichée automatiquement**, comme dans le REPL.

````python
# Plusieurs expressions — seule la dernière est affichée
2 + 3
4 * 5
10 - 7
````

Pour forcer l'affichage de plusieurs valeurs dans une même cellule, on utilise `print(...)` — c'est la règle en situation script.

````python
print(2 + 3)
print(4 * 5)
print(10 - 7)
````

## 10. Modes commande et édition

Jupyter distingue **deux modes** d'interaction avec une cellule :

| Mode | Bordure | Action |
|---|---|---|
| **Édition** | Verte | Taper du code, modifier la cellule |
| **Commande** | Bleue | Déplacer, ajouter, supprimer, taguer des cellules |

- `Échap` → passe en mode **commande**.
- `Entrée` → passe en mode **édition** sur la cellule sélectionnée.
- Les **raccourcis clavier** ne sont actifs qu'en mode commande *(sinon vous taperiez du code !)*.


## 11. Raccourcis clavier indispensables

Connaître ces raccourcis divise par trois le temps passé à manipuler un notebook.

### En mode commande (bordure bleue)

| Raccourci | Effet |
|---|---|
| `A` | Insérer une cellule **A**u-dessus |
| `B` | Insérer une cellule en dessous (**B**elow) |
| `D`, `D` | Supprimer la cellule (**D**elete, deux fois) |
| `M` | Transformer en **M**arkdown |
| `Y` | Transformer en code (p**Y**thon) |
| `↑` / `↓` | Cellule précédente / suivante |
| `Entrée` | Passer en mode édition |
| `Shift + Entrée` | Exécuter la cellule et passer à la suivante |
| `Ctrl + Entrée` | Exécuter **sans** bouger |
| `Alt + Entrée` | Exécuter et insérer une cellule dessous |
| `Z` | An**Z**nuler la dernière action (undo) |


### En mode édition (bordure verte)

| Raccourci | Effet |
|---|---|
| `Tab` | Compléter / indenter |
| `Shift + Tab` | Afficher la documentation de la fonction sous le curseur |
| `Ctrl + /` | Commenter / décommenter la ligne |
| `Échap` | Revenir en mode commande |

### Les trois à mémoriser **absolument**

1. `Shift + Entrée` — exécuter et passer à la suivante.
2. `A` / `B` en mode commande — insérer au-dessus / en dessous.
3. `Échap` — quitter l'édition.

Avec ces trois raccourcis, on navigue dans un notebook sans toucher la souris.


````python
# Essayez maintenant : sélectionnez cette cellule, appuyez sur Échap
# puis sur A pour insérer une cellule au-dessus.
print("Les raccourcis, c'est la vie")
````

## Commandes magiques Jupyter *(bonus)*

Jupyter ajoute des **« magic commands »** préfixées par `%` (ligne) ou `%%` (cellule) qui ne sont pas du Python standard mais des aides pratiques.

| Magic | Effet |
|---|---|
| `%time expression` | Temps d'exécution d'une expression |
| `%timeit expression` | Benchmark (moyenne sur plusieurs exécutions) |
| `%%writefile nom.py` | Écrire la cellule dans un fichier |
| `%ls` / `%pwd` | Commandes shell équivalentes |
| `!commande` | Exécuter une commande shell |

On les verra en détail plus tard — pour l'instant, on se limite à savoir qu'elles existent.


````python
%time 2 ** 1000
````

## 12. Synthèse

### REPL vs script vs notebook

| Outil | Quand l'utiliser |
|---|---|
| **REPL** | Tester une idée en 10 secondes, vérifier une syntaxe |
| **Script** | Livrer un programme exécutable, automatiser une tâche |
| **Notebook** | Apprendre, explorer des données, documenter une démarche |

### Raccourcis vitaux

- `Shift + Entrée` — exécuter la cellule
- `A` / `B` — insérer au-dessus / en dessous *(mode commande)*
- `D D` — supprimer la cellule *(mode commande)*
- `M` / `Y` — basculer Markdown / Code
- `Échap` / `Entrée` — mode commande / édition
- `Shift + Tab` sur un nom — afficher la doc

### Règles à retenir

1. **L'ordre des cellules dans un notebook n'est pas forcément l'ordre d'exécution.** Toujours relancer « Kernel → Restart & Run All » avant de conclure que ça marche.
2. **Un notebook n'est pas un script** : pour livrer, on extrait le code dans un `.py`.
3. **On versionne les notebooks** (Git), mais l'état du kernel n'est pas versionné — seul le fichier `.ipynb` l'est.

### Exercice 8 — Magic command *(difficile)*

<details>

```python

# %time fait une seule mesure.
# %timeit fait plusieurs exécutions pour donner une moyenne fiable.
```
