## 1. Qu'est-ce qu'une variable ?

Une **variable** est un **nom** qu'on donne à une valeur, pour pouvoir la réutiliser et la modifier.

Analogie : une étiquette collée sur une boîte. La boîte contient une valeur, l'étiquette permet de la retrouver.


## 2. L'opérateur d'affectation `=`

Le signe `=` en Python n'est **pas** l'égalité mathématique. C'est une **affectation** : « mettre la valeur de droite dans la variable de gauche ».


````python
age = 30
````

````python
print(age)
````

Une fois déclarée, la variable peut être utilisée à la place de la valeur.

````python
annee = 2026
````

````python
print(annee)
````

On peut **réaffecter** une variable : elle prend une nouvelle valeur.

````python
age = 30
````

````python
age = 31
````

````python
print(age)
````

### Affectations multiples

Python permet d'affecter plusieurs variables en une ligne :


````python
x = y = z = 0
````

````python
print(x, y, z)
````

Ou de déballer plusieurs valeurs à la fois :

````python
a, b = 10, 20
````

````python
print(a, b)
````

## 3. La fonction `type()`

`type(x)` renvoie le **type** de la valeur `x`. C'est la question « qu'est-ce que c'est ? » posée à Python.


````python
type(42)
````

````python
type(3.14)
````

````python
type("bonjour")
````

````python
type(True)
````

````python
type(None)
````

## 4. Les entiers (`int`)

Un **entier** est un nombre sans partie décimale. En Python, les entiers ont une **précision illimitée** — ils peuvent être aussi grands que la mémoire le permet.


````python
entier = 42
````

````python
print(entier)
````

````python
print(type(entier))
````

On peut utiliser des **underscores** pour rendre les grands nombres lisibles :

````python
population = 1_400_000_000
````

````python
print(population)
````

Les négatifs fonctionnent comme on s'y attend :

````python
temperature = -15
````

````python
print(temperature)
````

## 5. Les flottants (`float`)

Un **flottant** représente un nombre réel (avec partie décimale). Le séparateur est le **point**, pas la virgule.


````python
prix = 19.99
````

````python
print(prix)
````

````python
print(type(prix))
````

La notation scientifique est également supportée :

````python
grand = 1.5e9
````

````python
print(grand)
````

**Attention** : `4.` est valide et équivaut à `4.0`.

````python
x = 4.
````

````python
print(x)
````

````python
print(type(x))
````

## 6. Les chaînes de caractères (`str`)

Une **chaîne de caractères** représente du texte. On l'écrit entre guillemets : `"..."` ou `'...'`. Les deux sont équivalents.


````python
prenom = "Alice"
````

````python
print(prenom)
````

````python
print(type(prenom))
````

Les deux styles de guillemets marchent, et permettent d'inclure l'autre style dans la chaîne :

````python
phrase = 'Elle a dit "bonjour".'
````

````python
print(phrase)
````

On étudiera les chaînes **en profondeur** dans `03_Types_de_donnees/03_Chaines.ipynb`. Pour l'instant, on sait juste qu'elles existent et qu'on peut les stocker dans des variables.

## 7. Les booléens (`bool`)

Un **booléen** ne peut avoir que **deux valeurs** : `True` ou `False`. Attention à la majuscule — `true` (minuscule) n'existe pas en Python.


````python
est_admin = True
````

````python
print(est_admin)
````

````python
print(type(est_admin))
````

````python
est_actif = False
````

````python
print(est_actif)
````

Les booléens sont le **résultat** des comparaisons :

````python
resultat = 3 < 5
````

````python
print(resultat)
````

## 8. La valeur spéciale `None`

`None` représente **« rien »**, « aucune valeur », « absence ». Son type est `NoneType`.

On l'utilise quand une variable n'a **pas encore** de valeur, ou quand une fonction **ne retourne rien**.


````python
resultat = None
````

````python
print(resultat)
````

````python
print(type(resultat))
````

Pour tester si une variable vaut `None`, on utilise `is` et **pas** `==` :

````python
x = None
````

````python
print(x is None)
````

## 9. Conversion entre types

Python permet de **convertir** une valeur d'un type à un autre avec les fonctions du même nom que le type cible : `int(...)`, `float(...)`, `str(...)`, `bool(...)`.


### De chaîne à nombre

````python
texte = "42"
````

````python
nombre = int(texte)
````

````python
print(nombre)
````

````python
print(type(nombre))
````

````python
texte2 = "3.14"
````

````python
pi = float(texte2)
````

````python
print(pi)
````

**Attention** : `int("3.14")` lève une erreur. `int(...)` n'accepte que des chaînes représentant des **entiers**.

### De nombre à chaîne

````python
n = 42
````

````python
s = str(n)
````

````python
print(s)
````

````python
print(type(s))
````

### Entre int et float

````python
print(int(3.9))
````

`int()` sur un flottant **tronque** (supprime la partie décimale), il n'arrondit pas. On verra `round()` en détail dans le notebook sur les nombres.

````python
print(float(7))
````

### Vers bool

````python
print(bool(0))
````

````python
print(bool(1))
````

````python
print(bool(""))
````

````python
print(bool("non"))
````

On approfondira le « truthiness » dans le notebook sur les booléens. Retenez pour l'instant :

- `0`, `0.0`, `""`, `None` → `False`
- tout le reste → `True`


## 10. Règles de nommage

En Python, un **nom de variable** doit :

1. **commencer par une lettre** ou un `_` (pas par un chiffre) ;
2. ne contenir que des **lettres, chiffres, underscores** ;
3. **ne pas être un mot-clé** Python (`if`, `for`, `class`, ...) ;
4. respecter la casse : `age` et `Age` sont **deux variables différentes**.


### Convention PEP 8 — snake_case

Pour les variables et fonctions, on écrit en **snake_case** (minuscules, mots séparés par des underscores) :

| ❌ À éviter | ✅ Recommandé |
|---|---|
| `monAge` | `mon_age` |
| `NombreDeLignes` | `nombre_de_lignes` |
| `x1` | `premier_x` (si le contexte s'y prête) |

Les constantes s'écrivent en `MAJUSCULES_AVEC_UNDERSCORES`.


````python
mon_age = 30
````

````python
nombre_maximum = 100
````

````python
PI = 3.14159
````

### Mots-clés réservés

Python interdit certains noms — ce sont les **mots-clés** du langage :

```

break    class   continue def    del   elif     else    except
finally  for     from     global if    import   in      is
lambda   nonlocal not      or     pass  raise    return  try
```

## 11. Synthèse

### Les cinq types primitifs

| Type | Exemple | `type(...)` |
|---|---|---|
| Entier | `42`, `-7`, `1_000_000` | `int` |
| Flottant | `3.14`, `-0.5`, `1.5e9` | `float` |
| Chaîne | `"bonjour"`, `'hi'` | `str` |
| Booléen | `True`, `False` | `bool` |
| Absence | `None` | `NoneType` |

### Règles à retenir

1. **`=` est une affectation, pas une égalité.** L'égalité se teste avec `==` *(prochain notebook)*.
2. **On compare à `None` avec `is`**, pas `==`.
3. **`type(x)` répond « c'est quoi ? »**, `isinstance(x, T)` répond « c'en est un ? » *(vu plus tard)*.
4. **Les entiers Python ont une précision illimitée.**
5. **`int("3.14")` lève une erreur** — il faut passer par `float(...)` d'abord si besoin.
6. **snake_case** pour les noms de variables, `MAJUSCULES` pour les constantes.
