## 1. À quoi ça sert ?

Les *type hints* ajoutent à Python une **documentation structurée** sur les types attendus et renvoyés. Ils servent à :

- aider l'**éditeur** (autocomplétion, détection d'erreurs) ;
- aider le **lecteur** du code ;
- permettre à un outil comme `mypy` de **vérifier statiquement** la cohérence ;
- rendre le code plus facile à maintenir sur la durée.

À partir de maintenant, **tous les exercices sont typés**.

## 3. Types composites modernes (Python 3.9+)

Depuis Python 3.9, on utilise directement `list`, `dict`, `tuple`, `set` avec des crochets. Plus besoin d'importer `List`, `Dict`, etc. depuis `typing`.

````python
def somme(valeurs: list[int]) -> int:
    total = 0
    for v in valeurs:
        total = total + v
    return total
````

````python
somme([1, 2, 3, 4])
````

````python
def compter(mots: list[str]) -> dict[str, int]:
    resultat: dict[str, int] = {}
    for mot in mots:
        resultat[mot] = resultat.get(mot, 0) + 1
    return resultat
````

````python
compter(['a', 'b', 'a', 'c', 'b', 'a'])
````

### Tuples de taille fixe ou variable

````python
def premier(valeurs: tuple[int, ...]) -> int:
    """Premier élément d'un tuple de taille variable d'ints."""
    return valeurs[0]
````

## 4. Optionnels : `int | None` (Python 3.10+)

On écrit un type « ou l'autre » avec `|`. Très utile pour un paramètre qui peut être `None`.

````python
def age_ou_inconnu(age: int | None) -> str:
    if age is None:
        return 'âge inconnu'
    return f'{age} ans'
````

````python
age_ou_inconnu(30)
````

````python
age_ou_inconnu(None)
````

### Équivalent historique avec `Optional[int]`

Vous rencontrerez encore `from typing import Optional` dans du code ancien. `Optional[int]` est exactement `int | None`. Préférez la forme `|` dans le code neuf.

## 5. Les hints ne sont **pas** exécutés

Python n'impose **rien** à l'exécution. Si vous donnez un `str` à une fonction qui attend un `int`, Python ne lève pas d'erreur de type — il tente le calcul, qui peut réussir ou échouer.

````python
def double(x: int) -> int:
    return x * 2

# Ça marche, même si ce n'est pas un int
double('ab')
````

C'est pour ça qu'un **vérificateur statique** comme `mypy` est utile.

---


## 6. `mypy` et `mypy --strict`

`mypy` analyse votre code **sans l'exécuter** et signale les incohérences. Le mode `--strict` exige que **toutes** les fonctions soient annotées.

### Installation

```bash
uv add --dev mypy
uv run mypy --strict mon_fichier.py
```

Dans un notebook on ne lance pas `mypy` directement, mais la **règle du cours** est que tout exercice à partir d'ici doit être **`mypy --strict` vert**.

## 8. Synthèse

| Intention | Annotation |
|---|---|
| Entier | `int` |
| Flottant | `float` |
| Chaîne | `str` |
| Booléen | `bool` |
| Rien | `None` |
| Liste d'entiers | `list[int]` |
| Dict chaîne→entier | `dict[str, int]` |
| Tuple taille fixe | `tuple[float, float]` |
| Tuple variable homogène | `tuple[int, ...]` |
| Set | `set[str]` |
| Optionnel | `int \| None` |

### Règles

1. Toute fonction publique est annotée paramètres + retour.
2. `-> None` quand la fonction ne renvoie rien utile.
3. Préférer `int | None` à `Optional[int]` (Python 3.10+).
4. Préférer `list[int]` à `List[int]` (Python 3.9+).

## Mini-exemples supplémentaires

### Fonctions simples annotées

````python
def incrementer(x: int) -> int:
    return x + 1

incrementer(41)
````

````python
def majuscule(s: str) -> str:
    return s.upper()

majuscule('hello')
````

````python
def ou_bien(a: bool, b: bool) -> bool:
    return a or b

ou_bien(True, False)
````

### Listes et dicts annotés

````python
def doubler(valeurs: list[int]) -> list[int]:
    return [v * 2 for v in valeurs]

doubler([1, 2, 3])
````

````python
def inverser_map(d: dict[str, int]) -> dict[int, str]:
    return {v: k for k, v in d.items()}

inverser_map({'a': 1, 'b': 2})
````

### `set` et `tuple`

````python
def ensemble_mots(phrase: str) -> set[str]:
    return set(phrase.split())

ensemble_mots('le chat le chien')
````

````python
def extremes(valeurs: list[int]) -> tuple[int, int]:
    return (min(valeurs), max(valeurs))

extremes([4, 1, 9, 2])
````

### Optionnel

````python
def premier_element(valeurs: list[int]) -> int | None:
    if len(valeurs) == 0:
        return None
    return valeurs[0]

print(premier_element([]))
print(premier_element([1, 2, 3]))
````

### `Any` — porte de sortie (à éviter en général)

````python
from typing import Any

def echo(obj: Any) -> Any:
    return obj

print(echo(42))
print(echo('coucou'))
````

`Any` désactive la vérification de type pour ce nom. À réserver aux frontières du programme (I/O externe, `json.load`, etc.).
