## 1. Définir une fonction

Une **fonction** encapsule un bout de code sous un nom, pour le réutiliser. On la définit avec `def`.

````python
def bonjour():
    print('Bonjour !')
````

````python
bonjour()
````

La fonction existe comme objet dans la mémoire tant qu'on ne la redéfinit pas.

## 2. Paramètres, arguments, retour

````python
def carre(x):
    return x * x
````

````python
carre(5)
````

````python
carre(7)
````

- **paramètre** : le nom `x` dans la définition.
- **argument** : la valeur `5` passée à l'appel.
- **retour** : le résultat que la fonction renvoie avec `return`.

Sans `return`, une fonction renvoie implicitement `None`.

````python
def salue(nom):
    print('Bonjour', nom)

resultat = salue('Alice')
print('retour :', resultat)
````

## 3. Paramètres positionnels vs nommés

````python
def soustraire(a, b):
    return a - b
````

````python
soustraire(10, 3)       # positionnels
````

````python
soustraire(a=10, b=3)   # nommés
````

````python
soustraire(b=3, a=10)   # nommés, ordre libre
````

## 4. Valeurs par défaut

On peut donner une **valeur par défaut** à un paramètre. Il devient alors optionnel à l'appel.

````python
def saluer(nom='inconnu'):
    print('Bonjour', nom)
````

````python
saluer()
````

````python
saluer('Alice')
````

### Règle : les paramètres avec valeur par défaut viennent **après** les paramètres sans valeur.

````python
def addition(a, b=0, c=0):
    return a + b + c
````

````python
addition(1)
````

````python
addition(1, 2)
````

````python
addition(1, 2, 3)
````

## 5. `*args` — nombre variable de positionnels

`*args` capture dans un **tuple** tous les arguments positionnels supplémentaires.

````python
def somme(*args):
    total = 0
    for x in args:
        total = total + x
    return total
````

````python
somme(1, 2, 3, 4)
````

````python
somme()
````

Le nom `args` est une convention. Vous pouvez écrire `*nombres` si c'est plus clair. C'est le `*` qui compte.

````python
def afficher(*nombres):
    print(nombres)

afficher(1, 2, 3)
````

---


## 6. `**kwargs` — nombre variable de nommés

`**kwargs` capture dans un **dictionnaire** tous les arguments nommés supplémentaires.

````python
def decrire(**kwargs):
    for cle, valeur in kwargs.items():
        print(cle, '=', valeur)
````

````python
decrire(nom='Alice', age=30, ville='Paris')
````

## 7. Ordre des paramètres

L'ordre canonique dans une signature est :

```
def f(pos1, pos2, *args, clef1=valeur, **kwargs):
    ...
```

positionnels → `*args` → nommés avec défaut → `**kwargs`.

````python
def afficher(titre, *lignes, separateur='---', **metadata):
    print(titre)
    print(separateur)
    for ligne in lignes:
        print('-', ligne)
    if metadata:
        print(separateur)
        for cle, valeur in metadata.items():
            print(cle, ':', valeur)
````

````python
afficher(
    'Rapport',
    'ligne 1',
    'ligne 2',
    auteur='Alice',
    date='2026-04-14',
)
````

## 8. ⚠️ Piège du *mutable default argument*

**Ne jamais** utiliser une valeur par défaut mutable (`[]`, `{}`, `set()`). La valeur par défaut est évaluée **une seule fois**, à la définition de la fonction, et partagée entre tous les appels.

````python
def ajouter_faux(element, cible=[]):
    cible.append(element)
    return cible
````

````python
ajouter_faux('a')
````

````python
ajouter_faux('b')  # 'a' est encore là !
````

````python
ajouter_faux('c')
````

### Bonne pratique : utiliser `None` comme sentinelle

On met `None` par défaut et on crée la liste **à l'intérieur** de la fonction.

````python
def ajouter_bon(element, cible=None):
    if cible is None:
        cible = []
    cible.append(element)
    return cible
````

````python
ajouter_bon('a')
````

````python
ajouter_bon('b')  # une nouvelle liste à chaque appel
````

## 9. Docstring

Une **docstring** documente une fonction. Elle est écrite en toute première instruction, entre triples guillemets.

````python
def carre(x):
    """Renvoie le carré de x."""
    return x * x
````

````python
help(carre)
````

````python
carre.__doc__
````

Nous reviendrons sur les conventions de docstrings (Google / NumPy) au chapitre bonnes pratiques.

## 10. Synthèse

| Élément | Syntaxe |
|---|---|
| Définition | `def f(...):` |
| Retour | `return valeur` |
| Paramètre avec défaut | `def f(x=0):` |
| Positionnels variables | `def f(*args):` |
| Nommés variables | `def f(**kwargs):` |
| Ordre | `pos, *args, nom=def, **kw` |
| Docstring | `"""..."""` en 1re instruction |

### Règles à retenir

1. Pas de *mutable default* — utiliser `None` et créer à l'intérieur.
2. `*args` ↔ tuple, `**kwargs` ↔ dict.
3. Paramètres sans défaut avant ceux avec défaut.
4. Docstring obligatoire dès qu'une fonction dépasse 3 lignes.

```python
def appliquer(operation, *valeurs):
    """Applique une opération nommée aux valeurs."""
    if len(valeurs) == 0:
        return None
    if operation == 'somme':
        return sum(valeurs)
    if operation == 'moyenne':
        return sum(valeurs) / len(valeurs)
    if operation == 'max':
        return max(valeurs)
    if operation == 'min':
        return min(valeurs)
    return None

print(appliquer('somme', 1, 2, 3, 4))
print(appliquer('moyenne', 10, 20, 30))
print(appliquer('inconnue', 1, 2))
```
