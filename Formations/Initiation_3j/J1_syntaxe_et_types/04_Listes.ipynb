## 1. Créer une liste

Une liste est une **séquence ordonnée et modifiable** d'éléments. On la crée avec des crochets `[...]`.

````python
fruits = ["pomme", "banane", "cerise"]
````

````python
print(fruits)
````

````python
type(fruits)
````

Une liste peut contenir des éléments de **types différents**.

````python
melange = [42, 3.14, "hello", True]
````

````python
print(melange)
````

La liste vide se crée avec `[]` ou `list()`.

````python
vide_1 = []
````

````python
vide_2 = list()
````

````python
print(vide_1, vide_2)
````

`list()` peut aussi convertir un itérable en liste.

````python
list(range(5))
````

````python
list("Python")
````

La fonction `len()` donne le nombre d'éléments.

````python
len(fruits)
````

## 2. Accéder aux éléments

### Index positif

Les éléments sont numérotés à partir de **0**.

````python
fruits[0]
````

````python
fruits[1]
````

````python
fruits[2]
````

### Index négatif

Un index négatif compte depuis la fin : `-1` est le dernier élément.

````python
fruits[-1]
````

````python
fruits[-2]
````

### Slicing (tranches)

La syntaxe `liste[debut:fin]` extrait une sous-liste. `debut` est inclus, `fin` est **exclu**.

````python
nombres = [10, 20, 30, 40, 50, 60, 70]
````

````python
nombres[1:4]
````

````python
nombres[:3]   # du début jusqu'à l'index 3 (exclu)
````

````python
nombres[4:]   # de l'index 4 jusqu'à la fin
````

````python
nombres[::2]  # un élément sur deux
````

````python
nombres[::-1] # liste inversée
````

## 3. Modifier une liste (mutabilité)

Contrairement aux chaînes de caractères, les listes sont **mutables** : on peut modifier un élément en place.

````python
fruits[1] = "fraise"
````

````python
print(fruits)
````

On peut aussi remplacer une tranche entière.

````python
nombres[1:3] = [200, 300]
````

````python
print(nombres)
````

## 4. Méthodes de liste — ajout et suppression

### `append()` — ajouter à la fin

````python
fruits.append("kiwi")
````

````python
print(fruits)
````

### `insert()` — ajouter à un index précis

````python
fruits.insert(1, "mangue")
````

````python
print(fruits)
````

### `extend()` — ajouter plusieurs éléments

````python
fruits.extend(["ananas", "litchi"])
````

````python
print(fruits)
````

L'opérateur `+` crée une **nouvelle** liste (ne modifie pas l'originale).

````python
a = [1, 2]
b = [3, 4]
c = a + b
````

````python
print(c)
````

````python
print(a)  # a n'a pas changé
````

### `pop()` — retirer par index (et récupérer la valeur)

````python
dernier = fruits.pop()
````

````python
print(dernier)
````

````python
print(fruits)
````

````python
deuxieme = fruits.pop(1)
````

````python
print(deuxieme)
````

### `remove()` — retirer par valeur

Supprime la **première** occurrence.

````python
fruits.remove("kiwi")
````

````python
print(fruits)
````

### `clear()` — vider la liste

````python
copie = ["a", "b", "c"]
copie.clear()
````

````python
print(copie)
````

### `del` — supprimer par index ou par tranche

````python
nums = [10, 20, 30, 40, 50]
del nums[2]
````

````python
print(nums)
````

### `index()` — trouver la position

## 6. Trier une liste

### `sort()` — tri en place (modifie la liste)

`sort()` ne retourne rien (`None`), il modifie la liste directement.

````python
notes.sort()
````

````python
print(notes)
````

Tri décroissant avec `reverse=True`.

````python
notes.sort(reverse=True)
````

````python
print(notes)
````

### `sorted()` — tri qui crée une nouvelle liste

`sorted()` retourne une **nouvelle** liste triée, l'originale ne change pas.

````python
original = [3, 1, 4, 1, 5, 9]
trie = sorted(original)
````

````python
print(trie)
````

````python
print(original)  # inchangé
````

### `reverse()` — inverser en place

````python
original.reverse()
````

````python
print(original)
````

## 7. Itérer sur une liste

### Boucle `for` simple

````python
couleurs = ["rouge", "vert", "bleu"]
for c in couleurs:
    print(c)
````

### `enumerate()` — index + valeur

````python
for i, c in enumerate(couleurs):
    print(f"{i} : {c}")
````

### `zip()` — parcourir deux listes en parallèle

````python
prenoms = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]
for prenom, age in zip(prenoms, ages):
    print(f"{prenom} a {age} ans")
````

### Fonctions utiles sur les listes numériques

````python
valeurs = [4, 8, 2, 9, 1]
````

````python
sum(valeurs)
````

````python
min(valeurs)
````

````python
max(valeurs)
````

## 8. Compréhensions de liste

Une **compréhension de liste** (list comprehension) crée une liste en une seule expression.

Syntaxe : `[expression for element in iterable]`

````python
carres = [x ** 2 for x in range(6)]
````

````python
print(carres)
````

Avec une **condition de filtrage** : `[expression for element in iterable if condition]`

````python
pairs = [x for x in range(10) if x % 2 == 0]
````

````python
print(pairs)
````

On peut appliquer une transformation et un filtre en même temps.

````python
longueurs = [len(m) for m in ["chat", "chien", "oiseau", "rat"]]
print(longueurs)
````

### Compréhension avec f-string

````python
etiquettes = [f"item_{i}" for i in range(1, 6)]
print(etiquettes)
````

### Équivalence avec une boucle classique

Chaque compréhension peut se réécrire en boucle (et vice-versa).

````python
# Boucle classique
resultat = []
for x in range(6):
    resultat.append(x ** 2)
print(resultat)
````

## 9. Pièges courants

### Piège 1 : copie par référence

Affecter une liste à une autre variable ne crée **pas** de copie.

````python
a = [1, 2, 3]
b = a          # b pointe vers la MÊME liste
b.append(4)
````

````python
print(a)  # a est aussi modifié !
````

Pour faire une **vraie copie**, utiliser `.copy()` ou le slicing `[:]`.

````python
a = [1, 2, 3]
b = a.copy()   # copie indépendante
b.append(4)
````

````python
print(a)  # a n'a pas changé
````

````python
print(b)
````

### Piège 3 : modifier une liste en la parcourant

````python
# Ne faites pas ça :
# for x in ma_liste:
#     if condition:
#         ma_liste.remove(x)  # comportement imprévisible

# Utilisez une compréhension :
nombres = [1, 2, 3, 4, 5, 6]
nombres = [x for x in nombres if x % 2 != 0]
print(nombres)
````

## 10. Synthèse

| Opération | Syntaxe | Modifie la liste ? |
|---|---|---|
| Créer | `[1, 2, 3]` ou `list(...)` | — |
| Accéder | `liste[i]`, `liste[a:b]` | Non |
| Modifier | `liste[i] = val` | **Oui** |
| Ajouter | `append()`, `insert()`, `extend()` | **Oui** |
| Supprimer | `pop()`, `remove()`, `del` | **Oui** |
| Chercher | `in`, `index()`, `count()` | Non |
| Trier | `sort()` | **Oui** |
| Trier (copie) | `sorted()` | Non (nouvelle liste) |
| Compréhension | `[expr for x in iter]` | Non (nouvelle liste) |

### Règles à retenir

1. Les listes sont **mutables** : les méthodes comme `append`, `sort`, `remove` modifient la liste en place.
2. `sorted()` crée une **nouvelle** liste ; `sort()` modifie l'existante.
3. Attention aux **copies par référence** : `b = a` ne copie pas, `b = a.copy()` copie.
4. Les compréhensions de liste sont plus **lisibles** et **rapides** qu'une boucle + `append`.
