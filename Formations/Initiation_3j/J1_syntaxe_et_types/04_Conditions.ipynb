## 1. L'instruction `if`

La forme la plus simple : « **si** la condition est vraie, exécuter ce bloc ».

```python
if <condition>:
    <instructions>
```

Deux éléments essentiels :

1. Les **deux points** `:` à la fin de la ligne `if`.
2. L'**indentation** du bloc (4 espaces par convention).


````python
age = 20
if age >= 18:
    print("Vous êtes majeur.")
````

Si la condition est fausse, le bloc est **ignoré**, et Python continue après.

````python
age = 15
if age >= 18:
    print("Vous êtes majeur.")
print("Fin du programme.")
````

## 2. `if` / `else`

Pour exécuter **un bloc ou un autre** selon la condition :


````python
age = 15
if age >= 18:
    print("Majeur")
else:
    print("Mineur")
````

## 3. `if` / `elif` / `else`

Pour enchaîner plusieurs conditions **mutuellement exclusives** :


````python
note = 14
if note >= 16:
    print("Très bien")
elif note >= 12:
    print("Bien")
elif note >= 10:
    print("Passable")
else:
    print("Insuffisant")
````

Python évalue les conditions **dans l'ordre** et s'arrête à la première vraie. Les suivantes ne sont pas testées.

**Règle** : mettre les conditions **les plus restrictives en premier** (ici, `>= 16` avant `>= 12`).


## 4. Indentation obligatoire

En Python, l'**indentation n'est pas décorative** : elle délimite les blocs. Un bloc mal indenté est une **erreur**.


````python
x = 10
if x > 5:
    print("grand")
    print("vraiment grand")
print("toujours affiché")
````

Les deux `print` indentés appartiennent au bloc `if`. Le troisième, non indenté, est **en dehors**.

### Convention : 4 espaces, pas de tabulations

PEP 8 recommande **4 espaces** par niveau. JupyterLab et les IDE modernes insèrent 4 espaces quand on tape `Tab`. **Ne mélangez jamais tabulations et espaces** dans un même fichier : Python 3 lève une `TabError`.


## 5. Conditions imbriquées

Un `if` peut contenir un autre `if`. On indente alors de 8 espaces (2 × 4).


````python
age = 20
actif = True

if age >= 18:
    if actif:
        print("Majeur et actif")
    else:
        print("Majeur mais inactif")
else:
    print("Mineur")
````

### Préférer la combinaison logique quand c'est possible

L'exemple ci-dessus peut souvent s'écrire plus lisiblement avec `and` :


````python
age = 20
actif = True

if age >= 18 and actif:
    print("Majeur et actif")
elif age >= 18:
    print("Majeur mais inactif")
else:
    print("Mineur")
````

## 6. Combinaisons logiques dans un `if`

Les opérateurs `and`, `or`, `not` permettent de combiner plusieurs conditions dans un même `if`.


````python
jour = 6  # 0 = lundi, 6 = dimanche
if jour == 5 or jour == 6:
    print("Week-end")
else:
    print("Semaine")
````

````python
x = 50
if 0 <= x <= 100:
    print("Dans l'intervalle")
else:
    print("Hors intervalle")
````

````python
connecte = True
admin = False
if connecte and not admin:
    print("Utilisateur standard connecté")
````

## 7. `match` / `case` *(introduction)*

Depuis Python 3.10, une nouvelle syntaxe **`match`/`case`** permet d'écrire des tests de valeur plus lisibles qu'une chaîne de `elif`. C'est particulièrement utile quand on compare une variable à **plusieurs valeurs possibles**.


````python
commande = "help"

match commande:
    case "start":
        print("Démarrage...")
    case "stop":
        print("Arrêt...")
    case "help":
        print("Commandes : start, stop, help")
    case _:
        print("Commande inconnue :", commande)
````

Le cas `_` est le **cas par défaut** (équivalent de `else`).

`match`/`case` a beaucoup de fonctionnalités avancées (patterns, destructuring, `|` pour OR) qui seront vues dans le cours Intermédiaire. Pour l'instant, on l'utilise comme un `if`/`elif`/`else` plus clair.


On peut combiner plusieurs valeurs dans un même `case` avec `|` :

````python
jour = 6

match jour:
    case 0 | 1 | 2 | 3 | 4:
        print("Semaine")
    case 5 | 6:
        print("Week-end")
    case _:
        print("Jour invalide")
````

## 8. Synthèse

### Formes syntaxiques

```python
# Simple
if cond:
    ...

# Double branche
if cond:
    ...
else:
    ...

# Multiple branches
if cond1:
    ...
elif cond2:
    ...
elif cond3:
    ...
else:
    ...

# match / case
match valeur:
    case motif1:
        ...
    case motif2 | motif3:
        ...
    case _:
        ...
```

### Règles à retenir

1. **Les deux points `:`** à la fin de la ligne `if`, `elif`, `else`, `case`.
2. **Indentation de 4 espaces** pour chaque niveau de bloc.
3. **`elif` et non `else if`** — mot-clé unique.
4. **`elif` est évalué dans l'ordre**, on met les conditions restrictives en premier.
5. **Préférer `and`/`or` aux imbrications** quand c'est lisible.
6. **`match`/`case`** pour tester une variable contre plusieurs valeurs.
