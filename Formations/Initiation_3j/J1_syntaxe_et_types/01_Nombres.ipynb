### Égalité `==` et identité `is`

Ces deux opérateurs ne font pas la même chose :

- `==` compare les **valeurs**.
- `is` compare les **identités** (« est-ce le même objet en mémoire ? »).


````python
entier == flottant
````

Les valeurs sont considérées égales parce que `42 == 42.0` mathématiquement.

````python
entier is flottant
````

En revanche, ce ne sont pas les mêmes objets : l'un est un `int`, l'autre un `float`. Retenez : **on compare des valeurs avec `==`, pas avec `is`**.

### Comparaison avec `None`

`None` est une valeur spéciale qui représente « rien ». On la compare **toujours** avec `is`, jamais avec `==`.


````python
entier is None
````

````python
entier == None
````

Les deux marchent ici, mais `is None` est le style recommandé (plus rapide et plus sûr).

## 4. Opérations arithmétiques

Les opérateurs classiques fonctionnent comme vous l'attendez — avec une subtilité importante sur la division.


````python
4 + 2
````

````python
4 - 2
````

````python
5 * 2
````

````python
5 / 2
````

⚠ **Attention** : l'opérateur `/` retourne **toujours** un `float`, même quand les deux opérandes sont entiers et que le résultat est exact (essayez `6 / 3`). Pour obtenir un entier, il faut `//`.

````python
5 // 2
````

````python
5 % 2
````

`//` est la **division entière** (on garde le quotient), `%` est le **modulo** (on garde le reste).

````python
5. / 2.
````

````python
5. // 2.
````

`//` entre deux `float` garde la sémantique « partie entière basse » mais retourne un `float`.

````python
4 ** 2
````

`**` est l'opérateur de puissance (`4²` = 16).

### Séparateurs numériques

## 6. Mélange `int` / `float` et promotion de type

Que se passe-t-il quand on mélange un `int` et un `float` dans une opération ? Python « promeut » automatiquement l'entier en flottant.


````python
entier + flottant
````

````python
type(entier + flottant)
````

````python
entier / flottant
````

````python
print(99 / 100)
````

---

### Ce qu'il est important de comprendre

Dans la plupart des langages à bas niveau (C, C++, Java, Go...), l'opération est programmée **au niveau du type** :

- `int + int` donne un `int` ;
- `float + float` donne un `float` ;
- `int / int` donne un `int` avec **troncature** (en C, `5 / 2` vaut `2`, pas `2.5`).

Python, lui, fait une analyse de types au moment de l'exécution et choisit automatiquement la bonne opération. C'est un confort énorme, mais il faut garder à l'esprit que :

- `/` retourne **toujours** un `float`, même entre deux `int` ;
- `//` retourne un `int` si les deux opérandes sont des `int`, un `float` sinon ;
- Le mélange `int + float` produit toujours un `float`.

**En pratique** : pour faire de l'arithmétique exacte sur des entiers, restez entre entiers et utilisez `//` plutôt que `/`.

## 7. Un piège à connaître : IEEE 754

Les `float` de Python sont implémentés selon la norme internationale **IEEE 754 double précision** (64 bits). Ils peuvent représenter approximativement des nombres de `1,8 × 10³⁰⁸` à `2,2 × 10⁻³⁰⁸`. Mais ils ne les représentent **pas toujours exactement**.

Voici le piège le plus célèbre en informatique :


````python
0.1 + 0.2
````

````python
0.1 + 0.2 == 0.3
````

Surpris ? Ce n'est pas un bug de Python — c'est une limite fondamentale du format binaire. `0.1`, `0.2` et `0.3` ne peuvent **pas** être représentés exactement en binaire, de la même façon que `1/3` ne peut pas être représenté exactement en décimal (`0,333...`).

L'erreur d'arrondi est minuscule (`~10⁻¹⁷`) mais elle existe. Elle s'accumule quand on fait beaucoup d'opérations, ce qui peut poser problème en finance ou en calcul scientifique précis.

> **Règle à retenir** : on ne compare **jamais** deux flottants avec `==`. On utilise `math.isclose(a, b)`.


````python
import math
````

````python
math.isclose(0.1 + 0.2, 0.3)
````

Pour voir le piège dans tous les langages de programmation, une page de référence : [0.30000000000000004.com](https://0.30000000000000004.com/).

### Arrondir : attention aux pièges

Plusieurs fonctions arrondissent, et elles ne font pas la même chose.


**`round(x)`** utilise l'« arrondi bancaire » : pour `.5`, on arrondit vers le **pair** le plus proche.

````python
round(2.5)  # 2 !! (pas 3)
````

````python
round(3.5)  # 4
````

````python
round(0.5)  # 0 !!
````

C'est déroutant, mais Python suit la norme IEEE 754 (*round half to even*) qui est statistiquement plus juste que l'arrondi scolaire. En pratique, pour des grandes séries de valeurs, cette règle évite les biais.

**`math.floor(x)`** : arrondi vers `-∞` (vers le bas).

````python
math.floor(2.7)
````

````python
math.floor(-2.3)
````

**`math.ceil(x)`** : arrondi vers `+∞` (vers le haut).

````python
math.ceil(2.3)
````

````python
math.ceil(-2.7)
````

**`math.trunc(x)`** : arrondi vers `0`.

````python
math.trunc(2.7)
````

````python
math.trunc(-2.7)
````

## 11. Synthèse

| Type | Utilisation | Exemple |
|---|---|---|
| `int` | Compteurs, indices, quantités entières | `42`, `1_000_000`, `2 ** 100` |
| `float` | Mesures, calculs scientifiques | `3.14`, `1.5e10`, `0.0` |
| `complex` | Physique, signal *(rarement)* | `1 + 2j` |

### Règles à retenir

1. **`/` retourne toujours un `float`**, même entre deux entiers. Pour diviser en entier, utiliser `//`.
2. **Jamais `==` entre flottants** — utiliser `math.isclose(a, b)`.
3. **`round()` fait du *banker's rounding*** (arrondi au pair le plus proche pour `.5`), pas l'arrondi scolaire.
4. **Les entiers Python ont une précision illimitée** — calculer `100!` ou `2 ** 10000` ne pose aucun problème.
5. Le **mélange `int + float` promeut vers `float`**. Pour rester en entier, rester entre entiers et utiliser `//`.
6. **`NaN` n'est jamais égal à lui-même** — utiliser `math.isnan()`.

- `cmath` : équivalent pour les complexes *(mentionné)*
