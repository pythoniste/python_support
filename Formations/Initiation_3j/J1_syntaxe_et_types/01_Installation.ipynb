## 1. Pourquoi Python 3.14 ?

Python 3.14, sorti en octobre 2025, est la **version stable actuelle**. Elle apporte :

- des **messages d'erreur** encore plus précis ;
- des améliorations de performance ;
- de nouvelles fonctionnalités du langage (PEP 649, PEP 750, etc.) ;
- le support d'une version **free-threaded** expérimentale *(PEP 703, sujet avancé)*.

**Règle** : sauf contrainte client, on se forme toujours sur la dernière version mineure stable.


## 2. Installer Python

Plusieurs approches selon votre système. **Une seule suffit.**

### Option A — Installeur officiel (le plus simple)

Télécharger le paquet adapté à votre OS depuis [python.org/downloads](https://www.python.org/downloads/).

- **Windows** : cocher « Add python.exe to PATH » avant de cliquer sur *Install Now*.
- **macOS** : lancer le `.pkg` téléchargé.
- **Linux** : utiliser votre gestionnaire de paquets (`apt`, `dnf`, `pacman`, ...) **ou** l'option B ci-dessous si votre distribution n'a pas encore Python 3.14.

### Option B — Via `uv` *(recommandé dès qu'on connaît `uv`)*

On verra `uv` en section 5. Une fois `uv` installé, un simple :

```
uv python install 3.14
```

suffit à télécharger et installer Python 3.14, **sans toucher au Python système**.


## 3. Vérifier l'installation

Ouvrir un terminal neuf et taper :

```
python --version
```

Vous devez voir quelque chose comme :

```
Python 3.14.0
```

Si vous voyez `Python 2.7` ou `command not found`, c'est que :

- Python n'est pas dans votre `PATH` *(Windows)* ;
- votre système a `python3` et pas `python` : essayer `python3 --version`.

### Commandes à connaître

| Commande | Effet |
|---|---|
| `python --version` | Affiche la version installée |
| `python -V` | Identique, version courte |
| `which python` *(Linux/macOS)* | Affiche le chemin de l'exécutable |
| `where python` *(Windows)* | Idem |


## 4. Pourquoi un environnement virtuel ?

Imaginez :

- Votre **projet A** a besoin de la version 2.0 d'une bibliothèque.
- Votre **projet B** a besoin de la version 3.0 de la même bibliothèque.
- Installer les deux dans le Python système : **conflit**.

Un **environnement virtuel** est un dossier qui contient :

- une copie isolée de Python ;
- ses propres paquets installés ;
- ses propres scripts console.

Un projet = un environnement. On **ne pollue jamais** le Python système.


## 5. Installer `uv`

`uv` est un **gestionnaire d'environnements et de paquets** écrit en Rust, 10 à 100 fois plus rapide que `pip` et capable d'installer Python lui-même.

### Installation

**Linux / macOS :**

```
curl -LsSf https://astral.sh/uv/install.sh | sh
```

**Windows (PowerShell) :**

```
powershell -c "irm https://astral.sh/uv/install.ps1 | iex"
```

**Vérification :**

```
uv --version
```

Doit afficher `uv 0.x.y` *(ou plus récent)*.


## 6. Créer un environnement avec `uv`

Se placer dans un dossier de travail, puis :

```
uv venv --python 3.14
```

`uv` :

1. télécharge Python 3.14 s'il n'est pas déjà présent ;
2. crée un dossier `.venv/` dans le répertoire courant ;
3. prêt à l'emploi.

### Arborescence typique

```
mon_projet/
├── .venv/                 # environnement virtuel, ne JAMAIS versionner
│   ├── bin/               # Linux/macOS
│   ├── Scripts/           # Windows
│   └── ...
├── hello.py               # votre premier script
└── ...
```

**Astuce** : ajouter `.venv/` à votre `.gitignore` pour ne pas le versionner par erreur.


## 7. Activer l'environnement

Avant de lancer `python` ou d'installer des paquets, il faut **activer** l'environnement virtuel. Cela modifie temporairement votre `PATH` pour que `python` pointe vers celui du `.venv/`.

**Linux / macOS :**

```
source .venv/bin/activate
```

**Windows (PowerShell) :**

```
.\.venv\Scripts\Activate.ps1
```

Votre invite change et montre `(.venv)` devant : c'est le signe que l'environnement est actif.

### Désactivation

```
deactivate
```

### Alternative sans activation

`uv` permet d'exécuter des commandes dans l'environnement **sans activer** :

```
uv run python hello.py
```

C'est souvent plus pratique quand on alterne entre projets.


## 8. Installer des paquets

Dans un environnement activé :

```
uv pip install requests
```

Ou, en style projet `uv` moderne :

```
uv add requests
```

Ce dernier met à jour votre `pyproject.toml` (vu en J5) et installe le paquet dans le `.venv/`.

### Commandes utiles

| Commande | Effet |
|---|---|
| `uv pip install X` | Installe le paquet X |
| `uv pip uninstall X` | Désinstalle X |
| `uv pip list` | Liste les paquets installés |
| `uv pip freeze` | Affiche la liste au format `requirements.txt` |


## 9. Premier script `hello.py`

Créer un fichier `hello.py` contenant **exactement** :

```python
print("Bonjour, Python !")
```

L'exécuter depuis le terminal :

```
python hello.py
```

*(ou `uv run python hello.py` si l'environnement n'est pas activé)*

Vous devez voir :

```
Bonjour, Python !
```

**Félicitations**, vous avez exécuté votre premier programme Python. Tout ce qu'on va voir dans la formation part de ce moment.


### Comprendre ce qui vient de se passer

1. Vous avez tapé du texte dans un fichier.
2. Vous avez demandé à `python` d'exécuter ce fichier.
3. Python a **lu** votre fichier ligne par ligne et **exécuté** chaque instruction.
4. La seule instruction `print(...)` a écrit du texte sur la sortie standard, que votre terminal affiche.

C'est exactement la même boucle **lire → exécuter → afficher** qu'on aura pendant toute la formation. Ce qui va changer, c'est la richesse de ce qu'on pourra écrire *avant* le `print`.

## 10. Synthèse

### Les étapes d'une installation propre

| Étape | Commande | But |
|---|---|---|
| 1 | `curl ... \| sh` | Installer `uv` |
| 2 | `uv python install 3.14` | Obtenir Python 3.14 |
| 3 | `uv venv --python 3.14` | Créer un environnement virtuel local |
| 4 | `source .venv/bin/activate` | Activer (ou `uv run ...`) |
| 5 | `uv pip install ...` | Installer des paquets |
| 6 | `python monscript.py` | Exécuter |

### Règles à retenir

1. **Jamais dans le Python système** — toujours dans un `.venv/`.
2. **Un projet = un environnement.**
3. **`uv` gère Python ET les paquets** — pas besoin de `pip` séparé.
4. **`.venv/` ne se versionne pas** — ajouter à `.gitignore`.
5. **`--python 3.14`** à la création du venv garantit la bonne version, même si le système en a une plus ancienne.
