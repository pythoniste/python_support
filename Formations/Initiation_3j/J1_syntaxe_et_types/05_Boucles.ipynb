## 1. La fonction `range`

`range` est un objet qui **représente une séquence d'entiers**. Il est conçu pour être parcouru par une boucle `for`.

Trois formes :

| Forme | Équivalent |
|---|---|
| `range(n)` | `0, 1, 2, ..., n-1` |
| `range(a, b)` | `a, a+1, ..., b-1` |
| `range(a, b, pas)` | `a, a+pas, a+2*pas, ...` tant que `< b` |

**Important** : la borne de fin est **exclue**.


````python
list(range(5))
````

*Remarque : on convertit ici en `list` uniquement pour **visualiser** le contenu. On verra `list` plus tard. `range` en lui-même n'est pas une liste.*

````python
list(range(2, 7))
````

````python
list(range(0, 20, 3))
````

````python
list(range(10, 0, -1))
````

## 2. La boucle `for`

La boucle `for` itère sur un objet qui contient plusieurs éléments. Avec `range`, on obtient le classique **« faire quelque chose N fois »**.

```python
for <variable> in <séquence>:
    <bloc>
```


````python
for i in range(5):
    print(i)
````

À chaque tour, `i` prend la valeur suivante de la séquence.

````python
for i in range(1, 6):
    print("Itération :", i)
````

On peut utiliser `i` dans un calcul :

````python
for i in range(1, 11):
    print(i, "×", 7, "=", i * 7)
````

**Convention** : si on n'a pas besoin de la variable de boucle, utiliser `_` :


````python
for _ in range(3):
    print("Bonjour")
````

## 3. Itérer avec un index

Pour l'instant on ne connaît que `range`, donc l'« index » et la « valeur » sont la même chose. Quand on verra les listes, on verra aussi `enumerate(...)`. En attendant, `range(N)` suffit.


````python
somme = 0
for i in range(1, 11):
    somme += i
print("Somme 1..10 :", somme)
````

## 4. La boucle `while`

La boucle `while` **répète tant qu'une condition reste vraie**.

```python
while <condition>:
    <bloc>
```

**Attention** : il faut s'assurer que la condition devient fausse un jour, sinon la boucle est infinie.


````python
n = 0
while n < 5:
    print("n =", n)
    n += 1
````

## 5. `break` — sortir d'une boucle

`break` **interrompt immédiatement** la boucle, quelle qu'elle soit (`for` ou `while`).


````python
for i in range(10):
    if i == 5:
        break
    print(i)
````

Ici, la boucle s'arrête à `i = 5`. Les valeurs affichées sont 0, 1, 2, 3, 4.

**Chercher et s'arrêter** : trouver le premier multiple de 7 entre 20 et 100.

````python
for i in range(20, 101):
    if i % 7 == 0:
        print("Trouvé :", i)
        break
````

## 6. `continue` — passer à l'itération suivante

`continue` **saute le reste du bloc** et passe à l'itération suivante.


````python
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
````

Ici, on saute les nombres pairs, donc on n'affiche que les impairs : 1, 3, 5, 7, 9.

## 7. Clause `else` de boucle

Particularité pythonique : un bloc `for` ou `while` peut avoir une clause **`else`** qui s'exécute **si la boucle se termine normalement** (sans `break`).


````python
for i in range(5):
    print(i)
else:
    print("Boucle terminée normalement")
````

Si on ajoute un `break`, l'`else` **n'est pas exécuté** :

````python
for i in range(5):
    if i == 3:
        break
    print(i)
else:
    print("Cette ligne ne s'affiche PAS")
print("Fin")
````

### Cas d'usage : recherche

`else` de boucle permet d'écrire élégamment « j'ai cherché et je n'ai rien trouvé » :


````python
cible = 17
for i in range(2, 20):
    if i == cible:
        print("Trouvé à la position", i)
        break
else:
    print("Pas trouvé")
````

## 8. Boucles imbriquées

Une boucle peut contenir une autre boucle. Chaque boucle a son propre indentation (4 espaces supplémentaires).


````python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, "×", j, "=", i * j)
    print("---")
````

**Attention** : `break` n'interrompt que la boucle **la plus proche**. Il ne sort pas des deux à la fois.

## 9. Synthèse

### Formes des boucles

```python
# for avec range
for i in range(N):
    ...

# while
while condition:
    ...

# for / else
for i in range(N):
    ...
    if trouve:
        break
else:
    print("non trouvé")
```

### Règles à retenir

1. **`range(n)` va de 0 à n-1** — borne supérieure exclue.
2. **`for` pour un nombre connu d'itérations**, **`while`** pour un nombre inconnu dépendant d'une condition.
3. **`break` arrête** la boucle, **`continue` passe** à l'itération suivante.
4. **`else` de boucle** s'exécute si **aucun `break`** n'a eu lieu.
5. **Attention aux boucles infinies** avec `while` — toujours vérifier que la condition peut devenir fausse.
