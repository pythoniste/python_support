## 1. Créer une chaîne

Trois délimiteurs équivalents :

| Forme | Quand l'utiliser |
|---|---|
| `"texte"` | Par défaut |
| `'texte'` | Si le contenu contient des `"` |
| `"""texte"""` | Pour du texte multi-lignes |

`"""..."""` est aussi le format des **docstrings** (documentation de fonction, vu plus tard).


````python
s = "bonjour"
print(s)
````

````python
s = 'hello'
print(s)
````

````python
s = "Elle a dit \"bonjour\"."
print(s)
````

````python
s = 'Elle a dit "bonjour".'
print(s)
````

## 2. Chaînes sur plusieurs lignes

Les triples guillemets conservent les retours à la ligne à l'intérieur :


````python
texte = """Première ligne
Deuxième ligne
Troisième ligne"""
print(texte)
````

On peut aussi utiliser `\n` dans une chaîne simple :

````python
print("L1\nL2\nL3")
````

## 3. Concaténation et répétition

- `+` concatène deux chaînes.
- `*` répète une chaîne.


````python
print("abc" + "def")
````

````python
print("ha" * 5)
````

````python
print("-" * 20)
````

**Attention** : `+` ne marche qu'entre deux chaînes. Pour insérer un nombre, il faut le convertir ou — mieux — utiliser une **f-string** (section 10).

## 4. Indexation et longueur

Une chaîne est **indexable** : chaque caractère a une position. Les index commencent à **0**.


````python
s = "Python"
print(s[0])
````

````python
s = "Python"
print(s[1])
````

````python
s = "Python"
print(s[5])
````

Les **index négatifs** comptent depuis la fin. `-1` est le dernier caractère.

````python
s = "Python"
print(s[-1])
````

````python
s = "Python"
print(s[-2])
````

La longueur s'obtient avec `len(...)` :

````python
s = "Python"
print(len(s))
````

Accéder à un index inexistant lève `IndexError` :

````python
s = "abc"
# s[10]  # IndexError — décommenter pour tester
````

### Immutabilité

Une chaîne est **immuable** : on ne peut pas modifier un caractère en place.


````python
s = "python"
# s[0] = "P"  # TypeError — décommenter pour tester
````

Pour « modifier » une chaîne, on en crée une nouvelle :

````python
s = "python"
s = "P" + s[1:]
print(s)
````

## 5. Slicing

La syntaxe `s[debut:fin]` renvoie une **sous-chaîne**. `debut` est inclus, `fin` est **exclu**.


````python
s = "abcdefgh"
print(s[0:3])
````

````python
s = "abcdefgh"
print(s[2:6])
````

````python
s = "abcdefgh"
print(s[:4])
````

````python
s = "abcdefgh"
print(s[4:])
````

````python
s = "abcdefgh"
print(s[-3:])
````

````python
s = "abcdefgh"
print(s[::2])
````

````python
s = "abcdefgh"
print(s[::-1])
````

La dernière forme `[::-1]` est l'idiome pour **inverser** une chaîne : le `pas -1` parcourt de la fin vers le début.

## 6. Itérer sur une chaîne

Une `for` peut parcourir directement les caractères d'une chaîne :


````python
for c in "Python":
    print(c)
````

## 7. Méthodes de transformation

Les méthodes sont appelées avec la notation `s.methode(args)`. Elles **renvoient une nouvelle chaîne** — jamais ne modifient l'originale *(puisque les chaînes sont immuables)*.


````python
s = "Bonjour"
print(s.upper())
````

````python
s = "Bonjour"
print(s.lower())
````

````python
s = "alice"
print(s.capitalize())
````

````python
s = "alice dupont"
print(s.title())
````

**Nettoyer les espaces autour** :

````python
s = "   texte   "
print(repr(s.strip()))
````

*`repr(...)` affiche les guillemets autour de la chaîne, pratique pour voir les bornes.*

````python
s = "---abc---"
print(s.strip("-"))
````

**Remplacer une sous-chaîne** :

````python
s = "bonjour le monde"
print(s.replace("monde", "Python"))
````

## 8. Méthodes de recherche

Elles renvoient soit un booléen, soit un index.


````python
s = "bonjour"
print(s.startswith("bon"))
````

````python
s = "photo.jpg"
print(s.endswith(".jpg"))
````

````python
s = "abracadabra"
print(s.find("cad"))
````

````python
s = "abracadabra"
print(s.find("xyz"))
````

`find` renvoie l'index de la première occurrence, ou `-1` si absent.

````python
s = "abracadabra"
print(s.count("a"))
````

````python
s = "42"
print(s.isdigit())
````

````python
s = "abc"
print(s.isalpha())
````

### L'opérateur `in`

Pour tester la présence d'une sous-chaîne, on peut utiliser directement `in` :


````python
print("cad" in "abracadabra")
````

````python
print("xyz" in "abracadabra")
````

## 9. Méthodes `split` et `join`

`split(separateur)` **découpe** une chaîne en plusieurs morceaux. Le résultat est une **liste** (qu'on verra en détail bientôt — ici, on se contente de l'itérer).


````python
s = "pomme,poire,banane"
print(s.split(","))
````

````python
s = "un deux trois"
print(s.split())
````

Sans argument, `split()` découpe sur n'importe quelle suite d'espaces.

`join(morceaux)` est l'**inverse** : colle plusieurs morceaux en une chaîne.

````python
morceaux = ["a", "b", "c"]
print("-".join(morceaux))
````

## 10. **f-strings** : le format moderne

Depuis Python 3.6, les **f-strings** (format strings) sont la façon recommandée d'insérer des valeurs dans une chaîne. La syntaxe est `f"...{expression}..."` — le préfixe `f` transforme les parties entre accolades en **expressions Python évaluées**.


````python
nom = "Alice"
age = 30
print(f"{nom} a {age} ans.")
````

````python
x = 2
y = 3
print(f"{x} + {y} = {x + y}")
````

L'expression entre `{...}` peut être **n'importe quelle expression Python** : appel de fonction, calcul, attribut, etc.

````python
import math
r = 5
print(f"Aire du cercle : {math.pi * r ** 2}")
````

### Debug : `{x=}`

Depuis Python 3.8, la syntaxe `f"{x=}"` affiche à la fois le **nom** et la **valeur** de `x`. Très pratique pour déboguer.


````python
x = 42
print(f"{x=}")
````

````python
prenom = "Alice"
age = 30
print(f"{prenom=} {age=}")
````

## 11. Options de formatage (largeur, alignement, décimales)

Entre les accolades, on peut ajouter un **format spec** après `:` pour contrôler l'affichage.


**Décimales** :

````python
pi = 3.141592653589793
print(f"{pi:.2f}")
print(f"{pi:.5f}")
````

**Largeur et alignement** :

````python
mot = "abc"
print(f"[{mot:>10}]")
print(f"[{mot:<10}]")
print(f"[{mot:^10}]")
````

**Largeur et remplissage** :

````python
n = 42
print(f"{n:05}")
````

**Séparateur de milliers** :

````python
n = 1000000
print(f'{n:,}')
````

**Pourcentage** :

````python
taux = 0.153
print(f'{taux:.1%}')
````

**Tableau d'alignement utile** :

````python
for i in range(1, 6):
    carre = i ** 2
    cube = i ** 3
    print(f"{i:>3} | {carre:>5} | {cube:>5}")
````

## 12. Raw strings

Une **raw string** est préfixée par `r` : les antislashes ne sont **pas** interprétés. Très pratique pour les chemins Windows, les expressions régulières, etc.


````python
print(r"C:\Users\alice")
````

````python
print(r"ligne 1\nligne 2")
````

Sans le `r`, `\n` serait interprété comme un retour à la ligne.

## 13. Synthèse

### Méthodes essentielles

| Méthode | Rôle |
|---|---|
| `upper()` / `lower()` | Changer la casse |
| `strip()` | Nettoyer les espaces de début/fin |
| `replace(a, b)` | Remplacer `a` par `b` |
| `split(sep)` | Découper en liste |
| `"sep".join(parts)` | Coller des morceaux |
| `startswith(p)` / `endswith(p)` | Tester début / fin |
| `find(sub)` | Position d'une sous-chaîne ou -1 |
| `count(sub)` | Nombre d'occurrences |
| `in` | Tester la présence |

### Règles à retenir

1. **Les chaînes sont immuables** : toutes les méthodes renvoient une **nouvelle** chaîne.
2. **Slicing `[a:b]` : `a` inclus, `b` exclu.**
3. **`[::-1]`** inverse une chaîne.
4. **`len(s)`** donne la longueur.
5. **f-strings** `f"..."` pour insérer des valeurs : `f"{x + y}"`.
6. **`f"{x=}"`** pour afficher nom + valeur — débug rapide.
7. **Format `:.2f`** pour 2 décimales, `:>10` pour alignement à droite sur 10 caractères.
8. **`r"..."`** pour les chemins Windows et les regex.
