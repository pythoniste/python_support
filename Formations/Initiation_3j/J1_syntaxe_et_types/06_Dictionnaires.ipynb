## 2. Lire une valeur : `d[k]` vs `d.get(k)`

Deux manières d'accéder à une valeur, avec des comportements différents si la clé est absente.

````python
contact = {'nom': 'Dupont', 'age': 30}
````

````python
contact['nom']
````

````python
# Clé absente : erreur
try:
    print(contact['ville'])
except KeyError as err:
    print('clé manquante :', err)
````

`.get` renvoie `None` (ou une valeur par défaut) si la clé est absente, sans lever d'erreur.

````python
contact.get('ville')
````

````python
print(contact.get('ville'))
````

````python
contact.get('ville', 'inconnue')
````

### Quelle méthode choisir ?

- `d[k]` : quand la clé **doit** être présente — une erreur indique un bug.
- `d.get(k)` : quand la clé est **optionnelle** — on gère le cas absent proprement.

## 3. Ajouter, modifier, supprimer

````python
contact = {'nom': 'Dupont'}
contact['age'] = 30
print(contact)
````

La même syntaxe ajoute une clé absente ou met à jour une clé existante.

````python
contact['age'] = 31
print(contact)
````

### `del` supprime une clé

````python
del contact['age']
print(contact)
````

### `.pop(k)` supprime ET renvoie la valeur

````python
contact = {'nom': 'Dupont', 'age': 30}
age = contact.pop('age')
print('age retiré :', age)
print('dict restant :', contact)
````

### `.setdefault(k, v)` — lire ou créer

````python
contact = {'nom': 'Dupont'}
contact.setdefault('age', 0)
print(contact)
````

Si la clé existe, `setdefault` renvoie sa valeur sans la changer.

## 4. Itération

````python
notes = {'Alice': 15, 'Bob': 12, 'Chloé': 18}
````

### Par défaut : itération sur les clés

````python
for nom in notes:
    print(nom)
````

### `.keys()`, `.values()`, `.items()`

````python
for nom in notes.keys():
    print(nom)
````

````python
for note in notes.values():
    print(note)
````

````python
for nom, note in notes.items():
    print(f'{nom} : {note}')
````

Ici on **déballe** directement le tuple `(clé, valeur)` renvoyé par `.items()`.

## 6. Fusion avec `|` (Python 3.9+)

L'opérateur `|` crée un **nouveau** dictionnaire qui fusionne deux dicts. Les clés de droite l'emportent en cas de conflit.

````python
base = {'a': 1, 'b': 2}
extra = {'b': 20, 'c': 3}
````

````python
fusion = base | extra
print(fusion)
````

`base` et `extra` ne sont pas modifiés.

````python
print(base)
print(extra)
````

### `|=` met à jour en place

````python
base |= extra
print(base)
````

## 7. Compréhensions de dict

Comme pour les listes, on peut construire un dictionnaire en une expression.

````python
carres = {n: n * n for n in range(1, 6)}
print(carres)
````

````python
# Inverser un dictionnaire
notes = {'Alice': 15, 'Bob': 12, 'Chloé': 18}
par_note = {note: nom for nom, note in notes.items()}
print(par_note)
````

Attention aux doublons de valeurs : si deux clés ont la même valeur, l'inversion ne conserve que la dernière.

````python
notes_bis = {'Alice': 15, 'Bob': 12, 'David': 15}
{note: nom for nom, note in notes_bis.items()}
````

### Filtrer avec un `if`

````python
notes = {'Alice': 15, 'Bob': 8, 'Chloé': 18}
reussites = {nom: note for nom, note in notes.items() if note >= 10}
print(reussites)
````

## 8. Les clés doivent être **hashables**

Une clé de dictionnaire doit être **immuable** (hashable) : `int`, `float`, `str`, `bool`, `tuple` de hashables, `None`…
Une `list` ne peut **pas** être utilisée comme clé.

````python
d = {(0, 0): 'origine', (1, 0): 'droite'}
print(d)
````

````python
try:
    d = {[0, 0]: 'origine'}
except TypeError as err:
    print('erreur :', err)
````

Les valeurs, en revanche, peuvent être de n'importe quel type.

````python
d = {'fruits': ['pomme', 'poire'], 'legumes': ['carotte']}
print(d)
````

## 9. Synthèse

| Opération | Syntaxe |
|---|---|
| Création | `{'a': 1, 'b': 2}` |
| Lecture stricte | `d['a']` (erreur si absent) |
| Lecture tolérante | `d.get('a')` ou `d.get('a', défaut)` |
| Ajout / mise à jour | `d['c'] = 3` |
| Suppression | `del d['a']` ou `d.pop('a')` |
| Itération clés | `for k in d:` |
| Itération valeurs | `for v in d.values():` |
| Itération couples | `for k, v in d.items():` |
| Fusion | `a \| b`, `a \|= b` |
| Compréhension | `{k: v for ... in ... if ...}` |

### Règles à retenir

1. Un dictionnaire conserve l'**ordre d'insertion** depuis Python 3.7.
2. Les clés doivent être **hashables** — immuables en pratique.
3. Préférez `d.get(k, défaut)` à `if k in d` suivi de `d[k]`.
4. `.items()` + déballage est la manière idiomatique d'itérer.
