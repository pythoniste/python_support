## 1. Créer un tuple

Un `tuple` est une séquence ordonnée et **immuable** de valeurs.

````python
point = (3, 4)
````

````python
print(point)
````

````python
type(point)
````

Les parenthèses ne sont en fait **pas obligatoires** : ce sont les virgules qui créent le tuple.

````python
point2 = 3, 4
````

````python
print(point2)
````

````python
type(point2)
````

### Tuple vide et tuple à un élément

````python
vide = ()
````

````python
print(vide, type(vide))
````

Pour un tuple d'un seul élément, la virgule est **obligatoire**, sinon Python voit une expression parenthésée.

````python
pas_un_tuple = (42)
print(type(pas_un_tuple))  # int
````

````python
un_tuple = (42,)
print(type(un_tuple))     # tuple
````

## 3. Immutabilité

Contrairement à une `list`, un `tuple` **ne peut pas être modifié** après création.

````python
point = (3, 4)
try:
    point[0] = 99
except TypeError as err:
    print('Erreur :', err)
````

On ne peut ni `.append`, ni `.pop`, ni réaffecter un indice.

### Pourquoi l'immutabilité est utile

- un tuple est **hashable** : on peut l'utiliser comme clé de dictionnaire ou élément d'un `set` (chapitres suivants) ;
- il signale au lecteur : *ces valeurs ne doivent pas changer* (ex. des coordonnées, une version `(3, 14, 0)`) ;
- il est légèrement plus petit et plus rapide à construire qu'une liste.

````python
import sys
sys.getsizeof((1, 2, 3)), sys.getsizeof([1, 2, 3])
````

## 4. Déballage (unpacking)

Le **déballage** permet d'assigner les éléments d'un tuple à plusieurs variables en une seule ligne.

````python
point = (3, 4)
x, y = point
print('x =', x, ' y =', y)
````

Le déballage marche aussi à la volée, sans passer par une variable intermédiaire.

````python
x, y = 3, 4
print(x, y)
````

### Swap élégant

````python
a = 1
b = 2
a, b = b, a
print('a =', a, ' b =', b)
````

### Déballage avec `*` (rest)

````python
premier, *reste = (1, 2, 3, 4, 5)
print(premier)
print(reste)
````

````python
debut, *milieu, fin = (1, 2, 3, 4, 5)
print(debut, milieu, fin)
````

`*milieu` capture une **liste** de ce qui reste entre `debut` et `fin`.

## 5. Tuples et retour multiple

Même si `def` n'est vu qu'au chapitre suivant, sachez que de nombreuses primitives Python renvoient déjà des tuples. C'est le moyen idiomatique de renvoyer **plusieurs valeurs**.

````python
quotient, reste = divmod(17, 5)
print('q =', quotient, ' r =', reste)
````

`divmod` renvoie un tuple `(quotient, reste)` que l'on déballe directement.

````python
chaine = 'a=1,b=2,c=3'
parts = chaine.split(',')
print(parts)
````

Ici `split` renvoie une liste — mais on peut déballer une liste exactement comme un tuple.

````python
a, b, c = parts
print(a, b, c)
````

## 9. Quand choisir `tuple` plutôt que `list`

| Situation | Choix recommandé |
|---|---|
| Collection homogène susceptible d'évoluer | `list` |
| Enregistrement figé (coordonnée, version, couple) | `tuple` |
| Clé de dictionnaire ou élément de `set` | `tuple` (immuable, hashable) |
| Valeurs de retour multiples | `tuple` (avec déballage) |

**Règle empirique :** si vous voulez écrire `.append`, utilisez une liste. Sinon, préférez un tuple : le lecteur comprend immédiatement que ces valeurs forment un **bloc figé**.

---


## 10. Synthèse

| Opération | Liste | Tuple |
|---|---|---|
| Création | `[1, 2, 3]` | `(1, 2, 3)` ou `1, 2, 3` |
| Indexation | `l[0]` | `t[0]` |
| Longueur | `len(l)` | `len(t)` |
| Mutable ? | ✅ | ❌ |
| Hashable ? | ❌ | ✅ (si éléments hashables) |
| Méthode `.append` | ✅ | ❌ |
| Déballage `a, b = x` | ✅ | ✅ |

### Règles à retenir

1. Ce sont les **virgules** qui créent un tuple, pas les parenthèses.
2. Un tuple d'un élément s'écrit `(42,)` — la virgule est obligatoire.
3. Un tuple est **immuable** : aucune méthode ne le modifie.
4. Le **déballage** `x, y = point` est l'idiome Python par excellence.
5. `divmod`, `enumerate`, `zip` renvoient des tuples — retenez le déballage.
