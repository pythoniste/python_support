## 1. JSON — rappel

**JavaScript Object Notation**. Format texte universel pour échanger des données structurées. Il gère nativement : nombres, chaînes, booléens, `null`, tableaux, objets.

---


## 2. Les 4 fonctions du module `json`

| Fonction | Entrée → sortie |
|---|---|
| `json.dumps` | objet Python → chaîne JSON |
| `json.loads` | chaîne JSON → objet Python |
| `json.dump` | objet Python → fichier JSON |
| `json.load` | fichier JSON → objet Python |

Les formes **sans `s`** travaillent avec des **fichiers**. Les formes **avec `s`** travaillent avec des **chaînes**.

````python
import json
````

### `dumps` / `loads`

````python
donnees = {'nom': 'Alice', 'age': 30, 'actif': True}
chaine = json.dumps(donnees)
print(chaine)
print(type(chaine))
````

````python
json.loads(chaine)
````

---


## 3. `json.dump` vers un fichier

````python
import json
from pathlib import Path

donnees = {'nom': 'Alice', 'age': 30, 'langues': ['fr', 'en']}

with open('/tmp/alice.json', 'w', encoding='utf-8') as f:
    json.dump(donnees, f)

print(Path('/tmp/alice.json').read_text(encoding='utf-8'))
````

---


## 4. `json.load` depuis un fichier

````python
with open('/tmp/alice.json', 'r', encoding='utf-8') as f:
    donnees = json.load(f)
print(donnees)
print(type(donnees))
````

---


## 5. Contrôle du formatage

### `indent` — JSON pretty-printed

Par défaut, `json.dump` produit du JSON compact. Passer `indent=2` rend le fichier lisible.

````python
donnees = {'nom': 'Alice', 'age': 30, 'langues': ['fr', 'en']}
print(json.dumps(donnees, indent=2))
````

### `ensure_ascii=False` — garder les caractères UTF-8

Par défaut, `json` échappe les caractères non-ASCII (`é` → `\u00e9`). Pour du JSON lisible en français, passer `ensure_ascii=False`.

````python
donnees = {'prenom': 'Chloé', 'ville': 'Marseille'}
print(json.dumps(donnees))
print(json.dumps(donnees, ensure_ascii=False))
````

## 6. Types sérialisables

| Python | JSON |
|---|---|
| `dict` | object |
| `list`, `tuple` | array |
| `str` | string |
| `int`, `float` | number |
| `True`, `False` | `true`, `false` |
| `None` | `null` |

⚠️ Les autres types (`set`, `datetime`, `Path`, classes custom) ne sont **pas** sérialisables directement. Il faut les convertir d'abord.

````python
try:
    json.dumps({'tags': {1, 2, 3}})
except TypeError as err:
    print('erreur :', err)
````

---


## 7. Gérer les datetimes

Convention standard : les datetimes sont sérialisés en **chaîne ISO 8601**.

````python
from datetime import datetime
from zoneinfo import ZoneInfo

rdv = datetime(2026, 4, 14, 9, 30, tzinfo=ZoneInfo('Europe/Paris'))
donnees = {'titre': 'réunion', 'date': rdv.isoformat()}
chaine = json.dumps(donnees, ensure_ascii=False, indent=2)
print(chaine)
````

````python
# Relecture
brut = json.loads(chaine)
print(brut)
parsed_date = datetime.fromisoformat(brut['date'])
print(parsed_date)
````

## 8. Synthèse

| Besoin | Forme |
|---|---|
| Objet → chaîne | `json.dumps(obj)` |
| Chaîne → objet | `json.loads(s)` |
| Objet → fichier | `json.dump(obj, f)` |
| Fichier → objet | `json.load(f)` |
| Joli format | `indent=2` |
| Accents | `ensure_ascii=False` |

### Règles

1. Toujours `encoding='utf-8'` sur le fichier.
2. `indent=2, ensure_ascii=False` pour du JSON destiné à l'humain.
3. Datetimes → ISO 8601 (`.isoformat()`).
4. Les sets deviennent des listes avant sérialisation.

## Mini-exemples supplémentaires

### JSON d'une liste de nombres

````python
import json
print(json.dumps([1, 2, 3, 4]))
````

### Imbrication profonde

````python
import json
data = {'a': {'b': {'c': [1, 2, {'d': 'fin'}]}}}
print(json.dumps(data, indent=2))
````

### `None` devient `null`

````python
import json
print(json.dumps({'x': None}))
````

### Le roundtrip perd les tuples

````python
import json
orig = {'coord': (3, 4)}
dump = json.dumps(orig)
load = json.loads(dump)
print(type(orig['coord']))
print(type(load['coord']))  # list, pas tuple
````

JSON ne distingue pas `list` et `tuple` : les deux deviennent un array, relu comme une `list`.

### `separators` — JSON compact

````python
import json
data = {'a': 1, 'b': 2}
print(json.dumps(data))
print(json.dumps(data, separators=(',', ':')))
````

### Valider une clé obligatoire

````python
import json

def lire_config(texte: str) -> dict[str, object]:
    config = json.loads(texte)
    if 'version' not in config:
        raise ValueError('clé `version` manquante')
    return config

print(lire_config('{"version": 1}'))
````
