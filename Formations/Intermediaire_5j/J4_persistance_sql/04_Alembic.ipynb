## 1. Pourquoi des migrations

En développement, le schéma de la base évolue. Sans outil de migration :

- on perd la trace des modifications ;
- chaque dev doit recréer sa base à la main ;
- le passage en production est un casse-tête.

**Alembic** gère des **révisions** versionnées, chacune avec un `upgrade()` et un `downgrade()`. Comme un `git` pour le schéma SQL.

---

## 2. `alembic init`

On montre les commandes en mode démonstratif. En pratique, on les exécute dans un terminal.

````python
# Commandes à exécuter dans un terminal (pas dans le notebook) :
#
# alembic init migrations
#
# Cela crée :
# migrations/
#   env.py          — configuration de l'environnement Alembic
#   script.py.mako  — template des révisions
#   versions/       — dossier des fichiers de migration
# alembic.ini       — config (URL de la base, etc.)

print('⬆️  Commandes à exécuter en terminal')
````

**Dans `alembic.ini`**, modifier `sqlalchemy.url` pour pointer vers votre base :

```ini
sqlalchemy.url = sqlite:///./data.db
```

## 3. Créer une révision

---

## 4. `upgrade` et `downgrade`

````python
# Appliquer toutes les migrations :
# alembic upgrade head
#
# Revenir en arrière d'une révision :
# alembic downgrade -1
#
# Voir l'état actuel :
# alembic current
#
# Voir l'historique :
# alembic history

print('⬆️  Commandes en terminal')
````

## 5. Auto-génération

Alembic peut **comparer** le modèle ORM au schéma réel et générer la migration automatiquement :

````python
# alembic revision --autogenerate -m "ajout colonne description"
#
# Prérequis : configurer target_metadata dans env.py
# target_metadata = Base.metadata

print('⬆️  Autogenerate en terminal')
````

---

## 6. Bonnes pratiques

- **Toujours écrire le `downgrade`** : pouvoir revenir en arrière est crucial.
- **Versionner les fichiers de migration** dans git.
- **Relire les migrations auto-générées** : Alembic ne gère pas tout (renommage, migration de données).
- **Tester les migrations** sur une base de test avant la prod.
- **Ne jamais modifier une migration déjà appliquée en prod.** Créer une nouvelle révision.

---

## Synthèse

| Commande | Rôle |
|---|---|
| `alembic init` | Initialiser l'arborescence |
| `alembic revision -m "..."` | Créer une migration |
| `alembic upgrade head` | Appliquer toutes les migrations |
| `alembic downgrade -1` | Revenir d'un cran |
| `alembic current` | État courant |
| `--autogenerate` | Comparer modèle vs base |


### Règles à retenir

1. **Migrations = git du schéma.** Versionnez-les.
2. **Toujours écrire `downgrade`.**
3. **Relire les migrations auto-générées** : elles sont un point de départ, pas un produit fini.
4. **Tester sur une base jetable** avant la prod.
