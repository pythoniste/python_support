## 1. Pourquoi `pyproject.toml`

C'est **le** fichier de configuration standard d'un projet Python (PEP 517, 518, 621). Il remplace `setup.py`, `setup.cfg`, `requirements.txt` comme source de vérité unique.

---

## 2. Structure d'un `pyproject.toml`

````python
# Contenu type d'un pyproject.toml
contenu = """
[project]
name = "reservation-salles"
version = "0.1.0"
description = "Gestion de réservation de salles"
requires-python = ">=3.12"
dependencies = [
    "pydantic>=2.0",
    "sqlalchemy>=2.0",
]

[project.optional-dependencies]
dev = [
    "pytest>=8.0",
    "ruff",
    "mypy",
]

[project.scripts]
reservation = "reservation_salles.cli:main"

[build-system]
requires = ["hatchling"]
build-backend = "hatchling.build"
"""
print(contenu)
````

---

## 3. `uv` : le gestionnaire moderne

`uv` (Astral, les créateurs de `ruff`) est un **gestionnaire de packages ultra-rapide**. Il remplace `pip`, `pip-tools`, `virtualenv`, `pipx` en une seule commande.

```bash
# Installer uv
curl -LsSf https://astral.sh/uv/install.sh | sh

# Créer un nouveau projet
uv init reservation-salles
cd reservation-salles
```

---

## 4. Workflow : `init`, `add`, `sync`, `lock`, `run`

```bash
# Ajouter une dépendance
uv add pydantic sqlalchemy
uv add --dev pytest ruff mypy

# Synchroniser (installe les dépendances exactes du lock)
uv sync

# Lock (met à jour le lock file)
uv lock

# Exécuter une commande dans l'environnement
uv run pytest
uv run python -m reservation_salles
```

Le fichier `uv.lock` est **déterministe** : il garantit que tous les développeurs ont exactement les mêmes versions. Comme `Pipfile.lock` mais en beaucoup mieux.

---

## 5. `uv build` : créer un wheel

```bash
uv build
# Crée dist/reservation_salles-0.1.0-py3-none-any.whl
# et dist/reservation_salles-0.1.0.tar.gz
```

Un **wheel** est un package installable avec `pip install fichier.whl`. C'est le format standard de distribution.

---

## 7. Workspace

Pour les monorepos (plusieurs packages dans un même dépôt).

```toml
# pyproject.toml racine
[tool.uv.workspace]
members = ["packages/*"]
```

Chaque sous-package a son propre `pyproject.toml`. `uv` gère les dépendances croisées automatiquement.

---

## Synthèse

| Commande | Rôle |
|---|---|
| `uv init` | Créer un projet |
| `uv add pkg` | Ajouter une dépendance |
| `uv add --dev pkg` | Dépendance de dev |
| `uv sync` | Installer depuis le lock |
| `uv lock` | Résoudre et verrouiller |
| `uv run cmd` | Exécuter dans l'env |
| `uv build` | Créer un wheel |


### Règles à retenir

1. **Un seul fichier** : `pyproject.toml`. Pas de `requirements.txt` séparé.
2. **`uv.lock` dans git** : reproductibilité garantie.
3. **`uv run pytest`** plutôt que d'activer le venv à la main.
4. **Points d'entrée** dans `[project.scripts]` pour les CLI.
