## 1. Installation et premier affichage

````python
from rich.console import Console
from rich import print as rprint

console = Console()
console.print('Hello [bold magenta]Rich[/bold magenta]!')
````

---

## 2. Console et styles

````python
console.print('[bold red]Erreur[/bold red] : fichier introuvable')
console.print('[italic green]Succes[/italic green] : operation terminee')
console.print('[underline blue]https://example.com[/underline blue]')
````

````python
from rich.text import Text

t = Text('Python est genial')
t.stylize('bold', 0, 6)
t.stylize('italic red', 11, 17)
console.print(t)
````

### Styles disponibles

`bold`, `italic`, `underline`, `strike`, `dim`, `reverse`, `blink`. Couleurs : `red`, `green`, `blue`, `yellow`, `magenta`, `cyan`, `white`, ou codes hex `#ff5733`.

## 3. Tableaux

````python
from rich.table import Table

table = Table(title='Langages de programmation')
table.add_column('Langage', style='cyan', justify='left')
table.add_column('Annee', style='magenta')
table.add_column('Paradigme', style='green')

table.add_row('Python', '1991', 'Multi-paradigme')
table.add_row('Rust', '2010', 'Systeme')
table.add_row('Go', '2009', 'Concurrent')
table.add_row('TypeScript', '2012', 'Oriente objet')

console.print(table)
````

## 4. Arbres

````python
from rich.tree import Tree

tree = Tree('projet/')
src = tree.add('src/')
src.add('main.py')
src.add('utils.py')
tests = tree.add('tests/')
tests.add('test_main.py')
tree.add('pyproject.toml')
tree.add('README.md')

console.print(tree)
````

## 5. Panneaux et colonnes

````python
from rich.panel import Panel

console.print(Panel('Contenu important', title='Attention', border_style='red'))
````

````python
from rich.columns import Columns

items = [Panel(f'Item {i}', expand=True) for i in range(6)]
console.print(Columns(items, equal=True))
````

---

## 6. Barres de progression

````python
from rich.progress import track
import time

for i in track(range(20), description='Traitement...'):
    time.sleep(0.05)
````

### Progress avance

````python
from rich.progress import Progress

with Progress() as progress:
    task1 = progress.add_task('[red]Telechargement', total=100)
    task2 = progress.add_task('[green]Traitement', total=100)
    while not progress.finished:
        progress.update(task1, advance=1)
        progress.update(task2, advance=0.5)
        time.sleep(0.02)
````

## 7. Syntaxe et Markdown

````python
from rich.syntax import Syntax

code_python = '''
def fibonacci(n: int) -> int:
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
'''

syntax = Syntax(code_python, 'python', theme='monokai', line_numbers=True)
console.print(syntax)
````

````python
from rich.markdown import Markdown

md_text = '''
# Titre

**Gras** et *italique*.

- Element 1
- Element 2

```python
print('hello')
```
'''
console.print(Markdown(md_text))
````

## 8. Logging avec Rich

````python
import logging
from rich.logging import RichHandler

logging.basicConfig(
    level=logging.DEBUG,
    format='%(message)s',
    handlers=[RichHandler(rich_tracebacks=True)],
)
log = logging.getLogger('demo')
log.info('Message info')
log.warning('Attention')
log.error('Erreur')
````

## 10. Live display

````python
from rich.live import Live

table = Table()
table.add_column('Iteration')
table.add_column('Valeur')

with Live(table, refresh_per_second=4) as live:
    for i in range(10):
        table.add_row(str(i), str(i * i))
        time.sleep(0.2)
````

## Synthese

| Composant | Usage |
|---|---|
| `Console.print` | Texte stylise avec balises |
| `Table` | Tableaux formmates |
| `Tree` | Arborescences |
| `Panel` | Encadrement |
| `track` / `Progress` | Barres de progression |
| `Syntax` | Coloration syntaxique |
| `Markdown` | Rendu Markdown |
| `RichHandler` | Integration logging |
| `inspect` | Introspection d'objets |
| `Live` | Affichage dynamique |

### Regles a retenir

- **`console.print`** remplace `print` pour le texte riche.
- **`track`** pour les barres de progression simples.
- **`RichHandler`** pour des logs lisibles en dev.
- **Combinable avec Typer** pour des CLI elegants.
