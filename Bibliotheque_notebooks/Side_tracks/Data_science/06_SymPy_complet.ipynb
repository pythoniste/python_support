## 13. Equations differentielles

````python
f = sp.Function('f')
eq = sp.Eq(f(x).diff(x) - f(x), 0)
eq
````

````python
sp.dsolve(eq, f(x))
````

````python
# Equation du second ordre
eq2 = sp.Eq(f(x).diff(x, 2) + f(x), 0)
sp.dsolve(eq2, f(x))
````

## 15. Lambdify et performance

`lambdify` convertit une expression SymPy en fonction Python numerique rapide.

````python
expr = x**2 + sp.sin(x)
f = sp.lambdify(x, expr, 'math')
f(2.0)
````

````python
# Avec numpy pour vectoriser
import numpy as np
f_np = sp.lambdify(x, expr, 'numpy')
f_np(np.array([0.0, 1.0, 2.0]))
````

## Synthese

| Operation | SymPy |
|---|---|
| Symbole | `sp.Symbol('x')` ou `sp.symbols('x y')` |
| Simplification | `sp.simplify`, `sp.trigsimp` |
| Developper / factoriser | `sp.expand`, `sp.factor` |
| Resolution | `sp.solve`, `sp.nonlinsolve` |
| Derivation | `sp.diff(expr, x)` |
| Integration | `sp.integrate(expr, x)` ou `(x, a, b)` |
| Limite | `sp.limit(expr, x, val)` |
| Serie | `sp.series(expr, x, 0, n)` |
| Matrice | `sp.Matrix([[...]])` |
| ODE | `sp.dsolve(eq, f(x))` |
| LaTeX | `sp.latex(expr)` |
| Fonction rapide | `sp.lambdify(x, expr, 'numpy')` |

### Regles a retenir

- **SymPy est exact** : manipulations symboliques, pas d'erreurs numeriques.
- **Distinguer `Symbol` et `Function`** : `Symbol` pour x, `Function` pour f(x).
- **`subs` pour evaluer**, `evalf` pour un resultat decimal.
- **Pour la performance**, passer par `lambdify` puis NumPy.
