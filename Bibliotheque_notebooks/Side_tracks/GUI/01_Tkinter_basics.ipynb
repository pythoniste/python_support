### Exemple avancé : calculatrice simple

Combinons plusieurs widgets pour créer une calculatrice basique.
Cet exemple illustre l'utilisation conjointe de `grid`, `Entry`, `Button` et `StringVar`.

````python
import tkinter as tk

def create_calculator():
    root = tk.Tk()
    root.title('Calculatrice')

    display_var = tk.StringVar(value='0')
    display = tk.Entry(root, textvariable=display_var, font=('Arial', 18),
                       justify='right', state='readonly')
    display.grid(row=0, column=0, columnspan=4, sticky='ew', padx=5, pady=5)

    expression = ['']

    def on_click(char):
        if char == 'C':
            expression[0] = ''
            display_var.set('0')
        elif char == '=':
            try:
                result = str(eval(expression[0]))
                display_var.set(result)
                expression[0] = result
            except Exception:
                display_var.set('Erreur')
                expression[0] = ''
        else:
            expression[0] += char
            display_var.set(expression[0])

    buttons = [
        ('7','8','9','/'),
        ('4','5','6','*'),
        ('1','2','3','-'),
        ('C','0','=','+'),
    ]
    for r, row in enumerate(buttons, start=1):
        for c, char in enumerate(row):
            btn = tk.Button(root, text=char, font=('Arial', 14), width=4,
                           command=lambda ch=char: on_click(ch))
            btn.grid(row=r, column=c, padx=2, pady=2)

    return root

# Décommenter pour tester :
# root = create_calculator()
# root.mainloop()
print('Code de la calculatrice prêt (décommentez pour lancer)')
````

### Exercice : formulaire d'inscription

Créez un formulaire d'inscription avec Tkinter comprenant :

1. Champs `Entry` pour : nom, prénom, email
2. Un `Spinbox` pour l'âge (18-99)
3. Des `Radiobutton` pour le genre (H/F/Autre)
4. Une `Checkbutton` pour accepter les CGU
5. Un bouton "Valider" qui affiche un résumé dans un `Label`

Utilisez `grid` pour la mise en page.

````python
import tkinter as tk

def create_registration_form():
    root = tk.Tk()
    root.title('Inscription')

    # Variables
    nom_var = tk.StringVar()
    prenom_var = tk.StringVar()
    email_var = tk.StringVar()
    age_var = tk.IntVar(value=25)
    genre_var = tk.StringVar(value='H')
    cgu_var = tk.BooleanVar()

    # Champs
    fields = [('Nom', nom_var), ('Prénom', prenom_var), ('Email', email_var)]
    for i, (label, var) in enumerate(fields):
        tk.Label(root, text=f'{label} :').grid(row=i, column=0, sticky='e', padx=5)
        tk.Entry(root, textvariable=var, width=30).grid(row=i, column=1, padx=5, pady=2)

    # Age
    tk.Label(root, text='Âge :').grid(row=3, column=0, sticky='e', padx=5)
    tk.Spinbox(root, from_=18, to=99, textvariable=age_var, width=5).grid(row=3, column=1, sticky='w', padx=5)

    # Genre
    tk.Label(root, text='Genre :').grid(row=4, column=0, sticky='e', padx=5)
    frame_genre = tk.Frame(root)
    frame_genre.grid(row=4, column=1, sticky='w')
    for txt, val in [('Homme', 'H'), ('Femme', 'F'), ('Autre', 'A')]:
        tk.Radiobutton(frame_genre, text=txt, variable=genre_var, value=val).pack(side='left')

    # CGU
    tk.Checkbutton(root, text="J'accepte les CGU", variable=cgu_var).grid(row=5, column=1, sticky='w')

    # Résultat
    result_label = tk.Label(root, text='', fg='green', wraplength=300)
    result_label.grid(row=7, column=0, columnspan=2, pady=10)

    def valider():
        if not cgu_var.get():
            result_label.config(text='Veuillez accepter les CGU', fg='red')
            return
        result_label.config(
            text=f'{prenom_var.get()} {nom_var.get()}, {age_var.get()} ans, {email_var.get()}',
            fg='green'
        )

    tk.Button(root, text='Valider', command=valider).grid(row=6, column=1, pady=5)
    return root

# Décommenter pour tester :
# root = create_registration_form()
# root.mainloop()
print('Formulaire prêt (décommentez pour lancer)')
````

### Exercice : liste dynamique avec Listbox

Créez une application avec :
- Un `Entry` pour saisir un élément
- Un bouton "Ajouter" pour l'ajouter à une `Listbox`
- Un bouton "Supprimer" pour retirer l'élément sélectionné
- Un `Label` affichant le nombre d'éléments

````python
import tkinter as tk

def create_dynamic_list():
    root = tk.Tk()
    root.title('Liste dynamique')

    entry_var = tk.StringVar()
    count_var = tk.StringVar(value='0 éléments')

    tk.Entry(root, textvariable=entry_var, width=30).pack(padx=10, pady=5)

    listbox = tk.Listbox(root, height=10, width=30)
    listbox.pack(padx=10)

    def update_count():
        n = listbox.size()
        count_var.set(f'{n} élément{"s" if n > 1 else ""}')

    def ajouter():
        text = entry_var.get().strip()
        if text:
            listbox.insert(tk.END, text)
            entry_var.set('')
            update_count()

    def supprimer():
        sel = listbox.curselection()
        if sel:
            listbox.delete(sel[0])
            update_count()

    frame_btn = tk.Frame(root)
    frame_btn.pack(pady=5)
    tk.Button(frame_btn, text='Ajouter', command=ajouter).pack(side='left', padx=5)
    tk.Button(frame_btn, text='Supprimer', command=supprimer).pack(side='left', padx=5)
    tk.Label(root, textvariable=count_var).pack(pady=5)

    return root

# root = create_dynamic_list()
# root.mainloop()
print('Liste dynamique prête (décommentez pour lancer)')
````

## Synthèse

| Élément | Tkinter | Notes |
|---|---|---|
| Fenêtre | `tk.Tk()` + `mainloop()` | Boucle bloquante |
| Label | `tk.Label(parent, text=)` | |
| Entry | `tk.Entry(parent, textvariable=)` | |
| Button | `tk.Button(parent, text=, command=)` | |
| Text | `tk.Text(parent, height=, width=)` | Multiligne |
| Check | `tk.Checkbutton(parent, variable=)` | `BooleanVar` |
| Radio | `tk.Radiobutton(parent, variable=, value=)` | `StringVar` partagée |
| Layout | `.pack()` / `.grid()` / `.place()` | Pas mixer pack+grid dans même parent |
| Variables | `StringVar`, `IntVar`, `BooleanVar` | `.get()`, `.set()`, `.trace_add()` |
| Événements | `.bind('<Event>', handler)` | |
| Dialogues | `messagebox.show*`, `filedialog.ask*` | |
| Menu | `tk.Menu` + `add_cascade/add_command` | |
| ttk | `ttk.Button`, `ttk.Entry`, `ttk.Combobox` | Thèmes natifs |

### Règles à retenir

- **Tkinter est inclus** dans la bibliothèque standard Python.
- **`ttk`** pour un rendu moderne et natif.
- **`pack` ou `grid`** dans le même conteneur, jamais les deux.
- **`mainloop()`** est bloquant : toute la logique doit être dans des callbacks.

### Approfondir : menus et sous-menus

Les applications Tkinter professionnelles utilisent des barres de menus
avec des sous-menus, raccourcis clavier et séparateurs.

````python
import tkinter as tk

def create_menu_app():
    root = tk.Tk()
    root.title('App avec menus')
    root.geometry('400x300')

    # Barre de menus
    menubar = tk.Menu(root)

    # Menu Fichier
    file_menu = tk.Menu(menubar, tearoff=0)
    file_menu.add_command(label='Nouveau', accelerator='Ctrl+N')
    file_menu.add_command(label='Ouvrir...', accelerator='Ctrl+O')
    file_menu.add_separator()
    file_menu.add_command(label='Quitter', command=root.quit, accelerator='Ctrl+Q')
    menubar.add_cascade(label='Fichier', menu=file_menu)

    # Menu Édition
    edit_menu = tk.Menu(menubar, tearoff=0)
    edit_menu.add_command(label='Copier', accelerator='Ctrl+C')
    edit_menu.add_command(label='Coller', accelerator='Ctrl+V')
    menubar.add_cascade(label='Edition', menu=edit_menu)

    # Menu Aide
    help_menu = tk.Menu(menubar, tearoff=0)
    help_menu.add_command(label='À propos',
        command=lambda: tk.messagebox.showinfo('À propos', 'App v1.0'))
    menubar.add_cascade(label='Aide', menu=help_menu)

    root.config(menu=menubar)
    return root

# root = create_menu_app()
# root.mainloop()
print('App avec menus prête')
````

3. Un clic gauche crée un cercle permanent à l'endroit du clic
4. Un clic droit efface le dernier cercle

````python
import tkinter as tk

def create_drawing_app():
    root = tk.Tk()
    root.title('Dessin')

    canvas = tk.Canvas(root, width=400, height=400, bg='white')
    canvas.pack()

    circles = []
    cursor_circle = None

    def on_move(event):
        nonlocal cursor_circle
        r = 15
        if cursor_circle:
            canvas.delete(cursor_circle)
        cursor_circle = canvas.create_oval(
            event.x-r, event.y-r, event.x+r, event.y+r,
            outline='blue', dash=(3, 3)
        )

    def on_click(event):
        r = 15
        c = canvas.create_oval(
            event.x-r, event.y-r, event.x+r, event.y+r,
            fill='lightblue', outline='blue'
        )
        circles.append(c)

    def on_right_click(event):
        if circles:
            canvas.delete(circles.pop())

    canvas.bind('<Motion>', on_move)
    canvas.bind('<Button-1>', on_click)
    canvas.bind('<Button-3>', on_right_click)

    return root

# root = create_drawing_app()
# root.mainloop()
print('App de dessin prête')
````

### Exercice : chronomètre

Créez un chronomètre avec :
- Un affichage MM:SS.ms
- Boutons Start / Stop / Reset
- Utilisation de `root.after()` pour la mise à jour périodique

````python
import tkinter as tk
import time

def create_stopwatch():
    root = tk.Tk()
    root.title('Chronomètre')

    state = {'running': False, 'start_time': 0, 'elapsed': 0}
    time_var = tk.StringVar(value='00:00.000')

    tk.Label(root, textvariable=time_var, font=('Courier', 36)).pack(pady=20)

    def update_display():
        if state['running']:
            elapsed = state['elapsed'] + (time.time() - state['start_time'])
            mins = int(elapsed // 60)
            secs = int(elapsed % 60)
            ms = int((elapsed % 1) * 1000)
            time_var.set(f'{mins:02d}:{secs:02d}.{ms:03d}')
            root.after(50, update_display)

    def start():
        if not state['running']:
            state['running'] = True
            state['start_time'] = time.time()
            update_display()

    def stop():
        if state['running']:
            state['elapsed'] += time.time() - state['start_time']
            state['running'] = False

    def reset():
        state['running'] = False
        state['elapsed'] = 0
        time_var.set('00:00.000')

    frame = tk.Frame(root)
    frame.pack(pady=10)
    tk.Button(frame, text='Start', command=start, width=8).pack(side='left', padx=5)
    tk.Button(frame, text='Stop', command=stop, width=8).pack(side='left', padx=5)
    tk.Button(frame, text='Reset', command=reset, width=8).pack(side='left', padx=5)

    return root

# root = create_stopwatch()
# root.mainloop()
print('Chronomètre prêt')
````

### Exercice bonus : Treeview pour données hiérarchiques

Le widget `ttk.Treeview` permet d'afficher des données hiérarchiques (arbre)
ou tabulaires avec tri et sélection.

````python
import tkinter as tk
from tkinter import ttk

def create_treeview_app():
    root = tk.Tk()
    root.title('Treeview')

    # Données hiérarchiques
    tree = ttk.Treeview(root, columns=('type', 'taille'), show='tree headings')
    tree.heading('#0', text='Nom')
    tree.heading('type', text='Type')
    tree.heading('taille', text='Taille')

    # Arborescence
    projet = tree.insert('', 'end', text='mon_projet', values=('dossier', ''))
    src = tree.insert(projet, 'end', text='src', values=('dossier', ''))
    tree.insert(src, 'end', text='main.py', values=('Python', '2.4 Ko'))
    tree.insert(src, 'end', text='utils.py', values=('Python', '1.1 Ko'))
    tests = tree.insert(projet, 'end', text='tests', values=('dossier', ''))
    tree.insert(tests, 'end', text='test_main.py', values=('Python', '0.8 Ko'))
    tree.insert(projet, 'end', text='README.md', values=('Markdown', '0.5 Ko'))
    tree.insert(projet, 'end', text='pyproject.toml', values=('TOML', '0.3 Ko'))

    tree.pack(fill='both', expand=True, padx=10, pady=10)

    # Sélection
    label = tk.Label(root, text='Sélection : (aucune)')
    label.pack(pady=5)
    tree.bind('<<TreeviewSelect>>', lambda e:
        label.config(text=f'Sélection : {tree.item(tree.selection()[0])["text"]}' if tree.selection() else ''))

    return root

# root = create_treeview_app()
# root.mainloop()
print('Treeview prête')
````
