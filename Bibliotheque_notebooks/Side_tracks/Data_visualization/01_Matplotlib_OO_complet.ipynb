## 1. Installation et import

`pip install matplotlib`

**Ne jamais** faire `from pylab import *` ni `import matplotlib.pyplot as plt` suivi d'appels globaux. On importe `pyplot` pour son `subplots()` et son `show()`, puis on manipule les objets `Figure` et `Axes`.


````python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.dpi'] = 90
````

---


## 2. Première figure


````python
x = np.linspace(0, 2*np.pi, 200)
y = np.sin(x)
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title('Sinus')
ax.set_xlabel('x')
ax.set_ylabel('sin(x)')
plt.show()
````

**Pattern canonique** : `fig, ax = plt.subplots()` → on appelle des méthodes sur `ax` → `plt.show()`.


---


## 3. Figure, Axes, Artists

- **Figure** : la fenêtre/conteneur racine. Contient des Axes.
- **Axes** : une zone de tracé (un graphique). Contient des Artists.
- **Artist** : tout ce qui se dessine (ligne, texte, rectangle, image...).

Toutes les méthodes de dessin sont sur `Axes`, pas sur `Figure`.


````python
fig = plt.figure(figsize=(6, 3))
ax = fig.add_subplot(1, 1, 1)
ax.plot([0, 1, 2], [0, 1, 4])
plt.show()
````

---


## 4. `subplots` et `subplot_mosaic`


````python
fig, axes = plt.subplots(1, 2, figsize=(8, 3))
axes[0].plot(x, np.sin(x))
axes[1].plot(x, np.cos(x))
axes[0].set_title('sin')
axes[1].set_title('cos')
plt.tight_layout()
plt.show()
````

````python
fig, axes = plt.subplots(2, 2, figsize=(6, 5), sharex=True)
axes[0, 0].plot(x, np.sin(x))
axes[0, 1].plot(x, np.cos(x))
axes[1, 0].plot(x, np.tan(x))
axes[1, 0].set_ylim(-5, 5)
axes[1, 1].plot(x, np.sin(x)*np.cos(x))
plt.tight_layout()
plt.show()
````

### `subplot_mosaic` pour des layouts complexes


````python
fig, axes = plt.subplot_mosaic('''
    AAB
    CDB
''', figsize=(8, 5))
axes['A'].plot(x, np.sin(x))
axes['B'].plot(x, np.cos(x))
axes['C'].plot(x, np.sin(x)**2)
axes['D'].plot(x, np.cos(x)**2)
plt.tight_layout()
plt.show()
````

---


## 5. Lignes et marqueurs


````python
fig, ax = plt.subplots()
ax.plot(x, np.sin(x), label='sin', linewidth=2, color='C0')
ax.plot(x, np.cos(x), label='cos', linestyle='--', color='C1')
ax.legend()
plt.show()
````

### Marqueurs


````python
xs = np.arange(10)
ys = xs**1.5
fig, ax = plt.subplots()
ax.plot(xs, ys, marker='o', linestyle='-', markersize=8)
plt.show()
````

---


## 6. Nuages de points


````python
rng = np.random.default_rng(0)
xs = rng.normal(size=200)
ys = xs + rng.normal(size=200) * 0.3
fig, ax = plt.subplots()
ax.scatter(xs, ys, alpha=0.5, s=30, c=xs, cmap='viridis')
ax.set_xlabel('x')
ax.set_ylabel('y')
plt.show()
````

---


## 7. Histogrammes


````python
rng = np.random.default_rng(0)
data = rng.normal(loc=0, scale=1, size=1000)
fig, ax = plt.subplots()
ax.hist(data, bins=30, edgecolor='black')
ax.set_title('Loi normale')
plt.show()
````

### Densité (KDE grossière avec histogramme normalisé)


````python
fig, ax = plt.subplots()
ax.hist(data, bins=30, density=True, alpha=0.5, label='histogramme')
ax.set_ylabel('densité')
ax.legend()
plt.show()
````

---


## 8. Barres


````python
villes = ['Paris', 'Lyon', 'Nice', 'Marseille']
pop = [2.1, 0.5, 0.34, 0.86]
fig, ax = plt.subplots()
ax.bar(villes, pop, color='steelblue')
ax.set_ylabel('Population (millions)')
plt.show()
````

### Barres horizontales (meilleur pour de longs labels)


````python
fig, ax = plt.subplots()
ax.barh(villes, pop, color='teal')
ax.set_xlabel('Population (millions)')
plt.show()
````

### Barres empilées


````python
annees = [2020, 2021, 2022]
prod_a = np.array([10, 12, 14])
prod_b = np.array([6, 8, 10])
fig, ax = plt.subplots()
ax.bar(annees, prod_a, label='A')
ax.bar(annees, prod_b, bottom=prod_a, label='B')
ax.legend()
plt.show()
````

## 9. Remplissage (`fill_between`)

## 13. Légendes


````python
fig, ax = plt.subplots()
ax.plot(x, np.sin(x), label='sin')
ax.plot(x, np.cos(x), label='cos')
ax.legend(loc='upper right', frameon=False)
plt.show()
````

### Légende hors du cadre


````python
fig, ax = plt.subplots()
ax.plot(x, np.sin(x), label='sin')
ax.plot(x, np.cos(x), label='cos')
ax.legend(loc='center left', bbox_to_anchor=(1.02, 0.5))
fig.tight_layout()
plt.show()
````

## 14. Titres, labels, annotations


````python
fig, ax = plt.subplots()
ax.plot(x, np.sin(x))
ax.set_title('Fonction sinus', fontsize=14)
ax.set_xlabel('x (rad)')
ax.set_ylabel('sin(x)')
ax.grid(True, alpha=0.3)
ax.annotate('maximum', xy=(np.pi/2, 1), xytext=(np.pi/2 + 0.5, 1.2),
    arrowprops={'arrowstyle': '->'})
plt.show()
````

## Synthèse

| Notion | À retenir |
|---|---|
| API OO | `fig, ax = plt.subplots()` → tout passe par `ax` |
| Figure | conteneur ; `Axes` = zone de tracé |
| `subplots(nrows, ncols)` | grille simple ; `subplot_mosaic` pour complexe |
| `plot`, `scatter`, `bar`, `hist`, `imshow` | types de base |
| `legend`, `annotate` | lisibilité |
| Styles | `plt.style.context('...')` ou `rcParams` |
| Export | `savefig(..., dpi=150, bbox_inches='tight')` |


### Règles à retenir

- **Jamais** `from pylab import *` ni API globale (`plt.plot`).
- **Toujours** `fig, ax = plt.subplots()` en premier.
- **Un graphique = un `Axes`**, une figure peut en contenir plusieurs.
- **`tight_layout`** ou `constrained_layout=True` pour éviter les chevauchements.
- **SVG/PDF** pour l'impression, PNG pour le web.

import matplotlib.pyplot as plt
import numpy as np
from collections.abc import Callable

def tracer_fonction(f: Callable[[float], float], a: float, b: float) -> plt.Figure:
    xs = np.linspace(a, b, 200)
    ys = np.vectorize(f)(xs)
    fig, ax = plt.subplots()
    ax.plot(xs, ys)
    ax.set_xlabel('x')

def histo_avec_moyenne(data: np.ndarray) -> plt.Figure:
    fig, ax = plt.subplots()
    ax.hist(data, bins=30, edgecolor='black')
    m = float(data.mean())
    ax.axvline(m, color='red', linestyle='--')
    ax.annotate(f'moyenne = {m:.2f}', xy=(m, 0), xytext=(m, 5),
