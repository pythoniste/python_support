## 6. Chemins (Path)

Les chemins permettent des formes libres via les commandes SVG `M`, `L`, `C`, `Q`, `A`, `Z`.

````python
d = dw.Drawing(200, 100)
path = dw.Path(stroke='purple', stroke_width=3, fill='none')
path.M(10, 50).Q(100, 10, 190, 50).L(190, 90).L(10, 90).Z()
d.append(path)
d
````

````python
# Courbe de Bezier cubique
d = dw.Drawing(200, 100)
p = dw.Path(stroke='blue', stroke_width=3, fill='none')
p.M(10, 80).C(50, 10, 150, 10, 190, 80)
d.append(p)
d
````

## 7. Couleurs et styles

## 8. Groupes et transformations

````python
d = dw.Drawing(300, 150)
g = dw.Group(transform='translate(50, 50) rotate(15)')
g.append(dw.Rectangle(-30, -20, 60, 40, fill='mediumseagreen'))
g.append(dw.Circle(0, 0, 10, fill='white'))
d.append(g)
d
````

````python
# Plusieurs rotations
d = dw.Drawing(300, 300)
for angle in range(0, 360, 30):
    g = dw.Group(transform=f'translate(150,150) rotate({angle})')
    g.append(dw.Rectangle(0, -5, 100, 10, fill='steelblue'))
    d.append(g)
d
````

---

## 9. Gradients

````python
d = dw.Drawing(300, 100)
gradient = dw.LinearGradient(0, 0, 1, 0)
gradient.add_stop(0, 'red')
gradient.add_stop(1, 'yellow')
d.append(gradient)
d.append(dw.Rectangle(10, 10, 280, 80, fill=gradient.paint()))
d
````

````python
# Gradient radial
d = dw.Drawing(200, 200)
rg = dw.RadialGradient(100, 100, 80)
rg.add_stop(0, 'white')
rg.add_stop(1, 'navy')
d.append(rg)
d.append(dw.Circle(100, 100, 80, fill=rg.paint()))
d
````

## 13. Graphique reutilisable

````python
def barchart_svg(valeurs: list[float], labels: list[str], width: int = 300, height: int = 200) -> dw.Drawing:
    d = dw.Drawing(width, height)
    max_v = max(valeurs)
    n = len(valeurs)
    bar_w = (width - 40) / n
    for i, (v, lb) in enumerate(zip(valeurs, labels)):
        h = (v / max_v) * (height - 40)
        x = 20 + i * bar_w
        y = height - 20 - h
        d.append(dw.Rectangle(x, y, bar_w - 5, h, fill='steelblue'))
        d.append(dw.Text(lb, 12, x, height - 5, fill='black'))
    return d

barchart_svg([3, 7, 5, 9, 4], ['A', 'B', 'C', 'D', 'E'])
````

## Synthese

| Element | drawsvg |
|---|---|
| Canvas | `dw.Drawing(w, h)` |
| Rectangle | `dw.Rectangle(x, y, w, h, fill=...)` |
| Cercle | `dw.Circle(cx, cy, r, fill=...)` |
| Ligne | `dw.Line(x1, y1, x2, y2, stroke=...)` |
| Polygone | `dw.Lines(..., close=True)` |
| Chemin | `dw.Path().M().L().C().Z()` |
| Texte | `dw.Text(s, size, x, y)` |
| Groupe | `dw.Group(transform='...')` |
| Gradient | `dw.LinearGradient` / `dw.RadialGradient` |
| Export | `.save_svg()`, `.save_png()` |

### Regles a retenir

- **Systeme de coords** : y pointe vers le bas (comme SVG).
- **Composer** : groupes + transformations.
- **Eviter XML a la main** : la classe Python est plus lisible.
- **Reutiliser** : encapsuler les graphiques dans des fonctions.

import drawsvg as dw
import math

def camembert(valeurs: list[float], colors: list[str]) -> dw.Drawing:
    d = dw.Drawing(300, 300)
    total = sum(valeurs)
    cx, cy, r = 150, 150, 120
    angle = -math.pi / 2
    for v, color in zip(valeurs, colors):
        slice_angle = 2 * math.pi * v / total
        x1 = cx + r * math.cos(angle)
        y1 = cy + r * math.sin(angle)
        angle += slice_angle
        x2 = cx + r * math.cos(angle)
        y2 = cy + r * math.sin(angle)
        large = 1 if slice_angle > math.pi else 0
        p = dw.Path(fill=color, stroke='white', stroke_width=2)
        p.M(cx, cy).L(x1, y1).A(r, r, 0, large, 1, x2, y2).Z()
        d.append(p)
    return d

import drawsvg as dw

def bar_chart(data: dict[str, float], width: int = 500, height: int = 300) -> dw.Drawing:
    d = dw.Drawing(width, height)
    d.append(dw.Rectangle(0, 0, width, height, fill='#fafafa'))
    items = list(data.items())
    max_v = max(data.values())
    margin_left = 80
    bar_h = (height - 40) / len(items)
    for i, (label, v) in enumerate(items):
        w = (v / max_v) * (width - margin_left - 60)
        y = 20 + i * bar_h
        d.append(dw.Rectangle(margin_left, y, w, bar_h - 10, fill='steelblue'))
        d.append(dw.Text(label, 14, 10, y + bar_h / 2, fill='black'))
        d.append(dw.Text(str(v), 14, margin_left + w + 5, y + bar_h / 2, fill='black'))
    return d
