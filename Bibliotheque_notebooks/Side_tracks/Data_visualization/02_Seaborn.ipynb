## 1. Installation et configuration

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

sns.__version__
````

````python
sns.set_theme(style='whitegrid', font_scale=1.1)
````

Seaborn propose 5 styles : `darkgrid`, `whitegrid`, `dark`, `white`, `ticks`. `whitegrid` est un bon defaut pour les presentations.

## 2. Jeux de donnees integres

Seaborn inclut plusieurs jeux de donnees classiques pour l'apprentissage.

````python
sns.get_dataset_names()[:10]
````

````python
tips = sns.load_dataset('tips')
tips.head()
````

````python
penguins = sns.load_dataset('penguins').dropna()
penguins.head()
````

---

## 3. Graphiques relationnels

### Scatterplot

````python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', style='sex')
plt.title('Pourboires vs addition')
plt.show()
````

### Lineplot

````python
fmri = sns.load_dataset('fmri')
sns.lineplot(data=fmri, x='timepoint', y='signal', hue='event', style='region')
plt.show()
````

`lineplot` affiche automatiquement l'intervalle de confiance a 95 % (zone ombree).

### relplot : interface unifiee

````python
sns.relplot(data=tips, x='total_bill', y='tip', hue='day', col='time', kind='scatter')
plt.show()
````

---

## 4. Graphiques de distribution

### Histogramme

````python
sns.histplot(data=penguins, x='flipper_length_mm', hue='species', kde=True)
plt.show()
````

### KDE (estimation par noyau)

````python
sns.kdeplot(data=penguins, x='flipper_length_mm', hue='species', fill=True)
plt.show()
````

### Boxplot

````python
sns.boxplot(data=penguins, x='species', y='body_mass_g')
plt.show()
````

### Violinplot

````python
sns.violinplot(data=penguins, x='species', y='body_mass_g', hue='sex', split=True)
plt.show()
````

### ECDF (fonction de repartition empirique)

````python
sns.ecdfplot(data=penguins, x='flipper_length_mm', hue='species')
plt.show()
````

---

## 5. Graphiques categoriques

### Barplot

````python
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', errorbar='sd')
plt.show()
````

### Countplot

````python
sns.countplot(data=tips, x='day', hue='time')
plt.show()
````

### Stripplot et Swarmplot

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.stripplot(data=tips, x='day', y='total_bill', ax=axes[0])
axes[0].set_title('stripplot')
sns.swarmplot(data=tips, x='day', y='total_bill', ax=axes[1], size=3)
axes[1].set_title('swarmplot')
plt.tight_layout()
plt.show()
````

### Combiner boxplot et stripplot

````python
fig, ax = plt.subplots()
sns.boxplot(data=tips, x='day', y='total_bill', ax=ax, fliersize=0)
sns.stripplot(data=tips, x='day', y='total_bill', ax=ax, color='black', alpha=0.3, size=3)
plt.show()
````

---

## 6. Graphiques matriciels (heatmap)

````python
corr = penguins.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Matrice de correlation')
plt.show()
````

### Clustermap

````python
sns.clustermap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.show()
````

---

## 7. FacetGrid et pairplot

### pairplot

````python
sns.pairplot(penguins, hue='species', diag_kind='kde')
plt.show()
````

`pairplot` cree une matrice de scatterplots pour chaque paire de variables numeriques.

### FacetGrid

````python
g = sns.FacetGrid(tips, col='time', row='sex', hue='smoker', height=3)
g.map_dataframe(sns.scatterplot, x='total_bill', y='tip')
g.add_legend()
plt.show()
````

### jointplot

````python
sns.jointplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', hue='species', kind='scatter')
plt.show()
````

---

## 8. Palettes de couleurs

````python
# Palettes qualitatives
for name in ['deep', 'muted', 'bright', 'pastel', 'dark', 'colorblind']:
    sns.palplot(sns.color_palette(name))
    plt.title(name)
    plt.show()
````

````python
# Palette sequentielle
sns.palplot(sns.color_palette('Blues', 8))
plt.title('Blues')
plt.show()
````

````python
# Palette divergente
sns.palplot(sns.color_palette('coolwarm', 10))
plt.title('coolwarm')
plt.show()
````

Pour l'accessibilite, privilegier la palette `colorblind`.

---

## 9. Personnalisation avec Matplotlib

Seaborn repose sur Matplotlib. On peut combiner les deux librement.

````python
fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(data=tips, x='total_bill', bins=20, kde=True, ax=ax)
ax.set_xlabel('Addition ($)', fontsize=12)
ax.set_ylabel('Frequence', fontsize=12)
ax.set_title('Distribution des additions', fontsize=14)
ax.axvline(tips['total_bill'].mean(), color='red', linestyle='--', label='Moyenne')
ax.legend()
plt.tight_layout()
plt.show()
````

### Sauvegarder en haute resolution

````python
fig, ax = plt.subplots()
sns.boxplot(data=tips, x='day', y='total_bill', ax=ax)
fig.savefig('/tmp/seaborn_example.png', dpi=150, bbox_inches='tight')
````

## Synthese

| Type | Fonction principale | Quand l'utiliser |
|---|---|---|
| Relationnel | `scatterplot`, `lineplot` | 2 variables continues |
| Distribution | `histplot`, `kdeplot`, `boxplot`, `violinplot` | Explorer une variable |
| Categorique | `barplot`, `countplot`, `stripplot` | Variable categorique x continue |
| Matriciel | `heatmap`, `clustermap` | Correlation, matrice |
| Multi-panneaux | `pairplot`, `FacetGrid`, `jointplot` | Vision d'ensemble |

### Regles a retenir

- **`set_theme`** en debut de notebook pour un style coherent.
- **`hue`** pour encoder une variable categorique par la couleur.
- **`col` / `row`** pour des sous-graphiques par categorie.
- **Palette `colorblind`** pour l'accessibilite.
- **Combiner avec Matplotlib** pour le controle fin (axes, annotations).
