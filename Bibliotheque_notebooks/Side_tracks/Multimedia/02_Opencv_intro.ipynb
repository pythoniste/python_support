## 8. Detection de formes

````python
# Trouver les contours sur une image binaire
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
print('Nb contours :', len(contours))
````

````python
# Dessiner les contours sur une copie
result = img.copy()
cv2.drawContours(result, contours, -1, (0, 255, 0), 2)
result.shape
````

## Synthese

| Operation | OpenCV |
|---|---|
| Lire | `cv2.imread(path)` |
| Ecrire | `cv2.imwrite(path, img)` |
| Couleur | `cv2.cvtColor(img, cv2.COLOR_BGR2GRAY/RGB/HSV)` |
| Resize | `cv2.resize(img, (w, h))` |
| Flou | `cv2.GaussianBlur(img, (k, k), 0)` |
| Canny | `cv2.Canny(gray, t1, t2)` |
| Contours | `cv2.findContours(binary, ...)` |
| Cascade | `CascadeClassifier(xml).detectMultiScale(gray)` |
| Dessiner | `cv2.rectangle/circle/line/putText` |
| Video | `VideoCapture`, `VideoWriter` |

### Regles a retenir

- **BGR et pas RGB** (attention avec matplotlib).
- **Images = np.ndarray uint8 (h, w, c)**.
- **`cv2.imshow` ne marche pas en notebook** : utiliser matplotlib ou sauver.
- **`astype(np.uint8)`** pour eviter les erreurs apres operations.

import cv2
import pathlib

def to_grayscale(path: str) -> str:
    p = pathlib.Path(path)
    img = cv2.imread(str(p))
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    out = str(p.with_stem(p.stem + '_gray'))
    cv2.imwrite(out, gray)
    return out
```

import cv2
import numpy as np

def rotate(path: str, angle: float) -> np.ndarray:
    img = cv2.imread(path)
    h, w = img.shape[:2]
    M = cv2.getRotationMatrix2D((w / 2, h / 2), angle, 1.0)
    return cv2.warpAffine(img, M, (w, h))
```

import cv2

def count_shapes(path: str) -> int:
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
    _, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY_INV)
