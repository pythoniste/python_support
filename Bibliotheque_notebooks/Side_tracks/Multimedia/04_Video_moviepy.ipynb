## 12. Pipeline complet

````python
def make_intro(input_path: str, title: str, output: str) -> None:
    from moviepy import VideoFileClip, TextClip, CompositeVideoClip
    clip = VideoFileClip(input_path).subclipped(0, 10).resized(height=720)
    txt = (
        TextClip(text=title, font_size=60, color='white')
        .with_duration(3)
        .with_position('center')
    )
    final = CompositeVideoClip([clip, txt])
    final.write_videofile(output)

print('pipeline defini')
````

## Synthese

| Operation | moviepy |
|---|---|
| Charger | `VideoFileClip(path)` |
| Decouper | `.subclipped(start, end)` |
| Concatener | `concatenate_videoclips([c1, c2])` |
| Redimensionner | `.resized(0.5)` ou `.resized(height=720)` |
| Crop | `.cropped(x1, y1, x2, y2)` |
| Texte | `TextClip(text='...', font_size=...)` |
| Overlay | `CompositeVideoClip([bg, fg])` |
| Audio | `.with_audio(audio)`, `.audio` |
| Effets | `.fx(fadein, 2)`, `.fx(speedx, 2)` |
| Export | `.write_videofile(path, codec='libx264')` |

### Regles a retenir

- **Durations en secondes**, positions en pixels ou relatives (`'center'`).
- **CompositeVideoClip** empile de bas en haut.
- **`write_videofile`** choisit le codec via `codec=`.
- **ffmpeg obligatoire**.

def montage(clips_paths: list[str], output: str, title: str) -> None:
    subs = [VideoFileClip(p).subclipped(0, 5) for p in clips_paths]
    combined = concatenate_videoclips(subs)
    txt = (
        TextClip(text=title, font_size=60, color='white')

        .with_position('center')
    )
    CompositeVideoClip([combined, txt]).write_videofile(output)
