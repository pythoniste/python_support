### Exemple : résolution d'un système linéaire

Résolvons le système $Ax = b$ symboliquement et numériquement.

````python
from sympy import Matrix, symbols, Rational, pprint

A = Matrix([[2, 1, -1], [-3, -1, 2], [-2, 1, 2]])
b = Matrix([8, -11, -3])
x = A.solve(b)
pprint(x)  # Solution : [2, 3, -1]
````

### Exercice : valeurs propres et vecteurs propres

1. Créez la matrice $M = \begin{pmatrix} 4 & -2 \\ 1 & 1 \end{pmatrix}$
2. Calculez ses valeurs propres avec `.eigenvals()`
3. Calculez ses vecteurs propres avec `.eigenvects()`
4. Vérifiez que $M v = \lambda v$ pour chaque couple propre

````python
from sympy import Matrix, pprint

M = Matrix([[4, -2], [1, 1]])
print('Valeurs propres :', M.eigenvals())
print('Vecteurs propres :')
for eigenval, mult, eigenvecs in M.eigenvects():
    v = eigenvecs[0]
    print(f'  lambda={eigenval}, v={v.T}, M*v={M*v}, lambda*v={(eigenval*v).T}')
````

## Synthèse

| Opération | SymPy | Notes |
|---|---|---|
| Symboles | `sp.Symbol('x')` / `sp.symbols('x y')` | Avec hypothèses optionnelles |
| Factoriser | `sp.factor(expr)` | |
| Développer | `sp.expand(expr)` | |
| Simplifier | `sp.simplify(expr)` | |
| Résoudre | `sp.solve(eq, x)` | Réel et complexe |
| Système lin. | `sp.linsolve(eqs, vars)` | |
| Matrice | `sp.Matrix([[...]])` | |
| Déterminant | `M.det()` | |
| Inverse | `M.inv()` | |
| Valeurs propres | `M.eigenvals()` / `eigenvects()` | |
| Dérivée | `sp.diff(expr, x)` | |
| Intégrale | `sp.integrate(expr, x)` | Définie : `(x, a, b)` |
| Limite | `sp.limit(expr, x, a)` | |
| Série | `sp.series(expr, x, 0, n)` | |
| LaTeX | `sp.latex(expr)` | |

### Règles à retenir

- **SymPy = calcul exact** (pas flottant).
- **`sp.Rational(1, 3)`** au lieu de `1/3` pour les fractions exactes.
- **`evalf(n)`** pour obtenir des décimales.
- Pour du **numérique rapide**, convertir avec `sp.lambdify(x, expr, 'numpy')`.

### Exercice récapitulatif : étude complète d'une fonction

En utilisant SymPy, effectuez une étude complète de $f(x) = \frac{x^3 - 3x + 2}{x^2 - 1}$ :

1. Domaine de définition (points d'annulation du dénominateur)
2. Limites aux bornes du domaine
3. Dérivée et extrema
4. Dérivée seconde et points d'inflexion
5. Asymptotes
6. Tracé avec `sympy.plot`

````python
from sympy import symbols, limit, diff, solve, oo, simplify, plot, factor

x = symbols('x')
f = (x**3 - 3*x + 2) / (x**2 - 1)

# 1. Domaine
denom_zeros = solve(x**2 - 1, x)
print(f'Points interdits : {denom_zeros}')

# 2. Limites
for pt in [oo, -oo] + denom_zeros:
    lim = limit(f, x, pt)
    print(f'lim(x->{pt}) = {lim}')

# 3. Dérivée
f_prime = simplify(diff(f, x))
extrema = solve(f_prime, x)
print(f'f\'(x) = {f_prime}')
print(f'Extrema en x = {extrema}')

# 4. Factorisation
print(f'f factorisée : {factor(f)}')
````
