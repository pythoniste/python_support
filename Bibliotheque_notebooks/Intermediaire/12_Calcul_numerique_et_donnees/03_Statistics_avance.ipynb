## 1. Rappel

````python
from statistics import mean, stdev, median, quantiles

data = [4, 8, 15, 16, 23, 42]
mean(data), stdev(data), median(data)
````

---

## 2. `NormalDist`

````python
from statistics import NormalDist

nd = NormalDist(mu=100, sigma=15)  # QI
nd.cdf(130)  # P(X <= 130)
````

````python
nd.inv_cdf(0.95)  # valeur au 95e percentile
````

````python
nd.overlap(NormalDist(mu=110, sigma=15))  # chevauchement
````

````python
NormalDist.from_samples([4, 8, 15, 16, 23, 42])
````

---

## 3. Quartiles et percentiles

````python
quantiles(data, n=4)  # quartiles
````

````python
quantiles(data, n=10)  # déciles
````

````python
quantiles(data, n=100)  # percentiles
````

---

## 4. Corrélation

````python
from statistics import correlation

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]
correlation(x, y)
````

---

## 5. Régression linéaire

````python
from statistics import linear_regression

slope, intercept = linear_regression(x, y)
print(f'y = {slope:.2f}x + {intercept:.2f}')
````

````python
# Prédiction
for xi in [6, 7, 8]:
    print(f'x={xi} → y={slope * xi + intercept:.2f}')
````

---

## Synthèse

| Fonction | Rôle |
|---|---|
| `NormalDist(mu, sigma)` | Loi normale |
| `.cdf(x)` | Probabilité cumulative |
| `.inv_cdf(p)` | Quantile |
| `quantiles(data, n=4)` | Quartiles |
| `correlation(x, y)` | Corrélation de Pearson |
| `linear_regression(x, y)` | Pente et ordonnée |
