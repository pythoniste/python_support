## 1. Créer une `Fraction`

````python
from fractions import Fraction

Fraction(1, 3)
````

````python
Fraction('2/5')
````

````python
Fraction('0.1')  # exact, contrairement à Fraction(0.1)
````

````python
Fraction(0.1)  # imprécision du float
````

---

## 2. Arithmétique exacte

````python
Fraction(1, 3) + Fraction(1, 6)
````

````python
Fraction(2, 3) * Fraction(3, 4)
````

````python
Fraction(7, 2) // Fraction(3, 2)
````

---

## 3. Conversions

````python
float(Fraction(1, 3))
````

````python
Fraction(1, 3).limit_denominator(100)
````

````python
Fraction.from_float(3.14159)
````

````python
Fraction.from_decimal(Decimal('0.1'))
````

---

## 4. `Fraction` et `Decimal`

On peut mélanger `Fraction` et `int` mais pas directement `Fraction` et `float`.

````python
Fraction(1, 3) + 1  # ok, int promu en Fraction
````

---

## 5. Cas d'usage

`Fraction` est utile pour :

- les **probabilités** (ex. `1/52` pour un jeu de cartes)
- les **recettes de cuisine** (1/3 de tasse)
- les **algorithmes** qui nécessitent l'exactitude sans flottants

---

## Synthèse

| API | Rôle |
|---|---|
| `Fraction(num, den)` | Création |
| `Fraction('1/3')` | Depuis chaîne |
| `+`, `-`, `*`, `/` | Arithmétique exacte |
| `float(f)` | Conversion vers float |
| `.limit_denominator(max)` | Approximation rationelle |


### Règles à retenir

1. **`Fraction` pour l'exactitude** quand `float` est insuffisant.
2. **`Fraction('0.1')` pas `Fraction(0.1)`.** Même règle que `Decimal`.
3. **`Decimal` pour la finance, `Fraction` pour les ratios exacts.**

````python
# Votre code ici
````

<details>
<summary>📖 Voir la correction</summary>

```python
import math
from fractions import Fraction

fpi = Fraction(math.pi)
for max_d in [10, 100, 1000]:
    approx = fpi.limit_denominator(max_d)
```
