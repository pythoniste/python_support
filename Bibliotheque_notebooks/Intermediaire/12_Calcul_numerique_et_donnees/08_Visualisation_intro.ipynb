## 1. Matplotlib OO

Toujours utiliser l'API **orientée objet** (`fig, ax`), jamais `plt.plot()` directement.

````python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(8, 4))
x = np.linspace(0, 2 * np.pi, 100)
ax.plot(x, np.sin(x), label='sin(x)')
ax.plot(x, np.cos(x), label='cos(x)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Fonctions trigonométriques')
ax.legend()
plt.tight_layout()
plt.show()
````

---

## 2. Types de graphiques

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Bar
axes[0].bar(['Mars', 'Venus', 'Io'], [12, 6, 15])
axes[0].set_title('Capacité')

# Scatter
axes[1].scatter(np.random.rand(30), np.random.rand(30))
axes[1].set_title('Scatter')

# Histogram
axes[2].hist(np.random.randn(1000), bins=30)
axes[2].set_title('Distribution')

plt.tight_layout()
plt.show()
````

---

## 3. Personnalisation

````python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(range(10), [x**2 for x in range(10)], 'ro-', linewidth=2, markersize=8)
ax.set_title('Carrés', fontsize=14)
ax.grid(True, alpha=0.3)
plt.show()
````

---

## 4. Seaborn : haut niveau sur matplotlib

````python
import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    'salle': ['Mars']*5 + ['Venus']*5,
    'heures': np.random.randint(1, 8, 10),
})
fig, ax = plt.subplots(figsize=(6, 4))
sns.boxplot(data=df, x='salle', y='heures', ax=ax)
ax.set_title('Heures par salle')
plt.show()
````

---

## 5. Plotly : interactif

````python
import plotly.express as px

fig = px.bar(
    x=['Mars', 'Venus', 'Io'],
    y=[12, 6, 15],
    labels={'x': 'Salle', 'y': 'Capacité'},
    title='Capacité des salles',
)
fig.show()
````

---

## 6. Quand utiliser quoi

| Besoin | Outil |
|---|---|
| Graphique rapide en notebook | seaborn |
| Contrôle total (publication) | matplotlib OO |
| Interactif (dashboard, web) | plotly |
| Gros volume de points | matplotlib avec `rasterized=True` |


---

## Synthèse

| Outil | API | Forces |
|---|---|---|
| `matplotlib` | `fig, ax = plt.subplots()` | Contrôle total |
| `seaborn` | `sns.boxplot(data=df)` | Haut niveau, joli |
| `plotly` | `px.bar(...)` | Interactif |


### Règles à retenir

1. **API OO de matplotlib** : `fig, ax = plt.subplots()`. Pas `plt.plot()`.
2. **Seaborn pour l'exploration**, matplotlib pour la publication.
3. **Plotly pour l'interactivité.**
