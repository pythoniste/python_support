## 1. Pourquoi le calcul formel

Le calcul **numérique** (NumPy, SciPy) manipule des nombres flottants. Le calcul **formel** (SymPy) manipule des **expressions mathématiques exactes** : `x^2 + 2x + 1`, pas `9.00001`.

---

## 2. Symboles et expressions

````python
from sympy import symbols, sqrt, pi, Rational

x, y = symbols('x y')
expr = x**2 + 2*x + 1
expr
````

````python
sqrt(8)
````

````python
Rational(1, 3) + Rational(1, 6)
````

---

## 3. Simplification, développement, factorisation

````python
from sympy import expand, factor, simplify

expand((x + 1)**3)
````

````python
factor(x**3 + 3*x**2 + 3*x + 1)
````

````python
simplify((x**2 - 1) / (x - 1))
````

---

## 4. Dérivées

````python
from sympy import diff, sin, cos, exp

diff(x**3 + 2*x, x)
````

````python
diff(sin(x) * exp(x), x)
````

````python
diff(x**3, x, 2)  # dérivée seconde
````

---

## 5. Intégrales

````python
from sympy import integrate, oo

integrate(x**2, x)  # intégrale indéfinie
````

````python
integrate(x**2, (x, 0, 1))  # intégrale définie
````

````python
integrate(exp(-x**2), (x, -oo, oo))  # intégrale de Gauss
````

---

## 6. Résolution d'équations

````python
from sympy import solve, Eq

solve(x**2 - 4, x)
````

````python
solve(Eq(x**2 + x, 6), x)
````

````python
solve([x + y - 5, x - y - 1], [x, y])
````

---

## 7. Matrices symboliques

````python
from sympy import Matrix

M = Matrix([[1, 2], [3, 4]])
M.det()
````

````python
M.eigenvals()
````

---

## Synthèse

| Fonction | Rôle |
|---|---|
| `symbols('x y')` | Créer des symboles |
| `expand`, `factor`, `simplify` | Transformation d'expressions |
| `diff(expr, x)` | Dérivée |
| `integrate(expr, x)` | Intégrale |
| `solve(eq, x)` | Résolution |


### Règles à retenir

1. **SymPy pour l'exactitude**, NumPy pour le numérique rapide.
2. **`simplify` est coûteux** : utilisez `factor`, `expand`, `trigsimp` quand vous savez quelle forme vous voulez.
