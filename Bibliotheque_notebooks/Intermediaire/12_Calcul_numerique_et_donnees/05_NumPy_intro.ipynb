## 1. Pourquoi NumPy

NumPy est le fondement du calcul numérique en Python. Un `ndarray` est :

- **homogène** (tous les éléments du même type) ;
- **contigu en mémoire** (accès rapide, cache-friendly) ;
- **vectorisé** (opérations sans boucle Python, 10-100x plus rapide).

## 4. Slicing et indexation

````python
a = np.arange(10)
a[3:7]
````

````python
a[::2]  # un sur deux
````

````python
m = np.arange(12).reshape(3, 4)
m
````

````python
m[1, 2]  # ligne 1, colonne 2
````

````python
m[:, 1]  # toute la colonne 1
````

### Indexation booléenne

````python
a = np.array([10, 20, 30, 40, 50])
a[a > 25]
````

````python
a[a % 20 == 0]
````

---

## 5. Opérations vectorisées

Pas besoin de boucle : les opérations s'appliquent **élément par élément**.

````python
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])
a + b
````

````python
a * 2
````

````python
a ** 2
````

````python
np.sqrt(a)
````

---

## 6. Broadcasting

Quand deux arrays n'ont pas la même forme, NumPy **étire** le plus petit.

````python
m = np.ones((3, 4))
v = np.array([1, 2, 3, 4])
m + v
````

````python
m * np.array([[1], [2], [3]])
````

## 8. Reshaping et agrégation

````python
a = np.arange(12)
a.reshape(3, 4)
````

````python
m = np.arange(12).reshape(3, 4)
m.sum(), m.mean(), m.std()
````

````python
m.sum(axis=0)  # somme par colonne
````

````python
m.sum(axis=1)  # somme par ligne
````

---

## Synthèse

| Fonction | Rôle |
|---|---|
| `np.array([...])` | Créer un array |
| `np.zeros`, `ones`, `arange`, `linspace` | Constructeurs |
| `a[mask]` | Indexation booléenne |
| `a + b`, `a * 2` | Opérations vectorisées |
| Broadcasting | Adaptation automatique des formes |
| `.sum()`, `.mean()`, `.std()` | Agrégations |


### Règles à retenir

1. **Vectorisez, ne bouclez pas.** `a * 2` est 100x plus rapide qu'un `for`.
2. **Choisissez le bon dtype** pour économiser la mémoire.
3. **Broadcasting : comprendre les règles** évite des bugs subtils.
