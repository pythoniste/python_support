## 5. `groupby`

````python
reservations = pd.DataFrame({
    'salle': ['Mars', 'Venus', 'Mars', 'Io', 'Mars'],
    'heures': [2, 1, 3, 2, 1],
})
reservations.groupby('salle')['heures'].sum()
````

````python
reservations.groupby('salle').agg({'heures': ['sum', 'mean', 'count']})
````

---

## 6. `merge` (jointures)

````python
pd.merge(reservations, df[['nom', 'prix_h']], left_on='salle', right_on='nom')
````

## Synthèse

| API | Rôle |
|---|---|
| `pd.DataFrame({...})` | Créer un DataFrame |
| `df[col]`, `df.loc[]`, `df.iloc[]` | Sélection |
| `df[condition]`, `df.query()` | Filtrage |
| `df.groupby().agg()` | Agrégation |
| `pd.merge(left, right)` | Jointure |
| `pd.read_csv()` / `to_csv()` | I/O |


### Règles à retenir

1. **Pandas pour l'exploration et la transformation** de données tabulaires.
2. **Vectorisez** : `df['col'] * 2` plutôt qu'une boucle `for row in df.iterrows()`.
3. **`query()` est plus lisible** que la syntaxe `df[df[...]]` pour les filtres complexes.

### Exercice 3 — Merge et transformation *(difficile)*

Deux DataFrames : `commandes(id, produit_id, qte)` et `produits(id, nom, prix)`. Merger, ajouter une colonne `total = qte * prix`, trier par total décroissant.

````python
# Votre code ici
````

<details>
<summary>📖 Voir la correction</summary>

```python
import pandas as pd
produits = pd.DataFrame({'id': [1, 2], 'nom': ['A', 'B'], 'prix': [10.0, 25.0]})
commandes = pd.DataFrame({'id': [1,2,3], 'produit_id': [1,2,1], 'qte': [5,3,10]})
merged = pd.merge(commandes, produits, left_on='produit_id', right_on='id', suffixes=('_cmd', '_prod'))
merged['total'] = merged['qte'] * merged['prix']
print(merged.sort_values('total', ascending=False))
```
