## 1. `random` : pseudo-aléatoire reproductible

````python
import random

random.seed(42)  # reproductible
random.random()
````

````python
random.random()
````

````python
random.seed(42)  # même seed → même séquence
random.random()
````

`random` utilise le Mersenne Twister. **Reproductible** avec une seed, **pas cryptographiquement sûr**.

## 4. `secrets` : aléatoire cryptographique

Pour tout ce qui est lié à la **sécurité** : tokens, mots de passe, identifiants de session.

````python
import secrets

secrets.token_hex(16)  # 16 bytes → 32 car hexadécimaux
````

````python
secrets.token_urlsafe(16)
````

````python
secrets.randbelow(100)  # entier entre 0 et 99
````

````python
secrets.choice(['alpha', 'beta', 'gamma'])
````

---

## 5. Tokens et mots de passe

````python
import string

alphabet = string.ascii_letters + string.digits + string.punctuation
password = ''.join(secrets.choice(alphabet) for _ in range(16))
password
````

## 6. Quand utiliser quoi

| Besoin | Module |
|---|---|
| Simulation, jeu, test | `random` |
| Seed reproductible | `random.seed(n)` |
| Token de session, reset password | `secrets` |
| Clé API, mot de passe | `secrets` |
| Tirage de loto en prod | `secrets` |

**Règle : si c'est lié à la sécurité ou à l'argent, `secrets`. Sinon, `random`.**

---

## Synthèse

| Fonction | Module | Usage |
|---|---|---|
| `random()`, `gauss()`, `uniform()` | `random` | Simulation |
| `choice()`, `sample()`, `shuffle()` | `random` | Sélection aléatoire |
| `token_hex()`, `token_urlsafe()` | `secrets` | Tokens sécurisés |
| `randbelow(n)` | `secrets` | Entier crypto-safe |


### Règles à retenir

1. **`random` n'est PAS sûr** pour les tokens, mots de passe, clés.
2. **`secrets` pour tout ce qui touche à la sécurité.**
3. **`random.seed(n)` pour la reproductibilité** en test.

Écrire `estimer_pi(n: int) -> float` qui estime pi par Monte Carlo : tirer `n` points aléatoires dans [0,1]x[0,1], compter ceux dans le quart de cercle.

````python
# Votre code ici
````

<details>
<summary>📖 Voir la correction</summary>

```python
import random

def estimer_pi(n: int) -> float:
    dans_cercle = sum(1 for _ in range(n) if random.random()**2 + random.random()**2 <= 1)
    return 4 * dans_cercle / n

print(estimer_pi(1_000_000))
```
