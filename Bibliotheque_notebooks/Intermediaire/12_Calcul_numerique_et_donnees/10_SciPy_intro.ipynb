## 2. Optimisation : `minimize`

````python
from scipy.optimize import minimize

def f(x):
    return (x - 3)**2 + 1

result = minimize(f, x0=0)  # point de départ x=0
result.x, result.fun
````

````python
import numpy as np
from scipy.optimize import minimize

def rosenbrock(xy):
    x, y = xy
    return (1 - x)**2 + 100 * (y - x**2)**2

result = minimize(rosenbrock, x0=[0, 0])
result.x
````

---

## 3. Racines : `brentq`

````python
from scipy.optimize import brentq

def g(x):
    return x**3 - 2*x - 5

racine = brentq(g, 1, 3)  # cherche entre 1 et 3
racine, g(racine)
````

---

## 4. Interpolation : `interp1d`

````python
from scipy.interpolate import interp1d
import numpy as np

x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 1, 4, 9, 16])

f_interp = interp1d(x, y, kind='quadratic')
f_interp(2.5)
````

---

## 5. Intégration numérique : `quad`

````python
from scipy.integrate import quad
import numpy as np

result, error = quad(np.sin, 0, np.pi)
result, error
````

````python
result, _ = quad(lambda x: np.exp(-x**2), -np.inf, np.inf)
result, np.sqrt(np.pi)
````

## Synthèse

| Module | Fonction | Rôle |
|---|---|---|
| `optimize` | `minimize` | Minimum d'une fonction |
| `optimize` | `brentq` | Racine (zéro) |
| `interpolate` | `interp1d` | Interpolation |
| `integrate` | `quad` | Intégrale numérique |


### Règles à retenir

1. **SciPy pour le numérique**, SymPy pour l'exact.
2. **`minimize` prend un point de départ** : le résultat peut dépendre du choix initial.
3. **`quad` est très précis** pour les intégrales 1D.
