## 1. PEP 8 : pourquoi ?

**PEP 8** est le guide de style officiel pour the code Python. C'est la **lingua franca** of the communauté : en le following, votre code ressemble à celui de tout le monde, ce qui facilite sa lecture par n'importe qui.

Extrait fameux de PEP 8 : *« A Foolish Consistency is the Hobgoblin of Little Minds »* — ne transformez pas la cohérence en obsession : lisibilité d'abord.

## 2. Indentation et longueur de ligne

- **Indentation : 4 espaces**. Jamais de tabulations.
- **Longueur de ligne : 79 becauseactères** (ou 99 for thes bases de code modernes, ou 88 comme `black`/`ruff`).

````python
def resume(values: list[int]) -> dict[str, float]:
    """Renvoie un résumé statistique."""
    return {
        'min': float(min(values)),
        'max': float(max(values)),
        'average': sum(values) / len(values),
    }
````

- **2 lignes blanches** between définitions de functions/classs au niveau module ;
- **1 ligne blanche** between methods dans une class ;
- **1 ligne blanche** pour séparer les blocs logiques dans une function.

## 4. Nommage

| Type | Convention | Example |
|---|---|---|
| variable / function | `snake_case` | `compter_mots`, `prix_ht` |
| constante | `UPPER_SNAKE_CASE` | `TAUX_TVA`, `PI` |
| class | `PascalCase` | `Compte`, `HttpClient` |
| privé | `_prefixe` | `_helper`, `_cache` |
| évitez | `l`, `O`, `I` | (confusion avec 1 et 0) |

````python
# ❌ mauvais
def CalculerPrix(PrixHT, Taux):
    return PrixHT + PrixHT * Taux

# ✅ bon
TAUX_TVA = 0.20

def calculer_ttc(prix_ht: float, taux: float = TAUX_TVA) -> float:
    return prix_ht + prix_ht * taux
````

## 7. Docstrings — style Google

Each function publique a une **docstring**. Le style Google est lisible, simple et bien supporté par les outils (Sphinx, VS Code, Ruff).

````python
def retirer(solde: float, montant: float) -> float:
    """Retire un montant d'un solde.

    Args:
        solde: solde actuel (en euros).
        montant: montant à retirer (positif).

    Returns:
        Le new_val solde après retrait.

    Raises:
        ValueError: si `montant` est négatif ou dépasse le solde.
    """
    if montant < 0 or montant > solde:
        raise ValueError('montant invalide')
    return solde - montant
````

````python
help(retirer)
````

### Sections usuelles

| Section | Quand |
|---|---|
| `Args:` | list des parameters |
| `Returns:` | description du retour |
| `Raises:` | exceptions levées |
| `Yields:` | generators |
| `Examples:` | code démonstratif |

### Docstring avec `Examples`

````python
def square(x: int) -> int:
    """Renvoie le square de x.

    Args:
        x: un integer.

    Returns:
        x * x.

    Examples:
        >>> square(5)
        25
        >>> square(-3)
        9
    """
    return x * x

help(square)
````

Le bloc `Examples` suit le style `doctest` : les lignes `>>>` peuvent être exécutées automatiquement par `pytest --doctest-modules`.

### Commentaires vs docstrings

- **Commentaire** `# ...` : explique *pourquoi*, pas *quoi*. Pour les lecteurs du code.
- **Docstring** `"""..."""` : explique *quoi* et *comment l'usesr*. Pour les utilisateurs of the function.

## Cheat sheet — PEP 8 en 10 lignes

| Règle | Valeur |
|---|---|
| Indentation | 4 espaces |
| Longueur de ligne | 88 (ruff) ou 79 (PEP 8 strict) |
| Lignes blanches between functions top-level | 2 |
| Lignes blanches between methods | 1 |
| Variable / function | `snake_case` |
| Constante | `UPPER_SNAKE_CASE` |
| Class | `PascalCase` |
| Privé | `_prefixe` |
| Import par ligne | 1 |
| Espaces autour de `=` by default d'arg | **pas d'espace** |

````python
# Example PEP 8-compliant minimal
MAX_TENTATIVES = 3


def saluer(name: str, politesse: str = 'Bonjour') -> str:
    """Construit une salutation."""
    return f'{politesse} {name} !'


class Utilisateur:
    """Un utilisateur simple."""

    def __init__(self, name: str) -> None:
        self.name = name


print(saluer('Alice'))
````

### Checklist docstring

- [ ] Première ligne : une phrase impérative (« Renvoie…», « Calcule…»)
- [ ] Section `Args:` si parameters
- [ ] Section `Returns:` sauf si `-> None`
- [ ] Section `Raises:` si la functiwe can lever une exception
- [ ] `Examples:` si l'usage n'est pas évident
