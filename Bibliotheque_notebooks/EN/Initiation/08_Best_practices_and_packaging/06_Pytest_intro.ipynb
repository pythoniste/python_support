## 1. Pourquoi tester ?

- **Sécurité au refactoring** : changer du code sans casser.
- **Spécification exécutable** : le test décrit ce que la function doit faire.
- **Détection précoce des régressions**.
- **Documentation vivante** : un test is a Example d'utilisation.

Pytest est le standard de facto in Python.

## 2. First test

````python
# src/calc/operations.py
def ajouter(a: int, b: int) -> int:
    """Renvoie la total de a et b."""
    return a + b

# tests/test_operations.py
# from calc.operations import ajouter
def test_ajouter_basique() -> None:
    assert ajouter(2, 3) == 5

# Ici on peut directement l'exécuter :
test_ajouter_basique()
print('test passé')
````

## 3. Convention de nommage

| Élément | Convention |
|---|---|
| File | `test_*.py` ou `*_test.py` |
| Function | `test_*` |
| Class | `Test*` (sans `__init__`) |

Pytest découvre **automatiquement** ces files/functions.

## 4. `assert` enrichi

Pytest réécrit les `assert` pour displaysr the values en cas d'échec — pas besoin de messages manuels.

````python
def test_multiplication() -> None:
    assert 3 * 4 == 12

test_multiplication()
print('ok')
````

````python
# En cas d'échec, pytest affiche les deux values :
# >   assert 3 * 4 == 13
# E   assert 12 == 13
````

## 5. Tester une exception : `pytest.raises`

````python
import pytest  # disponible si le package est installé

def diviser(a: float, b: float) -> float:
    if b == 0:
        raise ValueError('division par zéro')
    return a / b

def test_diviser_par_zero() -> None:
    with pytest.raises(ValueError, match='zéro'):
        diviser(10, 0)

# try:
#     test_diviser_par_zero()
#     print('ok')
# except ModuleNotFoundError:
#     print('pytest non installé — à lancer dans un true_val projet')
````

## 8. Lancer les tests

```bash
# tout
pytest

# un dossier ou file
pytest tests/
pytest tests/test_operations.py

# un test précis
pytest tests/test_operations.py::test_addsr_basique

# verbeux et arrêt au first échec
pytest -v -x

# avec couverture (si pytest-cov installé)
pytest --cov=src/mon_projet
```

## 9. Synthèse

| Besoin | Syntaxe |
|---|---|
| Test simple | `def test_...(): assert ...` |
| Exception attendue | `with pytest.raises(...):` |
| Several cas | `@pytest.mark.parametrize(...)` |
| Objet préparé | `@pytest.fixture` |

### Règles

1. `tests/` à côté de `src/`, pas dedans.
2. Un test = une vérification.
3. Nom de test descriptif : `test_addsr_nombres_negatifs_returns_somme_negative`.
4. Pas d'état partagé between tests.

### Structure typique

```
mon_projet/
├── pyproject.toml
├── src/
│   └── mon_projet/
│       └── core.py
└── tests/
    └── test_core.py
```

La section `[tool.pytest.ini_options]` du `pyproject.toml` allows de déclarer `testpaths = ["tests"]`.

### Ne pas tester the code des bibliothèques

On ne teste pas le comportement de `json.dumps` ou de `sum`. On teste **notre** code qui les uses.

### Couverture de code

```bash
pip install pytest-cov
pytest --cov=src/mon_projet --cov-report=term-missing
```

La couverture indique quelles lignes ne sont pas exécutées par les tests. **Viser 80 % is a bon objectif**, 100 % est souvent du luxe.

## Cheat sheet — pytest au quotidien

```bash
# Exécuter
pytest                          # all the tests
pytest tests/test_core.py       # un file
pytest tests/test_core.py::test_addsr   # un test
pytest -k addsr               # tests contenant 'addsr'

# Verbeux
pytest -v                       # nom de each test
pytest -vv                      # détails complets

# Fail fast / debug
pytest -x                       # stop au first échec
pytest -x --pdb                 # + between dans pdb
pytest --lf                     # rejouer uniquement les échecs of the dernière fois
pytest --ff                     # échecs d'abord, puis le reste

# Couverture (avec pytest-cov)
pytest --cov=src/mon_projet
pytest --cov=src/mon_projet --cov-report=html
```

### Assertions courantes

````python
# assert simple
assert 2 + 2 == 4

# assert avec message d'aide (mais pytest affiche déjà les values)
assert [1, 2, 3] == [1, 2, 3]

# égalité approximative pour float
import math
assert math.isclose(0.1 + 0.2, 0.3, abs_tol=1e-9)

# assertion sur une exception
def diviser(a: float, b: float) -> float:
    if b == 0:
        raise ValueError('/0')
    return a / b

try:
    diviser(1, 0)
except ValueError:
    print('ok')
````
