## 1. Qu'est-ce qu'une variable ?

Une **variable** is a **nom** qu'on donne à une valeur, pour pouvoir la réusesr et la modifiesr.

Analogie : une étiquette collée sur une boîte. La boîte contains une valeur, l'étiquette allows of the retrouver.

````python
age = 30
````

````python
print(age)
````

Une fois déclarée, la variable peut être utilisée to the place de the value.

````python
annee = 2026
````

````python
print(annee)
````

We can **réaffecter** une variable : elle prend une nouvelle valeur.

````python
age = 30
````

````python
age = 31
````

````python
print(age)
````

Python allows d'affecter several variables en une ligne :

Ou de déballer several valeurs to the fois :

````python
a, b = 10, 20
````

````python
print(a, b)
````

## 10. Règles de nommage

In Python, un **nom de variable** doit :

1. **commencer par une lettre** ou un `_` (pas par un chiffre) ;
2. ne contenir que des **lettres, chiffres, underscores** ;
3. **ne pas être un mot-clé** Python (`if`, `for`, `class`, ...) ;
4. respecter la casse : `age` et `Age` sont **deux variables différentes**.


### Convention PEP 8 — snake_case

Pour les variables et functions, on écrit en **snake_case** (minuscules, mots séparés par des underscores) :

| ❌ À éviter | ✅ Recommandé |
|---|---|
| `monAge` | `mon_age` |
| `NombreDeLignes` | `nombre_de_lignes` |
| `x1` | `first_x` (si le contexte s'y prête) |

Les constantes s'écrivent en `MAJUSCULES_AVEC_UNDERSCORES`.


````python
mon_age = 30
````

````python
nombre_maximum = 100
````

````python
PI = 3.14159
````

### Mots-clés réservés

## 11. Synthèse

### Les cinq types primitifs

| Type | Example | `type(...)` |
|---|---|---|
| Integer | `42`, `-7`, `1_000_000` | `int` |
| Float | `3.14`, `-0.5`, `1.5e9` | `float` |
| Chaîne | `"bonjour"`, `'hi'` | `str` |
| Boolean | `True`, `False` | `bool` |
| Absence | `None` | `NoneType` |

### Règles Key takeaways

1. **`=` is ae affectation, pas une égalité.** L'égalité se teste avec `==` *(prochain notebook)*.
2. **On compares à `None` avec `is`**, pas `==`.
3. **`type(x)` répond « c'est quoi ? »**, `isinstance(x, T)` répond « c'en is a ? » *(vu plus tard)*.
4. **Les integers Python ont une précision illimitée.**
5. **`int("3.14")` lève une erreur** — il faut passer par `float(...)` d'abord si besoin.
6. **snake_case** for thes noms de variables, `MAJUSCULES` for thes constantes.
