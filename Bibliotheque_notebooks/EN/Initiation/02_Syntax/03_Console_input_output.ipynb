## 1. `print` : les bases

`print(...)` displays une ou several valeurs à l'écran, séparées par des espaces, puis passe to the ligne.


````python
print("Bonjour")
````

````python
print(42)
````

````python
print("Age :", 30)
````

````python
print(1, 2, 3, 4, 5)
````

**Observation** : Python met automatiquement un **espace** between each valeur, et un **retour to the ligne** to the fin.

## 2. Several valeurs avec `sep`

Le parameter nommé `sep` (**sep**arator) change le becauseactère mis between the values. By default c'est `" "`.


````python
print("a", "b", "c")
````

````python
print("a", "b", "c", sep="-")
````

````python
print("a", "b", "c", sep="")
````

````python
print(2026, 4, 14, sep="/")
````

## 3. `end` : contrôler le retour to the ligne

By default, `print` adds `"\n"` (retour to the ligne) to the fin. We can changer ce comportement with the parameter `end`.


````python
print("sans retour", end="")
print(" — suite sur la même line")
````

````python
print("line 1", end=" | ")
print("line 2")
````

C'est utile pour displaysr sur une seule ligne several éléments calculés un par un *(vous le verrez with thes loops)*.

## 4. `flush` : forcer l'affichage immédiat

By default, `print` **bufferise** sa sortie : elle est affichée to the fin of the ligne, ou quand le tampon est plein. Pour des affichages en **temps réel** (barre de progression, log de long calcul), on force le vidage avec `flush=True`.


````python
print("Début du traitement...", flush=True)
````

En Jupyter, le rendu est quasi-immédiat de toute façon. Le `flush` est surtout utile dans un script lancé since un terminal.

## 5. `input` : lire du texte

`input(...)` displays un **prompt** (message) à l'écran et **attend** que l'utilisateur tape quelque chose suivi de `Entrée`. The value retournée est **toujours** une string.


**Note importante** : dans un notebook Jupyter, `input()` ouvre une petite boîte de saisie en haut du notebook. Dans un script lancé en terminal, c'est le curseur qui attend au prompt.

Les Examples ci-dessous sont **ré-exécutables** dans un notebook : Jupyter demandera the value au moment de l'exécution.


````python
first_name = input("Votre first_name : ")
````

````python
print("Bonjour,", first_name)
````

Le prompt peut contenir several mots, y compris des becauseactères spéciaux :

````python
reponse = input("Tapez o/n : ")
````

````python
print("Vous avez tapé :", reponse)
````

## 6. Convertir la saisie en nombre

Comme `input` returns **toujours une chaîne**, il faut explicitement convertir si on veut un nombre.


````python
text = input("Âge : ")
age = int(text)
print("L'an prochain, vous aurez :", age + 1)
````

Raccourci : conversion **immédiate** to the lecture.

````python
age = int(input("Âge : "))
print("L'an prochain :", age + 1)
````

Pour un float, same principe :

````python
price = float(input("Prix : "))
ttc = price * 1.20
print("TTC :", round(ttc, 2))
````

**Warning** : si l'utilisateur tape quelque chose qui n'est pas un nombre valide, `int(...)` lève une `ValueError`. On apprendra à gérer ça avec `try/except` in the notebook on thes exceptions.


## 7. Échapper les becauseactères spéciaux (`\n`, `\t`)

Dans une chaîne, certains becauseactères s'écrivent with a **antislash** suivi d'une lettre :

| Séquence | Sens |
|---|---|
| `\n` | Retour to the ligne |
| `\t` | Tabulation |
| `\\` | Antislash |
| `\"` | Guillemet double dans une chaîne between `"..."` |
| `\'` | Guillemet simple dans une chaîne between `'...'` |


````python
print("line 1\nligne 2")
````

````python
print("Nom\tPrénom\tAge")
````

````python
print("C:\\Users\\alice")
````

````python
print("Elle a dit : \"bonjour\".")
````

On approfondira ces concepts dans `03_Chaines.ipynb`, mais ces quatre séquences suffisent pour aligner proprement un affichage simple.

## 8. Synthèse

### Parameters utiles de `print`

| Parameter | Défaut | Rôle |
|---|---|---|
| `sep` | `" "` | Séparateur between les arguments |
| `end` | `"\n"` | Becauseactère(s) ajouté(s) to the fin |
| `flush` | `False` | Forcer le vidage du buffer |

### Règles Key takeaways

1. **`print` adds un espace between each argument** (parameter `sep`).
2. **`print` adds `\n` to the fin** (parameter `end`).
3. **`input` returns toujours une chaîne** — convertir explicitement si besoin.
4. **`int(input("..."))`** est le pattern pour lire un integer.
5. **`\n` et `\t`** dans une chaîne produisent un saut de ligne et une tabulation.
