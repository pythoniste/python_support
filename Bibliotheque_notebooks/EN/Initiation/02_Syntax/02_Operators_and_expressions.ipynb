## 1. Operators arithmétiques

Python supporte les operators classiques plus trois spécifiques : `//`, `%` et `**`.


````python
4 + 2
````

````python
4 - 2
````

````python
4 * 2
````

````python
4 / 2
````

**Warning** : `/` returns **toujours** un `float`, same between deux integers. `4 / 2` vaut `2.0`, pas `2`.

````python
type(4 / 2)
````

Pour obtenir une division **entière**, usesr `//` :

````python
7 // 2
````

Le résultat de `//` between deux integers is a integer. Between un integer et un float, c'is a float.

````python
7.0 // 2
````

L'operator **modulo** `%` donne le reste of the division :

````python
7 % 3
````

Il sert souvent à tester la divisibilité :

````python
10 % 2
````

````python
11 % 2
````

`10 % 2 == 0` veut dire « 10 est divisible par 2 », therefore pair.

L'operator **puissance** `**` :

````python
2 ** 10
````

````python
3 ** 4
````

Il s'applique also aux floats :

````python
2 ** 0.5
````

`2 ** 0.5` est la racine becauserée de 2.

L'operator unaire `-` change le signe :

````python
-5
````

````python
x = 3
-x
````

## 2. Priorité et parenthèses

Comme en mathématiques, Python évalue les operators dans un **ordre de priorité**. Du plus prioritaire au moins prioritaire :


| Priorité | Operators |
|---|---|
| Plus haute | `**` |
|  | `-x` (unaire), `+x` |
|  | `*`, `/`, `//`, `%` |
|  | `+`, `-` (binaires) |
|  | `<`, `<=`, `>`, `>=`, `==`, `!=` |
|  | `not` |
|  | `and` |
| Plus basse | `or` |

**En pratique** : usesz des **parenthèses** pour être explicite. Personne ne vous reprochera des parenthèses en trop, mais on vous reprochera une expression ambiguë.


````python
2 + 3 * 4
````

`*` est prioritaire sur `+`, therefore `2 + 3 * 4` = `2 + 12` = `14`.

````python
(2 + 3) * 4
````

Avec parenthèses, on force l'addition en first.

````python
2 ** 3 ** 2
````

`**` est **associatif à droite** : `2 ** 3 ** 2` = `2 ** (3 ** 2)` = `2 ** 9` = `512`. C'is a Common pitfall.

````python
(2 ** 3) ** 2
````

## 3. Operators de comparaison

Une comparaison returns toujours un **boolean** (`True` ou `False`).


````python
3 < 5
````

````python
3 == 3
````

````python
3 != 4
````

````python
5 >= 5
````

````python
5 <= 4
````

### Comparaisons chaînées

Python autorise une écriture naturelle : `a < b < c` signifie `a < b and b < c`.

````python
x = 5
0 < x < 10
````

````python
y = 50
0 <= y <= 100
````

### Égalité `==` vs identité `is`

`==` compares les **valeurs**. `is` compares les **identités** en mémoire. Pour the values, usesr `==`. Pour `None`, `True`, `False`, usesr `is`.

````python
3 == 3.0
````

````python
3 is 3.0
````

````python
x = None
x is None
````

## 4. Operators logiques `and`, `or`, `not`

Python uses des **mots-clés** (`and`, `or`, `not`), pas des symboles (`&&`, `||`, `!`).


````python
True and True
````

````python
True and False
````

````python
True or False
````

````python
False or False
````

````python
not True
````

````python
not False
````

En combinaison avec des comparaisons :

````python
age = 20
age >= 18 and age < 65
````

````python
jour = 3
jour == 6 or jour == 0
````

````python
actif = False
not actif
````

## 5. Évaluation par court-circuit

Python évalue `and` et `or` **de gauche à droite** et s'arrête dès qu'il connaît le résultat :

- `False and ...` → `False` *(on n'évalue same pas le membre de droite)*.
- `True or ...` → `True` *(idem)*.

C'est ce qu'on appelle le **court-circuit**.


````python
False and (1 / 0 == 0)
````

Normalement `1 / 0` lève une erreur. Ici, Python ne l'évalue jamais parce que the result is déjà déterminé par `False`.

````python
True or (1 / 0 == 0)
````

Ce mécanisme est utilisé en pratique pour **protéger** des calculs risqués :

````python
x = 0
x != 0 and (10 / x > 1)
````

## 6. Affectations combinées (`+=`, `-=`, ...)

Python propose des raccourcis pour modifiesr une variable « en place » :


````python
counter = 0
counter += 1   # équivalent à counter = counter + 1
print(counter)
````

````python
total = 100
total -= 10
print(total)
````

````python
n = 5
n *= 3
print(n)
````

````python
n = 20
n //= 3
print(n)
````

All the operators arithmétiques ont une version combinée :

| Operator | Équivalent |
|---|---|
| `x += y` | `x = x + y` |
| `x -= y` | `x = x - y` |
| `x *= y` | `x = x * y` |
| `x /= y` | `x = x / y` |
| `x //= y` | `x = x // y` |
| `x %= y` | `x = x % y` |
| `x **= y` | `x = x ** y` |


## 7. Concaténation de chaînes

Sur les chaînes, `+` **colle** deux chaînes set, et `*` **répète** une chaîne.


````python
"bon" + "jour"
````

````python
"abc" * 3
````

**Warning** : `+` between une chaîne et un nombre lève une erreur. Il faut convertir explicitement :

````python
"Age : " + str(30)
````

## 8. Synthèse

### Priorité des operators (mémo court)

1. `**`
2. `-x` (unaire)
3. `*`, `/`, `//`, `%`
4. `+`, `-`
5. Comparaisons (`<`, `==`, ...)
6. `not`
7. `and`
8. `or`

### Règles Key takeaways

1. **`/` returns toujours un `float`** — usesr `//` pour l'integer.
2. **`a % b == 0`** teste la divisibilité.
3. **Les comparaisons returnsnt `True`/`False`**, directement stockables dans une variable.
4. **`and`/`or`** se court-circuitent — utile pour protéger un calcul.
5. **`==` pour the values, `is` pour `None`.**
6. **En cas de doute, usesr des parenthèses.**
