## 1. Method : isoler, reproduire, formuler

Avant tout outil, trois étapes mentales :

1. **Isoler** : quel est le plus petit morceau de code qui déclenche le bug ?
2. **Reproduire** : le faire apparaître *to the demande*, pas par hasard.
3. **Formuler** : décrire l'ébecauset between *ce qu'on attend* et *ce qu'on observe*.

Un bug bien formulé est à moitié résolu.

---


## 2. Print-debug

La method la plus rapide : addsr des `print` aux endroits clés, exécuter, comparesr, retirer.

````python
def fact(n: int) -> int:
    result = 1
    for i in range(1, n + 1):
        print('étape', i, 'result intermédiaire', result)
        result = result * i
    return result

print('final :', fact(5))
````

### Limites du print-debug

- on oublie des `print` qui traînent dans the code de prod ;
- on ne peut pas inspecter librement l'état *au moment* du bug ;
- on doit relancer le programme à each hypothèse.

Pour un bug simple, c'est parfait. Pour un bug coriace, passer au débogueur.

## 3. `f"{var=}"` — Tip Python 3.8+

L'écriture `f"{var=}"` displays to the fois le **nom** et la **valeur** of the variable. C'est la forme idéale pour un print de debug.

````python
x = 42
name = 'Alice'
print(f'{x=}')
print(f'{name=}')
print(f'{x + 1=}')
````

## 4. `breakpoint()` — le point d'arrêt universel

Since Python 3.7, `breakpoint()` est **la** manière recommandée d'insérer un point d'arrêt. Il invoque `pdb` by default (ou un other débogueur selon `PYTHONBREAKPOINT`).

### Example (ne pas exécuter dans un notebook automatique)

```python
def fact(n: int) -> int:
    resultat = 1
    for i in range(1, n + 1):
        breakpoint()   # on s'arrête ici
        resultat = resultat * i
    return resultat
```

À l'arrêt, Python ouvre un prompt `(Pdb) `. You can inspecter `i`, `resultat`, modifiesr des variables, continuer pas à pas.

## 7. Quand passer à `logging`

`print` et `breakpoint` are outils de **dev**. Dans un programme qui tourne en production, on uses `logging` — plus contrôlable (niveaux, files, format). Nous le voyons au chapitre Best practices (J5).

## 8. Synthèse

| Outil | Quand |
|---|---|
| `print(f'{var=}')` | Petit bug rapide |
| `breakpoint()` | Inspection interactive |
| `%debug` (Jupyter) | Post-mortem d'une exception |
| `logging` | Programme qui tourne en prod |

### Règles

1. Retirer les `print` de debug avant de commiter.
2. `breakpoint()` plutôt que `import pdb; pdb.set_trace()`.
3. Un bug est toujours plus rapide à résoudre with a test qui le reproduit.

    """Renvoie le maximum d'une list non vide."""

    """Ajoute `element` à `tags` (creates une nouvelle list si None)."""

    """Recherche dichotomique, returns l'indice ou -1."""

### `repr` vs `str`

````python
x = 'coucou'
print(str(x))   # pour l'utilisateur
print(repr(x))  # pour le debug, garde les guillemets
````

`repr` révèle les becauseactères invisibles — `\n`, `\t`, etc. Très utile quand un bug vient d'un espace insécable.

````python
mysterieux = 'hello\u00a0world'  # espace insécable
print(mysterieux)
print(repr(mysterieux))
````
