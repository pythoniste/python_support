## 1. Un module, c'est quoi exactement ?

Un **module** is a file `.py`. Un **package** is a dossier contenant un `__init__.py` (ou considéré comme un package namespace since Python 3.3).

Le nom du module = le nom du file sans `.py`.

````python
import math
math.sqrt(16)
````

````python
from math import sqrt
sqrt(16)
````

## 3. Absolu vs relatif

- **Import absolu** : chemin complet since la racine du projet — `from mon_projet.utils import helper`.
- **Import relatif** : since le module courant — `from .utils import helper`.

**Convention du cours :** on uses **uniquement des imports absolus**. Ils sont plus lisibles et résistent aux refactorings.

---


## 4. Structure d'un petit projet

```
mon_projet/
├── pyproject.toml
├── README.md
├── src/
│   └── mon_projet/
│       ├── __init__.py
│       ├── cli.py
│       ├── core.py
│       └── utils.py
└── tests/
    ├── __init__.py
    ├── test_core.py
    └── test_utils.py
```

### Le « src layout »

Placer le package dans `src/` évite les imports accidentels quand on lance Python since la racine. C'est la structure **recommandée** par la Python Packaging Authority (PyPA).

---


## 5. `__name__ == '__main__'`

Each module a une variable spéciale `__name__`. Elle vaut :

- `'__main__'` si le module est exécuté directement (`python mon_file.py`) ;
- le nom du module sinon (quand il est importé).

````python
print(__name__)  # 'main' dans certains runtimes, '__main__' dans un script
````

### Idiome classique

```python
def principal() -> None:
    print('démarrage...')

if __name__ == '__main__':
    principal()
```

Cela allows d'avoir du code **exécuté** si on lance le file, mais **pas** s'il est juste importé.

---


## 6. `sys.path`

Pour trouver un module, Python parcourt la list `sys.path`. Elle contains :

1. le dossier du script exécuté ;
2. le contenu de `PYTHONPATH` ;
3. les dossiers standard of the distribution (stdlib, site-packages).

Un module installé via `pip`/`uv` atterrit dans `site-packages`, accessible since n'importe où.

````python
import sys
sys.path[:3]
````

---


## 7. Pièges d'imports

### Imports circulaires

Si `a.py` importe `b.py` et `b.py` importe `a.py`, Pythwe can se retrouver à n'importer qu'à moitié l'un des deux modules. **Solution :** déplacer the code partagé dans un troisième module.

### Écraser un nom builtin

Ne nommez pas un file `random.py` ou `math.py` dans votre projet — il écrasera le module standard.

---


## 8. Synthèse

| Élément | Rôle |
|---|---|
| `module.py` | File .py = un module |
| `package/__init__.py` | Dossier = package |
| `import module` | Import d'un module complet |
| `from module import x` | Import sélectif |
| `as` | Alias |
| `__name__` | `'__main__'` si exécuté directement |
| `sys.path` | List des dossiers de recherche |

### Règles

1. Imports **absolus** uniquement.
2. Un seul **module** par file, un seul **sujet** par module.
3. `src/` layout pour tout projet sérieux.
4. `if __name__ == '__main__':` for the point d'entrée CLI.
