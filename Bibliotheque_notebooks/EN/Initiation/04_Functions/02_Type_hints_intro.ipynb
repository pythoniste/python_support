## 1. À quoi ça sert ?

Les *type hints* addsnt à Python une **documentation structurée** on thes types attendus et renvoyés. Ils servent à :

- aider l'**éditeur** (autocomplétion, détection d'erreurs) ;
- aider le **lecteur** du code ;
- allowstre à un outil comme `mypy` de **checksr statiquement** la cohérence ;
- rendre the code plus facile à maintenir on the durée.

À partir de maintenant, **all the Exercises sont typés**.

## 3. Types composites modernes (Python 3.9+)

Since Python 3.9, on uses directement `list`, `dict`, `tuple`, `set` avec des crochets. Plus besoin d'importer `List`, `Dict`, etc. since `typing`.

````python
def total(values: list[int]) -> int:
    total = 0
    for v in values:
        total = total + v
    return total
````

````python
total([1, 2, 3, 4])
````

````python
def compter(words: list[str]) -> dict[str, int]:
    result: dict[str, int] = {}
    for word in words:
        result[word] = result.get(word, 0) + 1
    return result
````

````python
compter(['a', 'b', 'a', 'c', 'b', 'a'])
````

## 4. Optionnels : `int | None` (Python 3.10+)

On écrit un type « ou l'other » avec `|`. Très utile pour un parameter qui peut être `None`.

````python
def age_ou_inconnu(age: int | None) -> str:
    if age is None:
        return 'age inconnu'
    return f'{age} ans'
````

````python
age_ou_inconnu(30)
````

````python
age_ou_inconnu(None)
````

### Équivalent historique avec `Optional[int]`

Vous rencontrerez encore `from typing import Optional` dans du code old. `Optional[int]` est exactement `int | None`. Préférez la forme `|` dans the code neuf.

## 5. Les hints ne sont **pas** exécutés

Python n'impose **rien** à l'exécution. Si vous donnez un `str` à une function qui attend un `int`, Python ne lève pas d'erreur de type — il tente le calcul, qui peut réussir ou échouer.

````python
def double(x: int) -> int:
    return x * 2

# Ça marche, même si ce n'est pas un int
double('ab')
````

C'est pour ça qu'un **vérificateur statique** comme `mypy` est utile.

---


## 6. `mypy` et `mypy --strict`

`mypy` analyse votre code **sans l'exécuter** et signale les incohérences. Le mode `--strict` exige que **toutes** les functions soient annotées.

### Installation

```bash
uv add --dev mypy
uv run mypy --strict mon_file.py
```

Dans un notebook on ne lance pas `mypy` directement, mais la **règle du cours** est que tout Exercise à partir d'ici doit être **`mypy --strict` vert**.

## 8. Synthèse

| Intention | Annotation |
|---|---|
| Integer | `int` |
| Float | `float` |
| Chaîne | `str` |
| Boolean | `bool` |
| Rien | `None` |
| List d'integers | `list[int]` |
| Dict chaîne→integer | `dict[str, int]` |
| Tuple taille fixe | `tuple[float, float]` |
| Tuple variable homogène | `tuple[int, ...]` |
| Set | `set[str]` |
| Optionnel | `int \| None` |

### Règles

1. Toute function publique est annotée parameters + retour.
2. `-> None` quand la function ne returns rien utile.
3. Préférer `int | None` à `Optional[int]` (Python 3.10+).
4. Préférer `list[int]` à `List[int]` (Python 3.9+).

### Lists et dicts annotés

````python
def doubler(values: list[int]) -> list[int]:
    return [v * 2 for v in values]

doubler([1, 2, 3])
````

````python
def inverser_map(d: dict[str, int]) -> dict[int, str]:
    return {v: k for k, v in d.items()}

inverser_map({'a': 1, 'b': 2})
````

### `set` et `tuple`

````python
def ensemble_mots(sentence: str) -> set[str]:
    return set(sentence.split())

ensemble_mots('le chat le chien')
````

````python
def extremes(values: list[int]) -> tuple[int, int]:
    return (min(values), max(values))

extremes([4, 1, 9, 2])
````

### Optionnel

````python
def premier_element(values: list[int]) -> int | None:
    if len(values) == 0:
        return None
    return values[0]

print(premier_element([]))
print(premier_element([1, 2, 3]))
````

### `Any` — porte de sortie (à éviter en général)

````python
from typing import Any

def echo(obj: Any) -> Any:
    return obj

print(echo(42))
print(echo('coucou'))
````

`Any` désactive la vérification de type pour ce nom. À réserver aux frontières du programme (I/O externe, `json.load`, etc.).
