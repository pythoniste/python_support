## 1. Rappel : `True`, `False`, `bool`

Python n'a que **deux valeurs booleannes** : `True` et `False` (toujours with ae majuscule).


````python
print(True)
````

````python
print(False)
````

**Fait amusant** : `bool` is ae **sous-class** de `int`. `True` équivaut à `1`, `False` équivaut à `0` dans un contexte numérique.

## 3. Truthiness : valeurs « vraies » et « fausses »

In Python, **toute valeur** peut être évaluée en contexte boolean (dans un `if`, un `while`, un `and`, etc.). The values considérées **fausses** sont :

- `False`
- `None`
- `0`, `0.0`, `0j`
- chaîne vide `""`
- conteneurs vides `[]`, `()`, `{}`, `set()` *(à connaître pour plus tard)*

**Tout le reste est vrai.**


````python
x = 0
if x:
    print("x est true_val")
else:
    print("x est false_val")
````

````python
x = 42
if x:
    print("x est true_val")
````

````python
name = ""
if name:
    print("name défini :", name)
else:
    print("name empty")
````

````python
name = "Alice"
if name:
    print("name défini :", name)
````

### Idiome : tester la présence d'une valeur

Au lieu d'écrire `if nom != "":`, on écrit simplement `if nom:`. C'est plus concis et plus pythonique.


## 4. `bool(x)` — tester la véracité explicitement

`bool(x)` returns `True` ou `False` selon la véracité de `x`. On s'en sert surtout pour **débugger**, rarement en production.


````python
print(bool(0))
````

````python
print(bool(1))
````

````python
print(bool(-5))
````

````python
print(bool(0.0))
````

````python
print(bool(""))
````

````python
print(bool("non"))
````

````python
print(bool(None))
````

## 5. Operators de court-circuit : `and` et `or` returnsnt des valeurs

**Détail important** : in Python, `and` et `or` ne returnsnt pas forcément `True`/`False`. Ils returnsnt **l'opérande qui a permis de décider**.


````python
print(3 and 5)
````

````python
print(0 and 5)
````

````python
print(3 or 5)
````

````python
print(0 or 5)
````

````python
print("" or "défaut")
````

### Règles précises

| Expression | Évaluation | Résultat |
|---|---|---|
| `a and b` | si `a` faux, s'arrête à `a` | `a` |
| `a and b` | sinon | `b` |
| `a or b` | si `a` vrai, s'arrête à `a` | `a` |
| `a or b` | sinon | `b` |

### Pattern : valeur by default

L'idiome `valeur = betweene or "by default"` uses `or` pour fournir une valeur si `betweene` est vide.


````python
entry = ""
name = entry or "Anonyme"
print("Bonjour,", name)
````

````python
entry = "Alice"
name = entry or "Anonyme"
print("Bonjour,", name)
````

**Warning** : ce pattern fait passer `0` ou `0.0` pour une valeur « vide », ce qui peut être faux selon le contexte. On verra mieux in the notebook on thes functions.

## 6. `not` — l'reverses logique

`not x` est `True` si `x` est faux, `False` si `x` est vrai.


````python
print(not True)
````

````python
print(not False)
````

````python
print(not 0)
````

````python
print(not 42)
````

````python
print(not "")
````

Idiome courant : `if not x:` signifie « si x est faux / vide / zéro / None ».

## 7. `any` et `all`

Ces deux primitives acceptent une **séquence** de valeurs et returnsnt un boolean :

- `all(séquence)` : `True` si **toutes** the values sont vraies. Équivalent de several `and` enchaînés.
- `any(séquence)` : `True` si **au moins une** valeur est vraie. Équivalent de several `or` enchaînés.

**Cas de base** :

- `all([])` → `True` *(vacuité)*
- `any([])` → `False`

Pour les usesr sans connaître les lists, we can passer une **expression génératrice** basée sur `range(...)` :


````python
print(all(i > 0 for i in range(1, 10)))
````

````python
print(all(i > 5 for i in range(1, 10)))
````

````python
print(any(i > 5 for i in range(1, 10)))
````

````python
print(any(i > 100 for i in range(1, 10)))
````

L'expression `(i > 5 for i in range(1, 10))` is a **generator** : elle produit des booleans un par un. `any`/`all` s'arrêtent dès qu'ils ont leur réponse (court-circuit). On approfondira les generators in the notebook on thes comprehensions.

### Cas idiomatique : checksr une propriété sur all the integers d'un intervalle

````python
# Tous les entiers de 1 à 100 sont-ils positifs ?
print(all(i > 0 for i in range(1, 101)))
````

````python
# Existe-t-il un multiple de 7 entre 50 et 60 ?
print(any(i % 7 == 0 for i in range(50, 61)))
````

## 8. Synthèse

### Valeurs fausses in Python

- `False`
- `None`
- `0`, `0.0`
- `""`
- conteneurs vides `[]`, `()`, `{}`, `set()`

**Tout le reste est vrai.**

### Règles Key takeaways

1. **`if x:`** est l'idiome pour tester la présence d'une valeur — plus concis que `if x != 0:`.
2. **`and`/`or` ne returnsnt pas nécessairement `True`/`False`** : ils returnsnt l'un de leurs opérandes.
3. **Pattern de valeur by default** : `nom = betweene or "défaut"` (with thes limites).
4. **`all` et `any`** sur une expression génératrice — pratique pour tester une propriété sur un intervalle.
5. **`not`** functionne sur n'importe quelle valeur, pas seulement sur un boolean strict.
6. **Comparer à `True`/`False` avec `==`** is a anti-pattern : écrire `if x:` plutôt que `if x == True:`.
