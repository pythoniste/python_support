## 2. Lire une valeur : `d[k]` vs `d.get(k)`

Deux manières d'accéder à une valeur, avec des comportements différents si the key est absente.

````python
contact = {'name': 'Dupont', 'age': 30}
````

````python
contact['name']
````

````python
# Key absente : error
try:
    print(contact['city'])
except KeyError as err:
    print('key manquante :', err)
````

`.get` returns `None` (ou une valeur by default) si the key est absente, sans lever d'erreur.

````python
contact.get('city')
````

````python
print(contact.get('city'))
````

````python
contact.get('city', 'inconnue')
````

### Quelle method choisir ?

- `d[k]` : quand the key **doit** être présente — une erreur indique un bug.
- `d.get(k)` : quand the key est **optionalle** — on gère le cas absent proprement.

## 3. Ajouter, modifiesr, removesr

````python
contact = {'name': 'Dupont'}
contact['age'] = 30
print(contact)
````

La same syntaxe adds une clé absente ou met à jour une clé existante.

````python
contact['age'] = 31
print(contact)
````

### `del` removes une clé

````python
del contact['age']
print(contact)
````

### `.pop(k)` removes ET returns the value

````python
contact = {'name': 'Dupont', 'age': 30}
age = contact.pop('age')
print('age retiré :', age)
print('dict restant :', contact)
````

### `.setdefault(k, v)` — lire ou createsr

````python
contact = {'name': 'Dupont'}
contact.setdefault('age', 0)
print(contact)
````

Si the key exists, `setdefault` returns sa valeur sans la changer.

## 4. Iteration

````python
grades = {'Alice': 15, 'Bob': 12, 'Chloé': 18}
````

### By default : iteration sur the keys

````python
for name in grades:
    print(name)
````

### `.keys()`, `.values()`, `.items()`

````python
for name in grades.keys():
    print(name)
````

````python
for grade in grades.values():
    print(grade)
````

````python
for name, grade in grades.items():
    print(f'{name} : {grade}')
````

Ici on **déballe** directement le tuple `(clé, valeur)` renvoyé par `.items()`.

### Tri par valeur (aperçu)

````python
notes_triees = sorted(grades.items(), key=lambda item: item[1])
for name, grade in notes_triees:
    print(f'{name} : {grade}')
````

`lambda` is ae function anonyme — nous y reviendrons au chapitre functions. Ici `key` indique selon quel critère sortsr.

## 6. Fusion avec `|` (Python 3.9+)

L'operator `|` creates un **new** dictionary qui fusionne deux dicts. The keys de droite l'emportent en cas de conflit.

````python
base = {'a': 1, 'b': 2}
extra = {'b': 20, 'c': 3}
````

````python
fusion = base | extra
print(fusion)
````

`base` et `extra` ne sont pas modifiés.

````python
print(base)
print(extra)
````

### `|=` met à jour en place

````python
base |= extra
print(base)
````

### Avant Python 3.9 : `{**a, **b}`

Vous rencontrerez encore cette syntaxe dans du code existant. Elle est équivalente à `a | b`.

````python
{**{'a': 1, 'b': 2}, **{'b': 20, 'c': 3}}
````

## 7. Comprehensions de dict

Comme for thes lists, we can construire un dictionary en une expression.

````python
carres = {n: n * n for n in range(1, 6)}
print(carres)
````

````python
# Inverser un dictionary
grades = {'Alice': 15, 'Bob': 12, 'Chloé': 18}
par_note = {grade: name for name, grade in grades.items()}
print(par_note)
````

Warning aux doublons de valeurs : si deux clés ont la same valeur, l'inversion ne conserve que la dernière.

````python
notes_bis = {'Alice': 15, 'Bob': 12, 'David': 15}
{grade: name for name, grade in notes_bis.items()}
````

### Filtrer with a `if`

````python
grades = {'Alice': 15, 'Bob': 8, 'Chloé': 18}
reussites = {name: grade for name, grade in grades.items() if grade >= 10}
print(reussites)
````

## 8. The keys doivent être **hashables**

Une clé de dictionary doit être **immutable** (hashable) : `int`, `float`, `str`, `bool`, `tuple` de hashables, `None`…
Une `list` ne peut **pas** être utilisée comme clé.

````python
d = {(0, 0): 'origine', (1, 0): 'droite'}
print(d)
````

````python
try:
    d = {[0, 0]: 'origine'}
except TypeError as err:
    print('error :', err)
````

The values, en revanche, peuvent être de n'importe quel type.

````python
d = {'fruits': ['pomme', 'poire'], 'legumes': ['carotte']}
print(d)
````

---


## 9. Synthèse

| Opération | Syntaxe |
|---|---|
| Création | `{'a': 1, 'b': 2}` |
| Lecture stricte | `d['a']` (erreur si absent) |
| Lecture tolérante | `d.get('a')` ou `d.get('a', défaut)` |
| Ajout / mise à jour | `d['c'] = 3` |
| Suppression | `del d['a']` ou `d.pop('a')` |
| Iteration clés | `for k in d:` |
| Iteration valeurs | `for v in d.values():` |
| Iteration couples | `for k, v in d.items():` |
| Fusion | `a \| b`, `a \|= b` |
| Comprehension | `{k: v for ... in ... if ...}` |

### Règles Key takeaways

1. Un dictionary conserve l'**ordre d'insertion** since Python 3.7.
2. The keys doivent être **hashables** — immutables en pratique.
3. Préférez `d.get(k, défaut)` à `if k in d` suivi de `d[k]`.
4. `.items()` + déballage est la manière idiomatique d'itérer.

Construire une loop interactive which allows à l'utilisateur de taper :

- `list` pour tout listr ;

```python
becausenet = {}

while True:
    commande = input('> ').strip()
    if commande == 'quit':
        break
    if commande == 'list':
        for nom, tel in becausenet.items():
            print(nom, ':', tel)
        continue
    parts = commande.split(maxsplit=2)
    if parts[0] == 'add' and len(parts) == 3:
        becausenet[parts[1]] = parts[2]
    elif parts[0] == 'get' and len(parts) == 2:
        print(becausenet.get(parts[1], 'inconnu'))
    elif parts[0] == 'del' and len(parts) == 2:
        becausenet.pop(parts[1], None)
    else:
        print('commande inconnue')
```
