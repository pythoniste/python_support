### Égalité `==` et identité `is`

Ces deux operators ne font pas la same chose :

- `==` compares les **valeurs**.
- `is` compares les **identités** (« est-ce le same objet en mémoire ? »).


````python
integer == floating
````

The values sont considérées égales parce que `42 == 42.0` mathématiquement.

````python
integer is floating
````

En revanche, ce ne sont pas les sames objets : l'un is a `int`, l'other un `float`. Retenez : **on compares des valeurs avec `==`, pas avec `is`**.

### Comparaison avec `None`

`None` is ae valeur spéciale which represents « rien ». On la compares **toujours** avec `is`, jamais avec `==`.


````python
integer is None
````

````python
integer == None
````

Les deux marchent ici, mais `is None` est le style recommandé (plus rapide et plus sûr).

### Décalages de bits

Les operators `<<` et `>>` décalent les bits d'un integer. Each décalage d'un cran à gauche multiplie par 2, each décalage à droite divise par 2 (en integer).


````python
4 >> 2  # 4 = 0b100, décalé de 2 à droite donne 0b1 = 1
````

````python
4 << 2  # 0b100 décalé de 2 à gauche donne 0b10000 = 16
````

````python
15 >> 2  # 0b1111 >> 2 = 0b11 = 3
````

````python
15 << 2  # 0b1111 << 2 = 0b111100 = 60
````

## 6. Mélange `int` / `float` et promotion de type

What happens quand on mélange un `int` et un `float` dans une opération ? Python « promeut » automatiquement l'integer en float.


````python
integer + floating
````

````python
type(integer + floating)
````

````python
integer / floating
````

````python
print(99 / 100)
````

---

### Ce qu'il est important de comprendre

Dans la plupart des langages à bas niveau (C, C++, Java, Go...), l'opération est programmée **au niveau du type** :

- `int + int` donne un `int` ;
- `float + float` donne un `float` ;
- `int / int` donne un `int` avec **troncature** (en C, `5 / 2` vaut `2`, pas `2.5`).

Python, lui, fait une analyse de types au moment de l'exécution et choisit automatiquement la bonne opération. C'is a confort énorme, mais il faut garder à l'esprit que :

- `/` returns **toujours** un `float`, same between deux `int` ;
- `//` returns un `int` si les deux opérandes are `int`, un `float` sinon ;
- Le mélange `int + float` produit toujours un `float`.

**En pratique** : pour faire de l'arithmétique exacte sur des integers, restez between integers et usesz `//` plutôt que `/`.

## 7. Un piège à connaître : IEEE 754

Les `float` de Python sont implémentés selon la norme internationale **IEEE 754 double précision** (64 bits). Ils peuvent representsr approximativement des nombres de `1,8 × 10³⁰⁸` à `2,2 × 10⁻³⁰⁸`. Mais ils ne les representsnt **pas toujours exactement**.

Here is le piège le plus célèbre en informatique :


````python
0.1 + 0.2
````

````python
0.1 + 0.2 == 0.3
````

Surpris ? Ce n'est pas un bug de Python — c'is ae limite fondamentale du format binaire. `0.1`, `0.2` et `0.3` ne peuvent **pas** être représentés exactement en binaire, of the same façon que `1/3` ne peut pas être représenté exactement en décimal (`0,333...`).

L'erreur d'arrondi est minuscule (`~10⁻¹⁷`) mais elle exists. Elle s'accumule quand on fait beaucoup d'opérations, ce qui peut poser problème en finance ou en calcul scientifique précis.

> **Règle Key takeaways** : on ne compares **jamais** deux floats avec `==`. On uses `math.isclose(a, b)`.


````python
import math
````

````python
math.isclose(0.1 + 0.2, 0.3)
````

Pour voir le piège dans all the langages de programmation, une page de référence : [0.30000000000000004.com](https://0.30000000000000004.com/).

### Arrondir : Warning aux pièges

Several functions arrondissent, et elles ne font pas la same chose.


**`round(x)`** uses l'« arrondi bancaire » : pour `.5`, on arrondit vers le **pair** le plus proche.

````python
round(2.5)  # 2 !! (pas 3)
````

````python
round(3.5)  # 4
````

````python
round(0.5)  # 0 !!
````

C'est déroutant, mais Python suit la norme IEEE 754 (*round half to even*) qui est statistiquement plus juste que l'arrondi scolaire. En pratique, pour des grandes séries de valeurs, cette règle évite les biais.

**`math.floor(x)`** : arrondi vers `-∞` (vers le bas).

````python
math.floor(2.7)
````

````python
math.floor(-2.3)
````

**`math.ceil(x)`** : arrondi vers `+∞` (vers le haut).

````python
math.ceil(2.3)
````

````python
math.ceil(-2.7)
````

**`math.trunc(x)`** : arrondi vers `0`.

````python
math.trunc(2.7)
````

````python
math.trunc(-2.7)
````

### Exercise 8 — Équation du second degré *(difficile)*

Écrire un script qui résout une équation of the forme `a × x² + b × x + c = 0`. Il doit :

1. demander les trois coefficients `a`, `b`, `c` à l'utilisateur ;
2. computesr le discriminant `Δ = b² − 4ac` ;
3. gérer all the cas followings :

| Cas | Action |
|---|---|
| `a == 0` et `b == 0` | displaysr « ce n'est pas une équation » |
| `a == 0` (mais `b != 0`) | équation linéaire, displaysr la Solution `-c / b` |
| `Δ ≈ 0` (avec `math.isclose`) | displaysr la racine double |
| `Δ < 0` | displaysr « pas de racine réelle » |
| `Δ > 0` | displaysr les deux racines |

**Indice :** usesz `math.isclose(delta, 0, abs_tol=1e-12)` pour détecter le discriminant nul proprement.


````python
# Votre code ici
````

<details>
<summary>📖 Voir la Solution</summary>

```python
import math

a = float(input("a : "))
b = float(input("b : "))
c = float(input("c : "))

if a == 0 and b == 0:
    print("Ce n'est pas une équation.")
elif a == 0:
    print("Équation linéaire, Solution :", -c / b)
else:
    delta = b * b - 4 * a * c
    if math.isclose(delta, 0, abs_tol=1e-12):
        print("Racine double :", -b / (2 * a))
    elif delta < 0:
        print("Pas de racine réelle.")
    else:
        racine = math.sqrt(delta)
        x1 = (-b - racine) / (2 * a)
        x2 = (-b + racine) / (2 * a)
        print("Deux racines :", x1, "et", x2)
```

</details>
