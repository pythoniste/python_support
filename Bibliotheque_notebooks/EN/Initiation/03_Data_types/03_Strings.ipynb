## 1. Créer une chaîne

Trois délimiteurs équivalents :

| Forme | Quand l'usesr |
|---|---|
| `"texte"` | By default |
| `'texte'` | Si le contenu contains des `"` |
| `"""texte"""` | Pour du texte multi-lignes |

`"""..."""` est also le format des **docstrings** (documentation de function, vu plus tard).


````python
s = "bonjour"
print(s)
````

````python
s = 'hello'
print(s)
````

````python
s = "Elle a dit \"bonjour\"."
print(s)
````

````python
s = 'Elle a dit "bonjour".'
print(s)
````

We can also usesr `\n` dans une chaîne simple :

## 3. Concaténation et répétition

- `+` concatène deux chaînes.
- `*` répète une chaîne.


````python
print("abc" + "def")
````

````python
print("ha" * 5)
````

````python
print("-" * 20)
````

**Warning** : `+` ne marche qu'between deux chaînes. Pour insérer un nombre, il faut le convertir ou — mieux — usesr une **f-string** (section 10).

## 4. Indexation et longueur

Une chaîne est **indexable** : each becauseactère a une position. Les index commencent à **0**.


````python
s = "Python"
print(s[0])
````

````python
s = "Python"
print(s[1])
````

````python
s = "Python"
print(s[5])
````

Les **index négatifs** comptent since la fin. `-1` est le last becauseactère.

````python
s = "Python"
print(s[-1])
````

````python
s = "Python"
print(s[-2])
````

La longueur s'obtient avec `len(...)` :

````python
s = "Python"
print(len(s))
````

Accéder à un index inexistant lève `IndexError` :

````python
s = "abc"
# s[10]  # IndexError — décommenter pour tester
````

### Immutabilité

Une chaîne est **immutable** : on ne peut pas modifiesr un becauseactère en place.


````python
s = "python"
# s[0] = "P"  # TypeError — décommenter pour tester
````

Pour « modifiesr » une chaîne, on en creates une nouvelle :

````python
s = "python"
s = "P" + s[1:]
print(s)
````

## 5. Slicing

The syntax `s[debut:fin]` returns une **sous-chaîne**. `debut` est inclus, `fin` est **exclu**.


````python
s = "abcdefgh"
print(s[0:3])
````

````python
s = "abcdefgh"
print(s[2:6])
````

````python
s = "abcdefgh"
print(s[:4])
````

````python
s = "abcdefgh"
print(s[4:])
````

````python
s = "abcdefgh"
print(s[-3:])
````

````python
s = "abcdefgh"
print(s[::2])
````

````python
s = "abcdefgh"
print(s[::-1])
````

La dernière forme `[::-1]` est l'idiome pour **reversesr** une chaîne : le `pas -1` parcourt of the fin vers le début.

## 6. Itérer sur une chaîne

## 7. Methods de transformation

Les methods sont appelées with the notation `s.methode(args)`. Elles **returnsnt une nouvelle chaîne** — jamais ne modifiesnt l'originale *(puisque les chaînes sont immutables)*.


````python
s = "Bonjour"
print(s.upper())
````

````python
s = "Bonjour"
print(s.lower())
````

````python
s = "alice"
print(s.capitalize())
````

````python
s = "alice dupont"
print(s.title())
````

**Nettoyer les espaces autour** :

````python
s = "   text   "
print(repr(s.strip()))
````

*`repr(...)` displays les guillemets autour of the chaîne, pratique pour voir les bornes.*

````python
s = "---abc---"
print(s.strip("-"))
````

**Remplacer une sous-chaîne** :

````python
s = "bonjour le monde"
print(s.replace("monde", "Python"))
````

## 8. Methods de recherche

Elles returnsnt soit un boolean, soit un index.


````python
s = "bonjour"
print(s.startswith("bon"))
````

````python
s = "photo.jpg"
print(s.endswith(".jpg"))
````

````python
s = "abracadabra"
print(s.find("cad"))
````

````python
s = "abracadabra"
print(s.find("xyz"))
````

`find` returns l'index of the première occurrence, ou `-1` si absent.

````python
s = "abracadabra"
print(s.count("a"))
````

````python
s = "42"
print(s.isdigit())
````

````python
s = "abc"
print(s.isalpha())
````

### L'operator `in`

Pour tester la présence d'une sous-chaîne, we can usesr directement `in` :


````python
print("cad" in "abracadabra")
````

````python
print("xyz" in "abracadabra")
````

## 9. Methods `split` et `join`

`split(separateur)` **découpe** une chaîne en several morceaux. The result is une **list** (qu'on verra en détail bientôt — ici, on se contente de l'itérer).


````python
s = "pomme,poire,banane"
print(s.split(","))
````

````python
s = "un deux trois"
print(s.split())
````

Sans argument, `split()` découpe sur n'importe quelle suite d'espaces.

`join(morceaux)` est l'**reverses** : colle several morceaux en une chaîne.

````python
morceaux = ["a", "b", "c"]
print("-".join(morceaux))
````

## 10. **f-strings** : le format moderne

Since Python 3.6, les **f-strings** (format strings) sont la façon recommandée d'insérer des valeurs dans une chaîne. The syntax est `f"...{expression}..."` — le préfixe `f` transforme les parties between accolades en **expressions Python évaluées**.


````python
name = "Alice"
age = 30
print(f"{name} a {age} ans.")
````

````python
x = 2
y = 3
print(f"{x} + {y} = {x + y}")
````

L'expression between `{...}` peut être **n'importe quelle expression Python** : appel de function, calcul, attribute, etc.

````python
import math
r = 5
print(f"Aire du cercle : {math.pi * r ** 2}")
````

### Debug : `{x=}`

Since Python 3.8, the syntax `f"{x=}"` displays to the fois le **nom** et la **valeur** de `x`. Très pratique pour déboguer.


````python
x = 42
print(f"{x=}")
````

````python
first_name = "Alice"
age = 30
print(f"{first_name=} {age=}")
````

## 11. Options de formatage (largeur, alignement, décimales)

Between les accolades, we can addsr un **format spec** après `:` pour contrôler l'affichage.


**Décimales** :

````python
pi = 3.141592653589793
print(f"{pi:.2f}")
print(f"{pi:.5f}")
````

**Largeur et alignement** :

````python
word = "abc"
print(f"[{word:>10}]")
print(f"[{word:<10}]")
print(f"[{word:^10}]")
````

**Largeur et remplissage** :

````python
n = 42
print(f"{n:05}")
````

**Séparateur de milliers** :

````python
n = 1000000
print(f'{n:,}')
````

**Pourcentage** :

````python
taux = 0.153
print(f'{taux:.1%}')
````

**Tableau d'alignement utile** :

````python
for i in range(1, 6):
    square = i ** 2
    cube = i ** 3
    print(f"{i:>3} | {square:>5} | {cube:>5}")
````

## 12. Raw strings

Une **raw string** est préfixée par `r` : les antislashes ne sont **pas** interprétés. Très pratique for thes chemins Windows, les expressions régulières, etc.


````python
print(r"C:\Users\alice")
````

````python
print(r"line 1\nligne 2")
````

Sans le `r`, `\n` serait interprété comme un retour to the ligne.

## 13. Synthèse

### Methods essentielles

| Method | Rôle |
|---|---|
| `upper()` / `lower()` | Changer la casse |
| `strip()` | Nettoyer les espaces de début/fin |
| `replace(a, b)` | Remplacer `a` par `b` |
| `split(sep)` | Découper en list |
| `"sep".join(parts)` | Coller des morceaux |
| `startswith(p)` / `endswith(p)` | Tester début / fin |
| `find(sub)` | Position d'une sous-chaîne ou -1 |
| `count(sub)` | Nombre d'occurrences |
| `in` | Tester la présence |

### Règles Key takeaways

1. **Les chaînes sont immutables** : all the methods returnsnt une **nouvelle** chaîne.
2. **Slicing `[a:b]` : `a` inclus, `b` exclu.**
3. **`[::-1]`** reverses une chaîne.
4. **`len(s)`** donne la longueur.
5. **f-strings** `f"..."` pour insérer des valeurs : `f"{x + y}"`.
6. **`f"{x=}"`** pour displaysr nom + valeur — débug rapide.
7. **Format `:.2f`** pour 2 décimales, `:>10` pour alignement à droite sur 10 becauseactères.
8. **`r"..."`** for thes chemins Windows et les regex.
