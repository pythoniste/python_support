## 2. Accéder aux éléments

### Index positif

Les éléments sont numérotés à partir de **0**.

````python
fruits[0]
````

````python
fruits[1]
````

````python
fruits[2]
````

### Index négatif

Un index négatif compte since la fin : `-1` est le last élément.

````python
fruits[-1]
````

````python
fruits[-2]
````

### Slicing (tranches)

The syntax `list[debut:fin]` extrait une sous-list. `debut` est inclus, `fin` est **exclu**.

````python
numbers = [10, 20, 30, 40, 50, 60, 70]
````

````python
numbers[1:4]
````

````python
numbers[:3]   # du début jusqu'à l'index 3 (exclu)
````

````python
numbers[4:]   # de l'index 4 jusqu'à la fin
````

````python
numbers[::2]  # un élément sur deux
````

````python
numbers[::-1] # list inversée
````

### Slicing with a pas négatif

````python
numbers[5:1:-1]  # de l'index 5 à l'index 2 (1 exclu), à reculons
````

````python
numbers[6:0:-2]  # de la fin vers le début, un sur deux
````

## 4. Methods de list — ajout et suppression

### `del` — removesr par index ou par tranche

## 6. Trier une list

### `sort()` — tri en place (modifies la list)

`sort()` ne returns rien (`None`), il modifies la list directement.

````python
grades.sort()
````

````python
print(grades)
````

Tri décroissant avec `reverse=True`.

````python
grades.sort(reverse=True)
````

````python
print(grades)
````

### `sorted()` — tri qui creates une nouvelle list

`sorted()` returns une **nouvelle** list triée, l'originale ne change pas.

````python
original = [3, 1, 4, 1, 5, 9]
trie = sorted(original)
````

````python
print(trie)
````

````python
print(original)  # inchangé
````

### `reverse()` — reversesr en place

````python
original.reverse()
````

````python
print(original)
````

### Tri avec clé personnalisée

We can passer une function `key` à `sort()` / `sorted()`. Par Example, sortsr des mots par longueur :

````python
words = ["banane", "kiwi", "fraise", "pomme"]
print(sorted(words, key=len))
````

## 8. Comprehensions de list

Une **comprehension de list** (list comprehension) creates une list en une seule expression.

Syntaxe : `[expression for element in iterable]`

````python
carres = [x ** 2 for x in range(6)]
````

````python
print(carres)
````

Avec une **condition de filtrage** : `[expression for element in iterable if condition]`

````python
pairs = [x for x in range(10) if x % 2 == 0]
````

````python
print(pairs)
````

We can appliquer une transformation et un filtre en same temps.

````python
longueurs = [len(m) for m in ["chat", "chien", "oiseau", "rat"]]
print(longueurs)
````

### Comprehension avec f-string

````python
etiquettes = [f"item_{i}" for i in range(1, 6)]
print(etiquettes)
````

### Équivalence with ae loop classique

Each comprehensiwe can se réécrire en loop (et vice-versa).

````python
# Loop classique
result = []
for x in range(6):
    result.append(x ** 2)
print(result)
````

## 9. Pièges courants

### Piège 1 : copies par référence

Affecter une list à une other variable ne creates **pas** de copies.

````python
a = [1, 2, 3]
b = a          # b pointe vers la MÊME my_list
b.append(4)
````

````python
print(a)  # a est aussi modifié !
````

Pour faire une **vraie copies**, usesr `.copy()` ou le slicing `[:]`.

````python
a = [1, 2, 3]
b = a.copy()   # copy indépendante
b.append(4)
````

````python
print(a)  # a n'a pas changé
````

````python
print(b)
````

### Piège 2 : multiplication de lists imbriquées

````python
# Piège : les sous-listes sont les MÊMES objets
grille = [[0] * 3] * 3
grille[0][0] = 1
print(grille)  # toutes les lines changent !
````

````python
# Solution : compréhension de my_list
grille = [[0] * 3 for _ in range(3)]
grille[0][0] = 1
print(grille)  # seule la première line change
````

### Piège 3 : modifiesr une list en la parcourant

````python
# Ne faites pas ça :
# for x in my_list:
#     if condition:
#         my_list.remove(x)  # comportement imprévisible

# Utilisez une compréhension :
numbers = [1, 2, 3, 4, 5, 6]
numbers = [x for x in numbers if x % 2 != 0]
print(numbers)
````

## 10. Synthèse

| Opération | Syntaxe | Modifie la list ? |
|---|---|---|
| Créer | `[1, 2, 3]` ou `list(...)` | — |
| Accéder | `list[i]`, `list[a:b]` | Non |
| Modifier | `list[i] = val` | **Oui** |
| Ajouter | `append()`, `insert()`, `extend()` | **Oui** |
| Supprimer | `pop()`, `remove()`, `del` | **Oui** |
| Chercher | `in`, `index()`, `count()` | Non |
| Trier | `sort()` | **Oui** |
| Trier (copies) | `sorted()` | Non (nouvelle list) |
| Comprehension | `[expr for x in iter]` | Non (nouvelle list) |

### Règles Key takeaways

1. Lists sont **mutables** : les methods comme `append`, `sort`, `remove` modifiesnt la list en place.
2. `sorted()` creates une **nouvelle** list ; `sort()` modifies l'existante.
3. Warning aux **copiess par référence** : `b = a` ne copies pas, `b = a.copy()` copies.
4. Les comprehensions de list sont plus **lisibles** et **rapides** qu'une loop + `append`.

courses = []

while True:
    choix = input("(a)jouter, (s)upprimer, (l)ister, (q)uitter : ")
    if choix == "q":
        print("Au revoir !")
        break
    elif choix == "a":
        article = input("Article à addsr : ")
        courses.append(article)
        print(f"'{article}' ajouté.")
    elif choix == "s":
        article = input("Article à removesr : ")

            courses.remove(article)
            print(f"'{article}' supprimé.")

            print(f"'{article}' n'est pas in the list.")
    elif choix == "l":
        if courses:
            for i, art in enumerate(courses, 1):
                print(f"  {i}. {art}")
        else:
