### Tuple vide et tuple à un élément

````python
empty = ()
````

````python
print(empty, type(empty))
````

Pour un tuple d'un seul élément, la virgule est **required**, sinon Python voit une expression parenthésée.

## 3. Immutabilité

Contrairement à une `list`, un `tuple` **ne peut pas être modifié** après création.

````python
point = (3, 4)
try:
    point[0] = 99
except TypeError as err:
    print('Erreur :', err)
````

On ne peut ni `.append`, ni `.pop`, ni réaffecter un indice.

### Pourquoi l'immutabilité est utile

- un tuple est **hashable** : we can l'usesr comme clé de dictionary ou élément d'un `set` (chapitres followings) ;
- il signale au lecteur : *ces valeurs ne doivent pas changer* (ex. des coorderedes, une version `(3, 14, 0)`) ;
- il est légèrement plus petit et plus rapide à construire qu'une list.

````python
import sys
sys.getsizeof((1, 2, 3)), sys.getsizeof([1, 2, 3])
````

## 4. Déballage (unpacking)

Le **déballage** allows d'assigner les éléments d'un tuple à several variables en une seule ligne.

````python
point = (3, 4)
x, y = point
print('x =', x, ' y =', y)
````

Le déballage marche also to the volée, sans passer par une variable intermédiaire.

````python
x, y = 3, 4
print(x, y)
````

### Swap élégant

````python
a = 1
b = 2
a, b = b, a
print('a =', a, ' b =', b)
````

### Déballage avec `*` (rest)

````python
first, *remainder = (1, 2, 3, 4, 5)
print(first)
print(remainder)
````

````python
debut, *milieu, fin = (1, 2, 3, 4, 5)
print(debut, milieu, fin)
````

`*milieu` capture une **list** de ce qui reste between `debut` et `fin`.

## 5. Tuples et retour multiple

Same si `def` n'est vu qu'au chapitre following, be aware that de nombreuses primitives Python returnsnt déjà des tuples. C'est le moyen idiomatique de renvoyer **several valeurs**.

````python
quotient, remainder = divmod(17, 5)
print('q =', quotient, ' r =', remainder)
````

`divmod` returns un tuple `(quotient, reste)` que l'on déballe directement.

````python
string = 'a=1,b=2,c=3'
parts = string.split(',')
print(parts)
````

Ici `split` returns une list — mais we can déballer une list exactement comme un tuple.

````python
a, b, c = parts
print(a, b, c)
````

## 8. `namedtuple` — tuples avec noms de champs

Un `namedtuple` is a tuple dont each élément a un **nom**. Très utile pour representsr des structures simples sans createsr de class.

````python
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 4)
print(p)
````

````python
p.x, p.y
````

````python
# Déballage classique toujours possible
x, y = p
print(x, y)
````

Dans les chapitres followings, nous verrons `dataclass` et `TypedDict` qui vont plus loin. `namedtuple` reste utile pour un besoin léger.

## 9. Quand choisir `tuple` plutôt que `list`

| Situation | Choix recommandé |
|---|---|
| Collection homogène susceptible d'évoluer | `list` |
| Enregistrement figé (coorderede, version, couple) | `tuple` |
| Clé de dictionary ou élément de `set` | `tuple` (immutable, hashable) |
| Valeurs de retour multiples | `tuple` (avec déballage) |

**Règle empirique :** si vous voulez écrire `.append`, usesz une list. Sinon, préférez un tuple : le lecteur comprend immédiatement que ces valeurs forment un **bloc figé**.

## 10. Synthèse

| Opération | List | Tuple |
|---|---|---|
| Création | `[1, 2, 3]` | `(1, 2, 3)` ou `1, 2, 3` |
| Indexation | `l[0]` | `t[0]` |
| Longueur | `len(l)` | `len(t)` |
| Mutable ? | ✅ | ❌ |
| Hashable ? | ❌ | ✅ (si éléments hashables) |
| Method `.append` | ✅ | ❌ |
| Déballage `a, b = x` | ✅ | ✅ |

### Règles Key takeaways

1. Ce sont les **virgules** qui createsnt un tuple, pas les parenthèses.
2. Un tuple d'un élément s'écrit `(42,)` — la virgule est required.
3. Un tuple est **immutable** : nonee method ne le modifies.
4. Le **déballage** `x, y = point` est l'idiome Python par excellence.
5. `divmod`, `enumerate`, `zip` returnsnt des tuples — retenez le déballage.
